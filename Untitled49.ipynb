```python

# ============================================================
# CFIS EFFICIENCY OPTIMIZATION — CÉLULA 1
# RECUPERAÇÃO E AUDITORIA DO PROJETO
# ============================================================

from pathlib import Path
import json
import hashlib
import platform
import sys
import importlib.metadata as metadata

import numpy as np
import pandas as pd

from google.colab import drive

drive.mount("/content/drive", force_remount=False)

PROJECT_DIR = Path(
    "/content/drive/MyDrive/CFIS_Artigo_Diogo"
)

DIRS = {
    "processed": PROJECT_DIR / "01_Datasets" / "Processed",
    "rankings": PROJECT_DIR / "02_Rankings",
    "performance": PROJECT_DIR / "03_Performance",
    "efficiency": PROJECT_DIR / "04_Efficiency",
    "logs": PROJECT_DIR / "09_Logs",
    "cache": PROJECT_DIR / "12_Cache",
}

STUDY_DIR = (
    DIRS["efficiency"] / "CFIS_Optimization"
)

for folder in [
    STUDY_DIR,
    STUDY_DIR / "01_Audit",
    STUDY_DIR / "02_Profiling",
    STUDY_DIR / "03_Optimized",
    STUDY_DIR / "04_Benchmarks",
    STUDY_DIR / "05_Validation",
]:
    folder.mkdir(parents=True, exist_ok=True)

SOURCE_PATH = (
    DIRS["logs"] / "CFIS_replication_v1.py"
)

EXPECTED_SHA256 = (
    "7925f673b88145d3aa44076ee05dbae7ac3f3370b8b3218392d6ffd0ce3ce5cf"
)

if not SOURCE_PATH.exists():
    raise FileNotFoundError(
        f"Implementação original não encontrada: {SOURCE_PATH}"
    )

source_bytes = SOURCE_PATH.read_bytes()

source_hash = hashlib.sha256(
    source_bytes
).hexdigest()

if source_hash != EXPECTED_SHA256:
    raise RuntimeError(
        "O código original não corresponde à versão "
        "auditada anteriormente. Não prosseguir "
        "sem investigar a divergência."
    )

prepared_datasets = {}
dataset_rows = []

for dataset_dir in sorted(
    DIRS["processed"].iterdir()
):
    if not dataset_dir.is_dir():
        continue

    required = [
        "X_train.parquet",
        "X_test.parquet",
        "y_train.parquet",
        "y_test.parquet",
        "metadata.json",
    ]

    if not all(
        (dataset_dir / filename).exists()
        for filename in required
    ):
        continue

    name = dataset_dir.name

    X_train = pd.read_parquet(
        dataset_dir / "X_train.parquet"
    )

    X_test = pd.read_parquet(
        dataset_dir / "X_test.parquet"
    )

    y_train_df = pd.read_parquet(
        dataset_dir / "y_train.parquet"
    )

    y_test_df = pd.read_parquet(
        dataset_dir / "y_test.parquet"
    )

    with open(
        dataset_dir / "metadata.json",
        encoding="utf-8"
    ) as f:
        dataset_metadata = json.load(f)

    if isinstance(y_train_df, pd.DataFrame):
        if y_train_df.shape[1] != 1:
            raise ValueError(
                f"Alvo de treino inválido: {name}"
            )
        y_train = y_train_df.iloc[:, 0]
    else:
        y_train = y_train_df

    if isinstance(y_test_df, pd.DataFrame):
        if y_test_df.shape[1] != 1:
            raise ValueError(
                f"Alvo de teste inválido: {name}"
            )
        y_test = y_test_df.iloc[:, 0]
    else:
        y_test = y_test_df

    if list(X_train.columns) != list(X_test.columns):
        raise ValueError(
            f"Características incompatíveis: {name}"
        )

    if (
        len(X_train) != len(y_train)
        or len(X_test) != len(y_test)
    ):
        raise ValueError(
            f"Dimensões incompatíveis: {name}"
        )

    task = dataset_metadata.get("task")

    if task not in ("classification", "regression"):
        raise ValueError(
            f"Tarefa não identificada: {name} ({task})"
        )

    prepared_datasets[name] = {
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test,
        "task": task,
        "feature_names": list(X_train.columns),
    }

    dataset_rows.append({
        "Dataset": name,
        "Task": task,
        "Train": len(X_train),
        "Test": len(X_test),
        "Features": X_train.shape[1],
        "Train_MiB": round(
            X_train.memory_usage(
                index=True, deep=True
            ).sum() / 2**20,
            3
        ),
    })

if len(prepared_datasets) != 12:
    raise RuntimeError(
        f"Esperados 12 datasets; "
        f"encontrados {len(prepared_datasets)}."
    )

inventory = pd.DataFrame(dataset_rows)

audit = {
    "study": "CFIS Efficiency Optimization",
    "source_path": str(SOURCE_PATH),
    "source_sha256": source_hash,
    "source_verified": True,
    "dataset_count": len(prepared_datasets),
    "datasets": dataset_rows,
    "environment": {
        "python": sys.version,
        "platform": platform.platform(),
        "numpy": np.__version__,
        "pandas": pd.__version__,
        "scikit_learn": metadata.version(
            "scikit-learn"
        ),
    },
    "constraints": {
        "preserve_original_cfis_formula": True,
        "preserve_original_scientific_parameters": True,
        "no_predictive_hyperparameter_tuning": True,
        "no_test_data_during_selection": True,
        "no_overwrite_of_prior_experiments": True,
    },
}

audit_path = (
    STUDY_DIR / "01_Audit" / "project_audit.json"
)

with open(
    audit_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        audit,
        f,
        indent=2,
        ensure_ascii=False
    )

print("=" * 72)
print("CFIS — NOVO ESTUDO DE OTIMIZAÇÃO")
print("=" * 72)

print("\nImplementação original: VERIFICADA")
print("SHA256:", source_hash)
print("Datasets recuperados:", len(prepared_datasets))

display(inventory)

print("\nAmbiente:")
print("Python:", platform.python_version())
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Scikit-learn:", metadata.version("scikit-learn"))

print("\nDiretório do estudo:", STUDY_DIR)
print("Auditoria salva:", audit_path)

print("\nCÉLULA 1 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS EFFICIENCY OPTIMIZATION — CÉLULA 2
# PROFILING DA IMPLEMENTAÇÃO ORIGINAL
# PILOTO: BREAST CANCER
# ============================================================

import ast
import cProfile
import gc
import importlib.util
import inspect
import io
import json
import os
import pstats
import time
import tracemalloc

import numpy as np
import pandas as pd

from pathlib import Path

STUDY_DIR = (
    Path(DIRS["efficiency"]) / "CFIS_Optimization"
)

PROFILE_DIR = STUDY_DIR / "02_Profiling"
PROFILE_DIR.mkdir(parents=True, exist_ok=True)

SOURCE_PATH = (
    Path(DIRS["logs"]) / "CFIS_replication_v1.py"
)

# ------------------------------------------------------------
# 1. Inspeção da implementação original
# ------------------------------------------------------------

source_text = SOURCE_PATH.read_text(
    encoding="utf-8"
)

tree = ast.parse(source_text)

print("=" * 72)
print("CFIS — INSPEÇÃO DO CÓDIGO ORIGINAL")
print("=" * 72)

functions = [
    node.name
    for node in tree.body
    if isinstance(
        node, (ast.FunctionDef, ast.AsyncFunctionDef)
    )
]

print("\nFunções encontradas:")
for name in functions:
    print(" -", name)

# Importação sem executar o bloco __main__, se houver.
spec = importlib.util.spec_from_file_location(
    "cfis_reference_module",
    SOURCE_PATH
)

module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

if not hasattr(module, "CFIS_feature_selection"):
    raise AttributeError(
        "Função CFIS_feature_selection não encontrada."
    )

cfis_function = module.CFIS_feature_selection

print("\nAssinatura:")
print(inspect.signature(cfis_function))

print("\nCódigo-fonte da função:")
print(inspect.getsource(cfis_function))

print("\n" + "=" * 72)
print("INSPEÇÃO CONCLUÍDA")
print("=" * 72)

print(
    "\nIMPORTANTE: esta célula apenas inspecionou "
    "a implementação. O profiling será executado "
    "após confirmarmos a assinatura da função."
)

print("\nCÉLULA 2 CONCLUÍDA.")
```

```python

import numpy as np
import pandas as pd
import warnings

from scipy import stats
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression, Lasso
from sklearn.inspection import permutation_importance

# Injeta as dependências no módulo original já importado.
dependencies = {
    "np": np,
    "pd": pd,
    "warnings": warnings,
    "stats": stats,
    "clone": clone,
    "StandardScaler": StandardScaler,
    "LogisticRegression": LogisticRegression,
    "Lasso": Lasso,
    "permutation_importance": permutation_importance,
}

cfis_function.__globals__.update(dependencies)

# Verifica se todas as dependências estão disponíveis.
missing = [
    name for name in dependencies
    if name not in cfis_function.__globals__
]

assert not missing, f"Dependências ausentes: {missing}"

print("Dependências do CFIS carregadas corretamente.")
print("Código original preservado.")
print("Pronto para executar a Célula 3.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 4
# PROFILING POR COMPONENTE — BREAST CANCER
# IMPLEMENTAÇÃO ORIGINAL PRESERVADA
# ============================================================

import time
import json
import functools
import numpy as np
import pandas as pd

from pathlib import Path

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10

data = prepared_datasets[DATASET_NAME]

PROFILE_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "02_Profiling"
)

PROFILE_DIR.mkdir(parents=True, exist_ok=True)

timings = {}

# ------------------------------------------------------------
# 1. INSTRUMENTAÇÃO TEMPORÁRIA
# ------------------------------------------------------------

globals_cfis = cfis_function.__globals__

def timed_wrapper(name, original_function):

    @functools.wraps(original_function)
    def wrapper(*args, **kwargs):

        start = time.perf_counter()

        try:
            return original_function(*args, **kwargs)
        finally:
            elapsed = time.perf_counter() - start

            timings.setdefault(name, []).append(elapsed)

    return wrapper


targets = {
    "StandardScaler": "scaler",
    "permutation_importance": "permutation",
    "LogisticRegression": "model_class",
    "Lasso": "model_regression",
}

originals = {}

# Instrumentar scaler.fit_transform por meio de subclasse.
OriginalScaler = globals_cfis["StandardScaler"]

class TimedScaler(OriginalScaler):

    def fit_transform(self, X, y=None, **fit_params):

        start = time.perf_counter()

        try:
            return super().fit_transform(
                X, y, **fit_params
            )
        finally:
            timings.setdefault(
                "StandardScaler.fit_transform", []
            ).append(
                time.perf_counter() - start
            )


# Instrumentar ajuste do modelo.
OriginalLogistic = globals_cfis["LogisticRegression"]
OriginalLasso = globals_cfis["Lasso"]

class TimedLogisticRegression(OriginalLogistic):

    def fit(self, X, y, sample_weight=None):

        start = time.perf_counter()

        try:
            return super().fit(
                X, y, sample_weight=sample_weight
            )
        finally:
            timings.setdefault(
                "LogisticRegression.fit", []
            ).append(
                time.perf_counter() - start
            )


class TimedLasso(OriginalLasso):

    def fit(self, X, y, sample_weight=None, check_input=True):

        start = time.perf_counter()

        try:
            return super().fit(
                X, y,
                sample_weight=sample_weight,
                check_input=check_input
            )
        finally:
            timings.setdefault(
                "Lasso.fit", []
            ).append(
                time.perf_counter() - start
            )


originals["StandardScaler"] = OriginalScaler
originals["LogisticRegression"] = OriginalLogistic
originals["Lasso"] = OriginalLasso
originals["permutation_importance"] = (
    globals_cfis["permutation_importance"]
)

globals_cfis["StandardScaler"] = TimedScaler
globals_cfis["LogisticRegression"] = TimedLogisticRegression
globals_cfis["Lasso"] = TimedLasso

globals_cfis["permutation_importance"] = timed_wrapper(
    "permutation_importance",
    originals["permutation_importance"]
)

# ------------------------------------------------------------
# 2. EXECUÇÃO
# ------------------------------------------------------------

try:

    start_total = time.perf_counter()

    result = cfis_function(
        data["X_train"],
        data["y_train"],
        task=data["task"],
        model=None,
        n_repeats=N_REPEATS,
        random_state=SEED
    )

    total_time = time.perf_counter() - start_total

finally:

    # Restaurar sempre as dependências originais.
    for name, original in originals.items():
        globals_cfis[name] = original

# ------------------------------------------------------------
# 3. RESULTADOS
# ------------------------------------------------------------

rows = []

for component, measurements in timings.items():

    rows.append({
        "Component": component,
        "Time_s": sum(measurements),
        "Calls": len(measurements),
        "Percentage_total": (
            100 * sum(measurements) / total_time
        )
    })

measured_time = sum(
    row["Time_s"] for row in rows
)

rows.append({
    "Component": "Other_operations",
    "Time_s": max(0.0, total_time - measured_time),
    "Calls": 1,
    "Percentage_total": (
        100 * max(0.0, total_time - measured_time)
        / total_time
    )
})

profile = pd.DataFrame(rows).sort_values(
    "Time_s",
    ascending=False
)

profile_path = (
    PROFILE_DIR
    / f"{DATASET_NAME}_component_profile.csv"
)

profile.to_csv(
    profile_path,
    index=False
)

print("=" * 72)
print("CFIS — PROFILING POR COMPONENTE")
print("=" * 72)

print(f"\nTempo total: {total_time:.4f} s")

display(profile.round(4))

print("\nTOP 10 — CFIS")
display(result.head(10))

print("\nPerfil salvo:", profile_path)
print("CÉLULA 4 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 4 (PROFILING DETALHADO)
# P: Permutation Importance
# C: Extração de coeficientes
# R: ANOVA / Pearson
# ============================================================

import time
import json
import functools
import inspect
import textwrap
import ast
import numpy as np
import pandas as pd

from pathlib import Path

DATASET_NAME = "Breast_Cancer"
N_REPEATS = 10
SEED = 42

PROFILE_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "02_Profiling"
)
PROFILE_DIR.mkdir(parents=True, exist_ok=True)

data = prepared_datasets[DATASET_NAME]

# ------------------------------------------------------------
# 1. INSTRUMENTAÇÃO DAS ETAPAS INTERNAS
# ------------------------------------------------------------
# O código da função original é copiado apenas para memória,
# onde inserimos marcadores de tempo entre instruções.
# O arquivo original no Drive não é alterado.
#
# Os tempos exclusivos são obtidos por diferença entre
# marcadores consecutivos. Assim evitamos sobreposição.

source = textwrap.dedent(
    inspect.getsource(cfis_function)
)

tree = ast.parse(source)
function_node = next(
    n for n in tree.body
    if isinstance(n, ast.FunctionDef)
    and n.name == cfis_function.__name__
)

# Relógio para cada instrução de nível superior.
# Isso inclui loops completos, condicionais e atribuições.
timing_records = []

def _cfis_mark(label):
    timing_records.append(
        (label, time.perf_counter())
    )

original_body = function_node.body
new_body = []

for index, statement in enumerate(original_body):

    label = (
        f"{index:02d}: "
        f"{type(statement).__name__} "
        f"(linha {statement.lineno})"
    )

    marker = ast.Expr(
        value=ast.Call(
            func=ast.Name(
                id="_cfis_mark",
                ctx=ast.Load()
            ),
            args=[ast.Constant(value=label)],
            keywords=[]
        )
    )

    new_body.append(marker)
    new_body.append(statement)

function_node.body = new_body

# Preservar assinatura e semântica da função.
ast.fix_missing_locations(tree)

namespace = dict(cfis_function.__globals__)
namespace["_cfis_mark"] = _cfis_mark

compiled = compile(
    tree,
    filename="<CFIS_PROFILING_IN_MEMORY>",
    mode="exec"
)

exec(compiled, namespace)

instrumented_cfis = namespace[
    cfis_function.__name__
]

# ------------------------------------------------------------
# 2. EXECUÇÃO INSTRUMENTADA
# ------------------------------------------------------------

start_total = time.perf_counter()

result = instrumented_cfis(
    data["X_train"],
    data["y_train"],
    task=data["task"],
    model=None,
    n_repeats=N_REPEATS,
    random_state=SEED
)

end_total = time.perf_counter()

# ------------------------------------------------------------
# 3. TEMPO POR INSTRUÇÃO
# ------------------------------------------------------------

statement_rows = []

for i, (label, start) in enumerate(timing_records):

    end = (
        timing_records[i + 1][1]
        if i + 1 < len(timing_records)
        else end_total
    )

    statement_rows.append({
        "Instruction": label,
        "Time_s": end - start
    })

statement_df = pd.DataFrame(statement_rows)

# ------------------------------------------------------------
# 4. CLASSIFICAÇÃO DAS ETAPAS
# ------------------------------------------------------------

# Mostrar o código original numerado para interpretação
# exata dos tempos, sem supor linhas ou operações.

source_lines = source.splitlines()

print("=" * 76)
print("CFIS — CÓDIGO ORIGINAL NUMERADO")
print("=" * 76)

for i, line in enumerate(source_lines, start=1):
    if line.strip():
        print(f"{i:03d} | {line}")

print("\n" + "=" * 76)
print("PROFILING POR INSTRUÇÃO DE NÍVEL SUPERIOR")
print("=" * 76)

statement_df["Percentage"] = (
    100 * statement_df["Time_s"]
    / (end_total - start_total)
)

display(
    statement_df.sort_values(
        "Time_s",
        ascending=False
    ).round(6)
)

# ------------------------------------------------------------
# 5. SALVAR RESULTADOS
# ------------------------------------------------------------

output_path = (
    PROFILE_DIR
    / f"{DATASET_NAME}_detailed_profile.csv"
)

statement_df.to_csv(
    output_path,
    index=False
)

summary = {
    "dataset": DATASET_NAME,
    "n_repeats": N_REPEATS,
    "seed": SEED,
    "total_time_s": end_total - start_total,
    "profiling_type": "AST_top_level_statement"
}

summary_path = (
    PROFILE_DIR
    / f"{DATASET_NAME}_detailed_profile.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        summary,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nTempo total:", round(
    end_total - start_total, 6
), "s")

print("\nArquivos salvos:")
print(output_path)
print(summary_path)

print("\nTOP 10 CFIS:")
display(result.head(10))

print("\nCÉLULA 4 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS EFFICIENCY OPTIMIZATION — CÉLULA 5
# AUDITORIA NUMÉRICA DOS COMPONENTES P, C E R
# ============================================================

import json
import time
import hashlib
import warnings
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
import scipy
import sklearn

from scipy import stats
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import permutation_importance

DATASET_NAME = "Breast_Cancer"
N_REPEATS = 10
SEED = 42
TOLERANCE = 1e-12

# ------------------------------------------------------------
# 1. CONFIGURAÇÃO
# ------------------------------------------------------------

BASE_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
)

REFERENCE_DIR = BASE_DIR / "01_Audit" / "Numerical_Reference"
REFERENCE_DIR.mkdir(parents=True, exist_ok=True)

data = prepared_datasets[DATASET_NAME]

X_train = data["X_train"]
y_train = data["y_train"]

feature_names = X_train.columns.tolist()

X_values = X_train.to_numpy(dtype=float)
y_values = np.asarray(y_train).ravel()

assert X_values.shape[0] == len(y_values)
assert np.isfinite(X_values).all()

# ------------------------------------------------------------
# 2. EXECUÇÃO DO CFIS ORIGINAL
# ------------------------------------------------------------

start = time.perf_counter()

reference = cfis_function(
    X_train,
    y_train,
    task="classification",
    model=None,
    n_repeats=N_REPEATS,
    random_state=SEED
)

reference_time = time.perf_counter() - start

# ------------------------------------------------------------
# 3. RECONSTRUÇÃO INDEPENDENTE DOS COMPONENTES
# ------------------------------------------------------------

X_scaled = StandardScaler().fit_transform(X_values)

estimator = LogisticRegression(
    random_state=SEED,
    max_iter=1000
)

estimator.fit(X_scaled, y_values)

# P — Permutation Importance
start = time.perf_counter()

perm_result = permutation_importance(
    estimator,
    X_scaled,
    y_values,
    n_repeats=N_REPEATS,
    random_state=SEED,
    scoring="accuracy",
    n_jobs=1
)

permutation_time = time.perf_counter() - start

P = np.nan_to_num(
    np.abs(perm_result.importances_mean),
    nan=0.0,
    posinf=0.0,
    neginf=0.0
)

# C — Coeficientes absolutos
coef = np.asarray(estimator.coef_)

if coef.ndim == 1:
    C = np.abs(coef)
elif coef.ndim == 2:
    C = np.mean(np.abs(coef), axis=0)
else:
    raise ValueError("Dimensão inesperada de coeficientes.")

# R — ANOVA
R = np.zeros(X_scaled.shape[1])
classes = np.unique(y_values)

with warnings.catch_warnings():
    warnings.simplefilter("ignore")

    for j in range(X_scaled.shape[1]):

        groups = [
            X_scaled[y_values == cls, j]
            for cls in classes
        ]

        if (
            all(len(g) > 0 for g in groups)
            and np.var(X_scaled[:, j]) > 0
        ):
            f_value, _ = stats.f_oneway(*groups)

            if np.isfinite(f_value):
                R[j] = max(0.0, float(f_value))

# ------------------------------------------------------------
# 4. FÓRMULA ORIGINAL
# ------------------------------------------------------------

def normalize_component(values):
    values = np.nan_to_num(
        np.asarray(values, dtype=float),
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    total = np.sum(values)

    if total > 0:
        return values / total

    return np.zeros_like(values)


P_norm = normalize_component(P)
C_norm = normalize_component(C)
R_norm = normalize_component(R)

CFIS = (P_norm + C_norm + R_norm) / 3.0
CFIS_percentage = normalize_component(CFIS) * 100.0

reconstructed = pd.DataFrame({
    "Feature": feature_names,
    "Permutation_Importance": P_norm,
    "Coefficient_Abs": C_norm,
    "Correlation_Abs": R_norm,
    "CFIS_Score": CFIS,
    "CFIS_Percentage": CFIS_percentage
}).sort_values(
    "CFIS_Score",
    ascending=False,
    kind="stable"
).reset_index(drop=True)

# ------------------------------------------------------------
# 5. AUDITORIA DE EQUIVALÊNCIA
# ------------------------------------------------------------

columns = [
    "Permutation_Importance",
    "Coefficient_Abs",
    "Correlation_Abs",
    "CFIS_Score",
    "CFIS_Percentage"
]

same_order = (
    reference["Feature"].tolist()
    == reconstructed["Feature"].tolist()
)

audit_rows = []

for column in columns:

    difference = np.abs(
        reference[column].to_numpy()
        - reconstructed[column].to_numpy()
    )

    audit_rows.append({
        "Component": column,
        "Max_Abs_Difference": float(difference.max()),
        "Equivalent": bool(
            np.allclose(
                reference[column].to_numpy(),
                reconstructed[column].to_numpy(),
                atol=TOLERANCE,
                rtol=0
            )
        )
    })

audit = pd.DataFrame(audit_rows)

all_equivalent = (
    same_order
    and audit["Equivalent"].all()
)

# ------------------------------------------------------------
# 6. SALVAR RESULTADOS
# ------------------------------------------------------------

# Identificador único evita sobrescrever auditorias.
run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

reference.to_csv(
    REFERENCE_DIR / f"{prefix}_cfis_reference.csv",
    index=False,
    float_format="%.17g"
)

reconstructed.to_csv(
    REFERENCE_DIR / f"{prefix}_reconstructed.csv",
    index=False,
    float_format="%.17g"
)

audit.to_csv(
    REFERENCE_DIR / f"{prefix}_audit.csv",
    index=False
)

np.savez_compressed(
    REFERENCE_DIR / f"{prefix}_components.npz",
    P_raw=P,
    C_raw=C,
    R_raw=R,
    P_normalized=P_norm,
    C_normalized=C_norm,
    R_normalized=R_norm,
    CFIS=CFIS,
    permutation_importances=perm_result.importances,
    permutation_mean=perm_result.importances_mean,
    permutation_std=perm_result.importances_std,
    feature_names=np.asarray(feature_names, dtype=str)
)

source_path = (
    Path(DIRS["logs"])
    / "CFIS_replication_v1.py"
)

source_sha256 = (
    hashlib.sha256(source_path.read_bytes()).hexdigest()
    if source_path.exists()
    else None
)

metadata = {
    "dataset": DATASET_NAME,
    "samples": int(X_values.shape[0]),
    "features": int(X_values.shape[1]),
    "n_repeats": N_REPEATS,
    "random_state": SEED,
    "task": "classification",
    "estimator": "LogisticRegression(max_iter=1000)",
    "scoring": "accuracy",
    "reference_time_s": reference_time,
    "permutation_time_s": permutation_time,
    "same_feature_order": same_order,
    "all_equivalent": bool(all_equivalent),
    "tolerance": TOLERANCE,
    "source_sha256": source_sha256,
    "numpy_version": np.__version__,
    "scipy_version": scipy.__version__,
    "sklearn_version": sklearn.__version__,
    "run_id": run_id
}

with open(
    REFERENCE_DIR / f"{prefix}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2, ensure_ascii=False)

# ------------------------------------------------------------
# 7. EXIBIÇÃO
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — AUDITORIA NUMÉRICA")
print("=" * 72)

print(f"Dataset: {DATASET_NAME}")
print(f"Amostras: {len(X_train)}")
print(f"Features: {X_train.shape[1]}")
print(f"Tempo CFIS original: {reference_time:.4f} s")
print(f"Tempo Permutation Importance: {permutation_time:.4f} s")

print("\nOrdem de características idêntica:", same_order)

print("\nEQUIVALÊNCIA POR COMPONENTE")
display(audit)

print("\nEQUIVALÊNCIA TOTAL:", all_equivalent)

print("\nTOP 10 CFIS ORIGINAL")
display(reference.head(10))

print("\nReferência salva em:")
print(REFERENCE_DIR)

assert all_equivalent, (
    "A reconstrução não reproduziu exatamente o CFIS. "
    "Investigar antes de otimizar."
)

print("\nCÉLULA 5 CONCLUÍDA COM SUCESSO.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 6
# PERMUTATION IMPORTANCE ACELERADA
# TESTE DE EQUIVALÊNCIA EXATA — CLASSIFICAÇÃO BINÁRIA
# ============================================================

import time
import json
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import permutation_importance
from sklearn.utils import check_random_state

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10

data = prepared_datasets[DATASET_NAME]

X = StandardScaler().fit_transform(
    data["X_train"].to_numpy(dtype=float)
)
y = np.asarray(data["y_train"]).ravel()

model = LogisticRegression(
    random_state=SEED,
    max_iter=1000
)
model.fit(X, y)

assert len(model.classes_) == 2
assert model.coef_.shape == (1, X.shape[1])

# ------------------------------------------------------------
# 1. IMPLEMENTAÇÃO OTIMIZADA
# ------------------------------------------------------------

def fast_binary_permutation_importance(
    estimator,
    X,
    y,
    n_repeats=10,
    random_state=42
):
    X = np.asarray(X)
    y = np.asarray(y)

    n_samples, n_features = X.shape

    coef = estimator.coef_[0]
    classes = estimator.classes_

    # Calculado uma única vez.
    base_scores = estimator.decision_function(X)

    baseline_predictions = estimator.predict(X)
    baseline_accuracy = np.mean(
        baseline_predictions == y
    )

    # Reproduz a geração da semente interna
    # utilizada pelo permutation_importance do sklearn.
    rng = check_random_state(random_state)
    internal_seed = rng.randint(
        np.iinfo(np.int32).max
    )

    importances = np.empty(
        (n_features, n_repeats),
        dtype=float
    )

    for j in range(n_features):

        feature_rng = np.random.RandomState(
            internal_seed
        )

        shuffle_idx = np.arange(n_samples)

        original_column = X[:, j]

        for r in range(n_repeats):

            # Índices embaralhados cumulativamente,
            # conforme a rotina do scikit-learn.
            feature_rng.shuffle(shuffle_idx)

            permuted_column = X[shuffle_idx, j]

            updated_scores = (
                base_scores
                + (permuted_column - original_column)
                * coef[j]
            )

            predictions = np.where(
                updated_scores > 0,
                classes[1],
                classes[0]
            )

            accuracy = np.mean(predictions == y)

            importances[j, r] = (
                baseline_accuracy - accuracy
            )

    return importances

# ------------------------------------------------------------
# 2. EXECUÇÃO DE REFERÊNCIA
# ------------------------------------------------------------

start = time.perf_counter()

reference = permutation_importance(
    model,
    X,
    y,
    n_repeats=N_REPEATS,
    random_state=SEED,
    scoring="accuracy",
    n_jobs=1
)

reference_time = time.perf_counter() - start

# ------------------------------------------------------------
# 3. EXECUÇÃO OTIMIZADA
# ------------------------------------------------------------

start = time.perf_counter()

optimized = fast_binary_permutation_importance(
    model,
    X,
    y,
    n_repeats=N_REPEATS,
    random_state=SEED
)

optimized_time = time.perf_counter() - start

# ------------------------------------------------------------
# 4. AUDITORIA POR FEATURE E REPETIÇÃO
# ------------------------------------------------------------

original_matrix = reference.importances

difference = np.abs(
    original_matrix - optimized
)

max_difference = float(
    np.max(difference)
)

exact_equal = bool(
    np.array_equal(original_matrix, optimized)
)

close_equal = bool(
    np.allclose(
        original_matrix,
        optimized,
        atol=1e-12,
        rtol=0
    )
)

reference_mean = reference.importances_mean
optimized_mean = optimized.mean(axis=1)

mean_difference = float(
    np.max(
        np.abs(
            reference_mean - optimized_mean
        )
    )
)

speedup = reference_time / optimized_time

# ------------------------------------------------------------
# 5. RESULTADOS
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — PERMUTATION IMPORTANCE OTIMIZADA")
print("=" * 72)

print(f"Dataset: {DATASET_NAME}")
print(f"Features: {X.shape[1]}")
print(f"Repetições: {N_REPEATS}")

print(f"\nSklearn: {reference_time:.6f} s")
print(f"Otimizado: {optimized_time:.6f} s")
print(f"Speedup exploratório: {speedup:.2f}x")

print("\nAUDITORIA NUMÉRICA")
print("Matrizes exatamente iguais:", exact_equal)
print("Equivalentes (1e-12):", close_equal)
print("Diferença máxima:", max_difference)
print("Diferença máxima das médias:", mean_difference)

comparison = pd.DataFrame({
    "Feature": data["X_train"].columns,
    "Reference_P": reference_mean,
    "Optimized_P": optimized_mean,
    "Abs_Difference": np.abs(
        reference_mean - optimized_mean
    )
})

display(
    comparison.sort_values(
        "Reference_P",
        ascending=False
    ).head(15)
)

# ------------------------------------------------------------
# 6. SALVAR EXPERIMENTO
# ------------------------------------------------------------

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "03_Optimized"
    / "Permutation_v1"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

comparison.to_csv(
    OUTPUT_DIR / f"{prefix}_comparison.csv",
    index=False
)

np.savez_compressed(
    OUTPUT_DIR / f"{prefix}_audit.npz",
    reference=original_matrix,
    optimized=optimized
)

metrics = {
    "dataset": DATASET_NAME,
    "n_repeats": N_REPEATS,
    "seed": SEED,
    "reference_time_s": reference_time,
    "optimized_time_s": optimized_time,
    "speedup_exploratory": speedup,
    "exact_equal": exact_equal,
    "close_equal": close_equal,
    "max_difference": max_difference,
    "mean_difference": mean_difference
}

with open(
    OUTPUT_DIR / f"{prefix}_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metrics, f, indent=2)

print("\nResultados salvos em:", OUTPUT_DIR)

if close_equal:
    print("\nAUDITORIA APROVADA.")
else:
    print("\nAUDITORIA NÃO APROVADA.")
    print("Investigar divergências antes de integrar ao CFIS.")

print("\nCÉLULA 6 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 7
# DIAGNÓSTICO DAS DIVERGÊNCIAS DA PERMUTATION IMPORTANCE
# ============================================================

import numpy as np
import pandas as pd
import inspect
import sklearn

from sklearn.inspection import permutation_importance
from sklearn.utils import check_random_state

# Reutiliza X, y, model e fast_binary_permutation_importance
# da Célula 6.

assert len(model.classes_) == 2

n_samples, n_features = X.shape

# ------------------------------------------------------------
# 1. INSPECIONAR IMPLEMENTAÇÃO INSTALADA
# ------------------------------------------------------------

import sklearn.inspection._permutation_importance as pi_module

print("=" * 72)
print("VERSÃO DO SCIKIT-LEARN")
print("=" * 72)
print(sklearn.__version__)

print("\nRotina interna de permutação:")
print(inspect.getsource(
    pi_module._calculate_permutation_scores
))

# ------------------------------------------------------------
# 2. RECONSTRUIR A SEQUÊNCIA DE PERMUTAÇÕES
# ------------------------------------------------------------

rng = check_random_state(SEED)
internal_seed = rng.randint(np.iinfo(np.int32).max)

feature_rng = check_random_state(internal_seed)

shuffle_idx = np.arange(n_samples)

permutations = []

for r in range(N_REPEATS):
    feature_rng.shuffle(shuffle_idx)
    permutations.append(shuffle_idx.copy())

# ------------------------------------------------------------
# 3. COMPARAR PREDIÇÕES POR DUAS ESTRATÉGIAS
# ------------------------------------------------------------

base_scores = model.decision_function(X)
coef = model.coef_[0]
classes = model.classes_

records = []

for j in range(n_features):

    for r, idx in enumerate(permutations):

        X_permuted = X.copy()
        X_permuted[:, j] = X[idx, j]

        # Predição padrão do scikit-learn.
        pred_sklearn = model.predict(X_permuted)

        # Predição com atualização incremental.
        updated_scores = (
            base_scores
            + (X[idx, j] - X[:, j]) * coef[j]
        )

        pred_fast = np.where(
            updated_scores > 0,
            classes[1],
            classes[0]
        )

        mismatches = int(np.sum(
            pred_sklearn != pred_fast
        ))

        records.append({
            "Feature_Index": j,
            "Repeat": r,
            "Prediction_Mismatches": mismatches,
            "Accuracy_Sklearn": np.mean(
                pred_sklearn == y
            ),
            "Accuracy_Fast": np.mean(
                pred_fast == y
            ),
            "Max_Score_Difference": float(
                np.max(
                    np.abs(
                        model.decision_function(X_permuted)
                        - updated_scores
                    )
                )
            )
        })

diagnostic = pd.DataFrame(records)

# ------------------------------------------------------------
# 4. RESUMO
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("DIAGNÓSTICO DE PREDIÇÕES")
print("=" * 72)

print(
    "Total de previsões divergentes:",
    diagnostic["Prediction_Mismatches"].sum()
)

print(
    "Combinações feature/repeat divergentes:",
    int(
        (diagnostic["Prediction_Mismatches"] > 0).sum()
    )
)

print(
    "Diferença máxima entre escores:",
    diagnostic["Max_Score_Difference"].max()
)

print("\nPRIMEIRAS DIVERGÊNCIAS")

display(
    diagnostic[
        diagnostic["Prediction_Mismatches"] > 0
    ].head(20)
)

print("\nCÉLULA 7 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 8
# PERMUTATION IMPORTANCE V2 — CORREÇÃO CUMULATIVA
# ============================================================

import time
import json
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.inspection import permutation_importance
from sklearn.utils import check_random_state

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10

# X, y e model foram preparados na Célula 6.
assert X.shape[0] == len(y)
assert len(model.classes_) == 2
assert model.coef_.shape == (1, X.shape[1])

def fast_binary_permutation_importance_v2(
    estimator, X, y, n_repeats=10, random_state=42
):
    X = np.asarray(X)
    y = np.asarray(y)

    n_samples, n_features = X.shape
    coef = estimator.coef_[0]
    classes = estimator.classes_

    base_scores = estimator.decision_function(X)
    baseline_accuracy = np.mean(estimator.predict(X) == y)

    # Mesma semente interna utilizada pelo scikit-learn.
    rng = check_random_state(random_state)
    internal_seed = rng.randint(np.iinfo(np.int32).max)

    importances = np.empty(
        (n_features, n_repeats), dtype=float
    )

    for j in range(n_features):

        feature_rng = check_random_state(internal_seed)
        shuffle_idx = np.arange(n_samples)

        # A coluna é modificada cumulativamente.
        current_column = X[:, j].copy()
        original_column = X[:, j]

        for r in range(n_repeats):

            feature_rng.shuffle(shuffle_idx)

            # CORREÇÃO: permutar a coluna já modificada.
            current_column = current_column[shuffle_idx].copy()

            updated_scores = (
                base_scores
                + (current_column - original_column) * coef[j]
            )

            predictions = np.where(
                updated_scores > 0,
                classes[1],
                classes[0]
            )

            accuracy = np.mean(predictions == y)

            importances[j, r] = (
                baseline_accuracy - accuracy
            )

    return importances


# ------------------------------------------------------------
# 1. REFERÊNCIA SCIKIT-LEARN
# ------------------------------------------------------------

start = time.perf_counter()

reference = permutation_importance(
    model,
    X,
    y,
    n_repeats=N_REPEATS,
    random_state=SEED,
    scoring="accuracy",
    n_jobs=1
)

reference_time = time.perf_counter() - start


# ------------------------------------------------------------
# 2. IMPLEMENTAÇÃO OTIMIZADA V2
# ------------------------------------------------------------

start = time.perf_counter()

optimized = fast_binary_permutation_importance_v2(
    model,
    X,
    y,
    n_repeats=N_REPEATS,
    random_state=SEED
)

optimized_time = time.perf_counter() - start


# ------------------------------------------------------------
# 3. AUDITORIA COMPLETA
# ------------------------------------------------------------

original_matrix = reference.importances

difference = np.abs(original_matrix - optimized)

exact_equal = bool(
    np.array_equal(original_matrix, optimized)
)

close_equal = bool(
    np.allclose(
        original_matrix,
        optimized,
        atol=1e-12,
        rtol=0
    )
)

max_difference = float(difference.max())

n_divergences = int(
    np.count_nonzero(difference > 1e-12)
)

reference_mean = reference.importances_mean
optimized_mean = optimized.mean(axis=1)

mean_difference = float(
    np.max(np.abs(reference_mean - optimized_mean))
)

speedup = reference_time / optimized_time

comparison = pd.DataFrame({
    "Feature": data["X_train"].columns,
    "Reference_P": reference_mean,
    "Optimized_P": optimized_mean,
    "Abs_Difference": np.abs(
        reference_mean - optimized_mean
    )
})

# ------------------------------------------------------------
# 4. EXIBIÇÃO
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — PERMUTATION IMPORTANCE V2")
print("=" * 72)

print(f"Dataset: {DATASET_NAME}")
print(f"Features: {X.shape[1]}")
print(f"Repetições: {N_REPEATS}")

print("\nTEMPOS")
print(f"Scikit-learn: {reference_time:.6f} s")
print(f"Otimizado V2: {optimized_time:.6f} s")
print(f"Speedup exploratório: {speedup:.2f}x")

print("\nAUDITORIA NUMÉRICA")
print("Matrizes exatamente iguais:", exact_equal)
print("Equivalentes (1e-12):", close_equal)
print("Diferença máxima:", max_difference)
print("Divergências > 1e-12:", n_divergences)
print("Diferença máxima das médias:", mean_difference)

print("\nCOMPARAÇÃO DAS 15 FEATURES MAIS IMPORTANTES")
display(
    comparison.sort_values(
        "Reference_P",
        ascending=False
    ).head(15)
)

# ------------------------------------------------------------
# 5. SALVAR EXPERIMENTO
# ------------------------------------------------------------

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "03_Optimized"
    / "Permutation_v2"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

comparison.to_csv(
    OUTPUT_DIR / f"{prefix}_comparison.csv",
    index=False,
    float_format="%.17g"
)

np.savez_compressed(
    OUTPUT_DIR / f"{prefix}_audit.npz",
    reference=original_matrix,
    optimized=optimized
)

metrics = {
    "dataset": DATASET_NAME,
    "n_repeats": N_REPEATS,
    "seed": SEED,
    "reference_time_s": reference_time,
    "optimized_time_s": optimized_time,
    "speedup_exploratory": speedup,
    "exact_equal": exact_equal,
    "close_equal": close_equal,
    "max_difference": max_difference,
    "n_divergences": n_divergences,
    "mean_difference": mean_difference
}

with open(
    OUTPUT_DIR / f"{prefix}_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metrics, f, indent=2)

print("\nResultados salvos em:", OUTPUT_DIR)

if exact_equal:
    print("\nAUDITORIA APROVADA — IGUALDADE EXATA.")
elif close_equal:
    print("\nAUDITORIA APROVADA — TOLERÂNCIA 1e-12.")
else:
    print("\nAUDITORIA NÃO APROVADA.")
    print("Não integrar ao CFIS antes da correção.")

print("\nCÉLULA 8 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 9
# INTEGRAÇÃO CONTROLADA DA PERMUTATION V2
# AUDITORIA DO CFIS COMPLETO — BREAST CANCER
# ============================================================

import json
import time
from datetime import datetime, timezone
from pathlib import Path
from types import SimpleNamespace

import numpy as np
import pandas as pd

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10

data = prepared_datasets[DATASET_NAME]

X_train = data["X_train"]
y_train = data["y_train"]

# ------------------------------------------------------------
# 1. PREPARAR SUBSTITUIÇÃO TEMPORÁRIA
# ------------------------------------------------------------

cfis_globals = cfis_function.__globals__

original_permutation = cfis_globals[
    "permutation_importance"
]

assert callable(fast_binary_permutation_importance_v2)

def optimized_permutation_adapter(
    estimator,
    X,
    y,
    n_repeats=10,
    random_state=42,
    scoring=None,
    n_jobs=None,
    **kwargs
):
    # V2 foi validada apenas para este cenário.
    if (
        scoring != "accuracy"
        or len(estimator.classes_) != 2
        or estimator.coef_.shape != (1, X.shape[1])
        or kwargs
    ):
        raise ValueError(
            "Cenário ainda não validado para Permutation V2."
        )

    values = fast_binary_permutation_importance_v2(
        estimator,
        X,
        y,
        n_repeats=n_repeats,
        random_state=random_state
    )

    return SimpleNamespace(
        importances=values,
        importances_mean=np.mean(values, axis=1),
        importances_std=np.std(values, axis=1)
    )

# ------------------------------------------------------------
# 2. EXECUTAR CFIS ORIGINAL
# ------------------------------------------------------------

start = time.perf_counter()

cfis_original = cfis_function(
    X_train,
    y_train,
    task="classification",
    model=None,
    n_repeats=N_REPEATS,
    random_state=SEED
)

original_time = time.perf_counter() - start

# ------------------------------------------------------------
# 3. EXECUTAR CFIS OTIMIZADO
# ------------------------------------------------------------

try:
    cfis_globals["permutation_importance"] = (
        optimized_permutation_adapter
    )

    start = time.perf_counter()

    cfis_optimized = cfis_function(
        X_train,
        y_train,
        task="classification",
        model=None,
        n_repeats=N_REPEATS,
        random_state=SEED
    )

    optimized_time = time.perf_counter() - start

finally:
    # Restaurar mesmo em caso de erro.
    cfis_globals["permutation_importance"] = (
        original_permutation
    )

assert (
    cfis_globals["permutation_importance"]
    is original_permutation
)

# ------------------------------------------------------------
# 4. AUDITORIA NUMÉRICA COMPLETA
# ------------------------------------------------------------

columns = [
    "Permutation_Importance",
    "Coefficient_Abs",
    "Correlation_Abs",
    "CFIS_Score",
    "CFIS_Percentage"
]

same_features = (
    cfis_original["Feature"].tolist()
    == cfis_optimized["Feature"].tolist()
)

audit_rows = []

for column in columns:

    a = cfis_original[column].to_numpy()
    b = cfis_optimized[column].to_numpy()

    audit_rows.append({
        "Component": column,
        "Exact_Equal": bool(np.array_equal(a, b)),
        "Max_Abs_Difference": float(
            np.max(np.abs(a - b))
        )
    })

audit = pd.DataFrame(audit_rows)

exact_equivalence = bool(
    same_features
    and audit["Exact_Equal"].all()
)

# ------------------------------------------------------------
# 5. SALVAR AUDITORIA
# ------------------------------------------------------------

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Full_CFIS_V2"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

cfis_original.to_csv(
    OUTPUT_DIR / f"{prefix}_original.csv",
    index=False,
    float_format="%.17g"
)

cfis_optimized.to_csv(
    OUTPUT_DIR / f"{prefix}_optimized.csv",
    index=False,
    float_format="%.17g"
)

audit.to_csv(
    OUTPUT_DIR / f"{prefix}_audit.csv",
    index=False
)

metrics = {
    "dataset": DATASET_NAME,
    "n_repeats": N_REPEATS,
    "seed": SEED,
    "original_time_s": original_time,
    "optimized_time_s": optimized_time,
    "speedup_exploratory": (
        original_time / optimized_time
    ),
    "same_feature_order": same_features,
    "exact_equivalence": exact_equivalence,
    "max_abs_difference": float(
        audit["Max_Abs_Difference"].max()
    ),
    "scope": "binary_logistic_regression"
}

with open(
    OUTPUT_DIR / f"{prefix}_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metrics,
        f,
        indent=2,
        ensure_ascii=False
    )

# ------------------------------------------------------------
# 6. RESULTADOS
# ------------------------------------------------------------

print("=" * 72)
print("CFIS COMPLETO — ORIGINAL VS OTIMIZADO V2")
print("=" * 72)

print(f"Dataset: {DATASET_NAME}")

print("\nDESEMPENHO EXPLORATÓRIO")
print(f"Original:  {original_time:.6f} s")
print(f"Otimizado: {optimized_time:.6f} s")
print(
    f"Speedup:   "
    f"{original_time / optimized_time:.2f}x"
)

print("\nAUDITORIA")
print("Mesma ordem das features:", same_features)

display(audit)

print("\nEquivalência exata:", exact_equivalence)

print("\nTOP 10 — CFIS OTIMIZADO")
display(cfis_optimized.head(10))

print("\nArquivos salvos em:", OUTPUT_DIR)

assert exact_equivalence, (
    "CFIS otimizado divergiu do original. "
    "Não considerar a integração validada."
)

print("\nCÉLULA 9 CONCLUÍDA COM SUCESSO.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 10
# BENCHMARK REPETIDO: TEMPO E CPU
# ORIGINAL VS OTIMIZADO V2
# ============================================================

import gc
import json
import time
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10
N_PAIRS = 7

data = prepared_datasets[DATASET_NAME]
X_train = data["X_train"]
y_train = data["y_train"]

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

assert callable(optimized_permutation_adapter)
assert callable(original_permutation)

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "04_Benchmarks"
    / "Timing_CPU_V2"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

records = []
reference_output = None

def run_benchmark(variant):

    global reference_output

    gc.collect()

    if variant == "Original":
        replacement = original_permutation
    else:
        replacement = optimized_permutation_adapter

    try:
        cfis_globals["permutation_importance"] = replacement

        cpu_start = time.process_time()
        wall_start = time.perf_counter()

        result = cfis_function(
            X_train,
            y_train,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        wall_elapsed = time.perf_counter() - wall_start
        cpu_elapsed = time.process_time() - cpu_start

    finally:
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

    if reference_output is None:
        reference_output = result.copy(deep=True)

    # Equivalência integral em todas as execuções.
    pd.testing.assert_frame_equal(
        result,
        reference_output,
        check_exact=True
    )

    return {
        "Variant": variant,
        "Wall_Time_s": wall_elapsed,
        "CPU_Time_s": cpu_elapsed
    }


# ------------------------------------------------------------
# 1. AQUECIMENTO
# ------------------------------------------------------------

print("Executando aquecimento...")

run_benchmark("Original")
run_benchmark("Optimized")

# ------------------------------------------------------------
# 2. BENCHMARK PAREADO
# ------------------------------------------------------------

print(f"Executando {N_PAIRS} pares de testes...")

for pair in range(1, N_PAIRS + 1):

    # Alternância da ordem de execução.
    order = (
        ["Original", "Optimized"]
        if pair % 2 == 1
        else ["Optimized", "Original"]
    )

    for variant in order:

        measurement = run_benchmark(variant)

        records.append({
            "Pair": pair,
            "Order": order.index(variant) + 1,
            **measurement
        })

    print(f"Par {pair}/{N_PAIRS} concluído.")

results = pd.DataFrame(records)

# ------------------------------------------------------------
# 3. ESTATÍSTICAS
# ------------------------------------------------------------

summary = (
    results.groupby("Variant")
    .agg(
        Wall_Mean_s=("Wall_Time_s", "mean"),
        Wall_Median_s=("Wall_Time_s", "median"),
        Wall_Std_s=("Wall_Time_s", "std"),
        CPU_Mean_s=("CPU_Time_s", "mean"),
        CPU_Median_s=("CPU_Time_s", "median")
    )
)

paired_wall = results.pivot(
    index="Pair",
    columns="Variant",
    values="Wall_Time_s"
)

paired_cpu = results.pivot(
    index="Pair",
    columns="Variant",
    values="CPU_Time_s"
)

paired = pd.DataFrame({
    "Pair": paired_wall.index,
    "Original_Wall_s": paired_wall["Original"].values,
    "Optimized_Wall_s": paired_wall["Optimized"].values,
    "Wall_Speedup": (
        paired_wall["Original"]
        / paired_wall["Optimized"]
    ).values,
    "Original_CPU_s": paired_cpu["Original"].values,
    "Optimized_CPU_s": paired_cpu["Optimized"].values,
    "CPU_Speedup": (
        paired_cpu["Original"]
        / paired_cpu["Optimized"]
    ).values
})

median_wall_speedup = float(
    paired["Wall_Speedup"].median()
)

median_cpu_speedup = float(
    paired["CPU_Speedup"].median()
)

# ------------------------------------------------------------
# 4. SALVAR RESULTADOS
# ------------------------------------------------------------

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

results.to_csv(
    OUTPUT_DIR / f"{prefix}_raw.csv",
    index=False
)

summary.to_csv(
    OUTPUT_DIR / f"{prefix}_summary.csv"
)

paired.to_csv(
    OUTPUT_DIR / f"{prefix}_paired.csv",
    index=False
)

metadata = {
    "dataset": DATASET_NAME,
    "pairs": N_PAIRS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "median_wall_speedup": median_wall_speedup,
    "median_cpu_speedup": median_cpu_speedup,
    "exact_equivalence_all_runs": True,
    "cpu_metric": "time.process_time",
    "memory_profiling_enabled": False
}

with open(
    OUTPUT_DIR / f"{prefix}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2)

# ------------------------------------------------------------
# 5. EXIBIÇÃO
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("CFIS — BENCHMARK REPETIDO")
print("=" * 72)

print("\nESTATÍSTICAS POR IMPLEMENTAÇÃO")
display(summary.round(6))

print("\nRESULTADOS POR PAR")
display(paired.round(4))

print(
    f"\nSpeedup mediano (Wall): "
    f"{median_wall_speedup:.2f}x"
)

print(
    f"Speedup mediano (CPU): "
    f"{median_cpu_speedup:.2f}x"
)

print("\nEquivalência exata em todas as execuções: True")
print("Resultados salvos em:", OUTPUT_DIR)
print("\nCÉLULA 10 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 11
# AUDITORIA DE MEMÓRIA — ORIGINAL VS V2
# ============================================================

import gc
import json
import time
import tracemalloc
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
import psutil

DATASET_NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10
N_PAIRS = 5

data = prepared_datasets[DATASET_NAME]
X_train = data["X_train"]
y_train = data["y_train"]

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

assert callable(optimized_permutation_adapter)
assert not tracemalloc.is_tracing(), (
    "Interrompa outra sessão tracemalloc antes deste teste."
)

process = psutil.Process()
records = []
reference_output = None

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "04_Benchmarks"
    / "Memory_V2"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


def measure_memory(variant):
    gc.collect()

    replacement = (
        original_permutation
        if variant == "Original"
        else optimized_permutation_adapter
    )

    rss_before = process.memory_info().rss

    try:
        cfis_globals["permutation_importance"] = replacement

        tracemalloc.start()

        start = time.perf_counter()

        result = cfis_function(
            X_train,
            y_train,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        elapsed = time.perf_counter() - start

        current_bytes, peak_bytes = (
            tracemalloc.get_traced_memory()
        )

    finally:
        tracemalloc.stop()
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

    rss_after = process.memory_info().rss

    return result, {
        "Variant": variant,
        "Wall_Time_s": elapsed,
        "Python_Peak_MiB": peak_bytes / 1024**2,
        "Python_Current_MiB": current_bytes / 1024**2,
        "RSS_Before_MiB": rss_before / 1024**2,
        "RSS_After_MiB": rss_after / 1024**2,
        "RSS_Delta_MiB": (
            rss_after - rss_before
        ) / 1024**2
    }


print("=" * 72)
print("CFIS — AUDITORIA DE MEMÓRIA")
print("=" * 72)

for pair in range(1, N_PAIRS + 1):

    order = (
        ["Original", "Optimized"]
        if pair % 2
        else ["Optimized", "Original"]
    )

    for position, variant in enumerate(order, 1):

        result, measurement = measure_memory(variant)

        if reference_output is None:
            reference_output = result.copy(deep=True)

        pd.testing.assert_frame_equal(
            result,
            reference_output,
            check_exact=True
        )

        records.append({
            "Pair": pair,
            "Order": position,
            **measurement
        })

        del result

    print(f"Par {pair}/{N_PAIRS} concluído.")

assert (
    cfis_globals["permutation_importance"]
    is original_permutation
)

results = pd.DataFrame(records)

summary = results.groupby("Variant").agg(
    Python_Peak_Median_MiB=(
        "Python_Peak_MiB", "median"
    ),
    Python_Peak_Mean_MiB=(
        "Python_Peak_MiB", "mean"
    ),
    RSS_Before_Median_MiB=(
        "RSS_Before_MiB", "median"
    ),
    RSS_After_Median_MiB=(
        "RSS_After_MiB", "median"
    ),
    RSS_Delta_Median_MiB=(
        "RSS_Delta_MiB", "median"
    )
)

paired = results.pivot(
    index="Pair",
    columns="Variant",
    values="Python_Peak_MiB"
)

paired_comparison = pd.DataFrame({
    "Pair": paired.index,
    "Original_Peak_MiB": paired["Original"].values,
    "Optimized_Peak_MiB": paired["Optimized"].values,
    "Peak_Ratio": (
        paired["Original"] / paired["Optimized"]
    ).values
})

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

prefix = f"{DATASET_NAME}_{run_id}"

results.to_csv(
    OUTPUT_DIR / f"{prefix}_raw.csv",
    index=False
)

summary.to_csv(
    OUTPUT_DIR / f"{prefix}_summary.csv"
)

paired_comparison.to_csv(
    OUTPUT_DIR / f"{prefix}_paired.csv",
    index=False
)

metadata = {
    "dataset": DATASET_NAME,
    "pairs": N_PAIRS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "exact_equivalence_all_runs": True,
    "python_peak_method": "tracemalloc",
    "rss_method": "psutil_before_after",
    "limitations": (
        "tracemalloc does not measure all native allocations; "
        "RSS before/after is not peak RSS; "
        "measurements are affected by allocator reuse."
    )
}

with open(
    OUTPUT_DIR / f"{prefix}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\nRESUMO DE MEMÓRIA")
display(summary.round(4))

print("\nCOMPARAÇÃO PAREADA — PICO PYTHON")
display(paired_comparison.round(4))

print("\nMEDIÇÕES INDIVIDUAIS")
display(results.round(4))

print("\nEquivalência exata: True")
print("Resultados salvos em:", OUTPUT_DIR)
print("\nCÉLULA 11 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 12
# PICO RSS EM SUBPROCESSOS ISOLADOS
# ============================================================

import os
import sys
import json
import time
import pickle
import tempfile
import subprocess
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import psutil

DATASET_NAME = "Breast_Cancer"
N_PAIRS = 3
SEED = 42
N_REPEATS = 10

data = prepared_datasets[DATASET_NAME]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "04_Benchmarks"
    / "Isolated_RSS_V2"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

# O código original é carregado de seu arquivo,
# sem modificar o conteúdo.
ORIGINAL_FILE = (
    Path(DIRS["logs"]) / "CFIS_replication_v1.py"
)
assert ORIGINAL_FILE.exists()

# O subprocesso recebe apenas dados de treinamento.
payload = {
    "X": data["X_train"],
    "y": data["y_train"],
    "seed": SEED,
    "n_repeats": N_REPEATS
}

worker_code = r'''
import sys, pickle, json, time, os, warnings
import numpy as np
import pandas as pd
import psutil
from scipy import stats
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression, Lasso
from sklearn.inspection import permutation_importance
from sklearn.utils import check_random_state
from types import SimpleNamespace

payload_path, source_path, variant, output_path = sys.argv[1:]

with open(payload_path, "rb") as f:
    payload = pickle.load(f)

namespace = {
    "__name__": "cfis_original_module",
    "pd": pd, "np": np, "warnings": warnings,
    "stats": stats, "clone": clone,
    "StandardScaler": StandardScaler,
    "LogisticRegression": LogisticRegression,
    "Lasso": Lasso,
    "permutation_importance": permutation_importance
}

with open(source_path, "r", encoding="utf-8") as f:
    source = f.read()

exec(compile(source, source_path, "exec"), namespace)
cfis = namespace["CFIS_feature_selection"]

def fast_v2(estimator, X, y, n_repeats=10, random_state=42,
            scoring=None, n_jobs=None, **kwargs):
    if scoring != "accuracy" or kwargs:
        raise ValueError("Cenário não suportado")
    if len(estimator.classes_) != 2:
        raise ValueError("Apenas classificação binária")
    X = np.asarray(X)
    y = np.asarray(y)
    n, p = X.shape
    coef = estimator.coef_[0]
    classes = estimator.classes_
    scores = estimator.decision_function(X)
    baseline = np.mean(estimator.predict(X) == y)
    seed = check_random_state(random_state).randint(
        np.iinfo(np.int32).max
    )
    values = np.empty((p, n_repeats))
    for j in range(p):
        rng = check_random_state(seed)
        idx = np.arange(n)
        original = X[:, j]
        current = original.copy()
        for r in range(n_repeats):
            rng.shuffle(idx)
            current = current[idx].copy()
            updated = scores + (current-original)*coef[j]
            pred = np.where(updated > 0, classes[1], classes[0])
            values[j, r] = baseline - np.mean(pred == y)
    return SimpleNamespace(
        importances=values,
        importances_mean=values.mean(axis=1),
        importances_std=values.std(axis=1)
    )

if variant == "Optimized":
    namespace["permutation_importance"] = fast_v2

# Registra pico de RSS do processo desde o início
# usando /proc/self/status (VmHWM, Linux).
start = time.perf_counter()
result = cfis(
    payload["X"], payload["y"],
    task="classification",
    model=None,
    n_repeats=payload["n_repeats"],
    random_state=payload["seed"]
)
elapsed = time.perf_counter() - start

with open("/proc/self/status") as f:
    status = f.read()

import re
match = re.search(r"^VmHWM:\s+(\d+)\s+kB", status, re.M)
peak_mib = int(match.group(1)) / 1024 if match else None

with open(output_path, "wb") as f:
    pickle.dump({
        "variant": variant,
        "wall_time_s": elapsed,
        "peak_rss_mib": peak_mib,
        "result": result
    }, f)
'''

records = []
reference = None

with tempfile.TemporaryDirectory() as tmp:
    tmp = Path(tmp)
    payload_path = tmp / "payload.pkl"
    worker_path = tmp / "worker.py"

    with open(payload_path, "wb") as f:
        pickle.dump(payload, f)

    worker_path.write_text(worker_code, encoding="utf-8")

    for pair in range(1, N_PAIRS + 1):
        order = (
            ["Original", "Optimized"]
            if pair % 2
            else ["Optimized", "Original"]
        )

        for variant in order:
            result_path = tmp / f"result_{pair}_{variant}.pkl"

            completed = subprocess.run(
                [
                    sys.executable,
                    str(worker_path),
                    str(payload_path),
                    str(ORIGINAL_FILE),
                    variant,
                    str(result_path)
                ],
                capture_output=True,
                text=True,
                timeout=180
            )

            if completed.returncode != 0:
                print(completed.stderr[-4000:])
                raise RuntimeError(
                    f"Falha no subprocesso {variant}"
                )

            with open(result_path, "rb") as f:
                measured = pickle.load(f)

            result = measured.pop("result")

            if reference is None:
                reference = result
            else:
                pd.testing.assert_frame_equal(
                    result, reference, check_exact=True
                )

            records.append({
                "Pair": pair,
                **measured
            })

        print(f"Par {pair}/{N_PAIRS} concluído.")

results = pd.DataFrame(records)

summary = results.groupby("variant").agg(
    Peak_RSS_Median_MiB=("peak_rss_mib", "median"),
    Peak_RSS_Mean_MiB=("peak_rss_mib", "mean"),
    Wall_Median_s=("wall_time_s", "median")
)

run_prefix = f"{DATASET_NAME}_{run_id}"

results.to_csv(
    OUTPUT_DIR / f"{run_prefix}_raw.csv",
    index=False
)
summary.to_csv(
    OUTPUT_DIR / f"{run_prefix}_summary.csv"
)

metadata = {
    "dataset": DATASET_NAME,
    "pairs": N_PAIRS,
    "n_repeats": N_REPEATS,
    "seed": SEED,
    "equivalence": "exact",
    "peak_rss_method": "Linux VmHWM",
    "scope": "whole subprocess lifetime",
    "limitations": (
        "Includes interpreter imports, dataset loading, "
        "and model execution; not algorithm-only peak."
    )
}

with open(
    OUTPUT_DIR / f"{run_prefix}_metadata.json",
    "w", encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2)

print("\n" + "=" * 72)
print("CFIS — PICO RSS ISOLADO")
print("=" * 72)

print("\nRESULTADOS INDIVIDUAIS")
display(results.round(4))

print("\nRESUMO")
display(summary.round(4))

print("\nEquivalência exata em todos os subprocessos: True")
print("Arquivos salvos em:", OUTPUT_DIR)
print("\nCÉLULA 12 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 13
# VALIDAÇÃO MULTIDATASET — CLASSIFICAÇÃO BINÁRIA
# ============================================================

import time
import json
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

SEED = 42
N_REPEATS = 10

DATASETS = [
    "Pima_Diabetes",
    "Credit_Approval",
    "Synthetic_30",
    "Synthetic_60",
    "Synthetic_120"
]

COMPONENTS = [
    "Permutation_Importance",
    "Coefficient_Abs",
    "Correlation_Abs",
    "CFIS_Score",
    "CFIS_Percentage"
]

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

assert callable(optimized_permutation_adapter)
assert callable(original_permutation)

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Binary_Multidataset_V2"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

records = []
audit_records = []

try:
    for dataset_name in DATASETS:

        data = prepared_datasets[dataset_name]

        assert data["task"] == "classification"
        assert len(np.unique(data["y_train"])) == 2

        X_train = data["X_train"]
        y_train = data["y_train"]

        print("\n" + "=" * 68)
        print(f"DATASET: {dataset_name}")
        print("=" * 68)

        # --------------------------------------------
        # CFIS ORIGINAL
        # --------------------------------------------

        cfis_globals["permutation_importance"] = (
            original_permutation
        )

        start = time.perf_counter()

        original = cfis_function(
            X_train,
            y_train,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        original_time = time.perf_counter() - start

        # --------------------------------------------
        # CFIS OTIMIZADO
        # --------------------------------------------

        cfis_globals["permutation_importance"] = (
            optimized_permutation_adapter
        )

        start = time.perf_counter()

        optimized = cfis_function(
            X_train,
            y_train,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        optimized_time = time.perf_counter() - start

        # Restaurar antes de qualquer auditoria.
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

        # --------------------------------------------
        # AUDITORIA
        # --------------------------------------------

        same_order = (
            original["Feature"].tolist()
            == optimized["Feature"].tolist()
        )

        all_components_exact = True
        max_difference = 0.0

        for component in COMPONENTS:

            # Comparação por nome da característica,
            # independentemente da posição no ranking.
            a = original.set_index("Feature")[component]
            b = optimized.set_index("Feature")[component]

            b = b.reindex(a.index)

            exact = bool(
                np.array_equal(
                    a.to_numpy(),
                    b.to_numpy()
                )
            )

            difference = float(
                np.max(
                    np.abs(
                        a.to_numpy() - b.to_numpy()
                    )
                )
            )

            all_components_exact &= exact
            max_difference = max(
                max_difference, difference
            )

            audit_records.append({
                "Dataset": dataset_name,
                "Component": component,
                "Exact_Equal": exact,
                "Max_Abs_Difference": difference
            })

        exact_equivalence = bool(
            same_order and all_components_exact
        )

        speedup = original_time / optimized_time

        records.append({
            "Dataset": dataset_name,
            "Samples": len(X_train),
            "Features": X_train.shape[1],
            "Original_Time_s": original_time,
            "Optimized_Time_s": optimized_time,
            "Speedup_Exploratory": speedup,
            "Same_Feature_Order": same_order,
            "Exact_Equivalence": exact_equivalence,
            "Max_Abs_Difference": max_difference
        })

        print(f"Original:  {original_time:.4f} s")
        print(f"Otimizado: {optimized_time:.4f} s")
        print(f"Speedup exploratório: {speedup:.2f}x")
        print("Ordem idêntica:", same_order)
        print("Equivalência exata:", exact_equivalence)
        print("Diferença máxima:", max_difference)

        # Preservar ambos os rankings para auditoria.
        original.to_csv(
            OUTPUT_DIR /
            f"{dataset_name}_{run_id}_original.csv",
            index=False,
            float_format="%.17g"
        )

        optimized.to_csv(
            OUTPUT_DIR /
            f"{dataset_name}_{run_id}_optimized.csv",
            index=False,
            float_format="%.17g"
        )

finally:
    cfis_globals["permutation_importance"] = (
        original_permutation
    )

# ------------------------------------------------------------
# RESULTADOS CONSOLIDADOS
# ------------------------------------------------------------

summary = pd.DataFrame(records)
audit = pd.DataFrame(audit_records)

summary.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False
)

audit.to_csv(
    OUTPUT_DIR / f"{run_id}_component_audit.csv",
    index=False
)

metadata = {
    "datasets": DATASETS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "model": "Original CFIS LogisticRegression",
    "scope": "binary_classification",
    "timing": "single-run exploratory",
    "all_exact": bool(
        summary["Exact_Equivalence"].all()
    )
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\n" + "=" * 72)
print("CFIS V2 — VALIDAÇÃO MULTIDATASET")
print("=" * 72)

display(summary.round(6))

print("\nAUDITORIA POR COMPONENTE")
display(audit)

print(
    "\nEQUIVALÊNCIA EXATA EM TODOS OS DATASETS:",
    metadata["all_exact"]
)

print("\nResultados salvos em:", OUTPUT_DIR)

if not metadata["all_exact"]:
    print(
        "\nATENÇÃO: há divergências. "
        "Não generalizar a otimização antes de investigá-las."
    )

print("\nCÉLULA 13 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 14A
# CORREÇÃO DA LISTA + AUDITORIA MULTICLASSE V3
# ============================================================

import time
import json
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.inspection import permutation_importance

SEED = 42
N_REPEATS = 10

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Multiclass_V3_Detailed_Audit"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

# Identificar somente datasets multiclasse já preparados.
available = []

for name, data in prepared_datasets.items():
    if data["task"] != "classification":
        continue

    n_classes = len(np.unique(data["y_train"]))

    if n_classes > 2:
        available.append(name)

available = sorted(available)

print("Datasets multiclasse disponíveis:")
for name in available:
    print(
        f" - {name}: "
        f"{len(np.unique(prepared_datasets[name]['y_train']))} classes, "
        f"{prepared_datasets[name]['X_train'].shape[1]} features"
    )

assert available, "Nenhum dataset multiclasse disponível."

records = []
component_records = []

try:
    for name in available:
        data = prepared_datasets[name]

        X = data["X_train"]
        y = np.asarray(data["y_train"]).ravel()

        # Mesmo escalonamento e estimador do CFIS original.
        X_scaled = StandardScaler().fit_transform(
            X.to_numpy(dtype=float)
            if isinstance(X, pd.DataFrame)
            else np.asarray(X, dtype=float)
        )

        model = LogisticRegression(
            random_state=SEED,
            max_iter=1000
        )
        model.fit(X_scaled, y)

        print("\n" + "=" * 68)
        print("DATASET:", name)
        print("=" * 68)

        # 1. Matriz original sklearn.
        start = time.perf_counter()

        sklearn_result = permutation_importance(
            model,
            X_scaled,
            y,
            n_repeats=N_REPEATS,
            random_state=SEED,
            scoring="accuracy",
            n_jobs=1
        )

        original_perm_time = time.perf_counter() - start

        # 2. Matriz V3.
        start = time.perf_counter()

        optimized_values = (
            fast_multiclass_permutation_importance_v3(
                model,
                X_scaled,
                y,
                n_repeats=N_REPEATS,
                random_state=SEED
            )
        )

        optimized_perm_time = time.perf_counter() - start

        original_values = sklearn_result.importances

        exact_matrix = bool(
            np.array_equal(
                original_values,
                optimized_values
            )
        )

        max_matrix_difference = float(
            np.max(
                np.abs(
                    original_values - optimized_values
                )
            )
        )

        divergent_values = int(
            np.count_nonzero(
                original_values != optimized_values
            )
        )

        # 3. CFIS completo.
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

        original_cfis = cfis_function(
            X, y,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        try:
            cfis_globals["permutation_importance"] = (
                optimized_multiclass_adapter
            )

            optimized_cfis = cfis_function(
                X, y,
                task="classification",
                model=None,
                n_repeats=N_REPEATS,
                random_state=SEED
            )

        finally:
            cfis_globals["permutation_importance"] = (
                original_permutation
            )

        same_order = (
            original_cfis["Feature"].tolist()
            == optimized_cfis["Feature"].tolist()
        )

        columns = [
            "Permutation_Importance",
            "Coefficient_Abs",
            "Correlation_Abs",
            "CFIS_Score",
            "CFIS_Percentage"
        ]

        exact_components = []

        for column in columns:
            a = original_cfis.set_index("Feature")[column]
            b = optimized_cfis.set_index("Feature")[column]
            b = b.reindex(a.index)

            exact = bool(
                np.array_equal(
                    a.to_numpy(),
                    b.to_numpy()
                )
            )

            max_diff = float(
                np.max(
                    np.abs(
                        a.to_numpy() - b.to_numpy()
                    )
                )
            )

            exact_components.append(exact)

            component_records.append({
                "Dataset": name,
                "Component": column,
                "Exact_Equal": exact,
                "Max_Abs_Difference": max_diff
            })

        exact_cfis = bool(
            same_order and all(exact_components)
        )

        records.append({
            "Dataset": name,
            "Samples": len(y),
            "Features": X_scaled.shape[1],
            "Classes": len(np.unique(y)),
            "Permutation_Exact": exact_matrix,
            "Permutation_Max_Difference": max_matrix_difference,
            "Permutation_Divergent_Values": divergent_values,
            "CFIS_Exact": exact_cfis,
            "Same_Feature_Order": same_order,
            "Original_Permutation_s": original_perm_time,
            "Optimized_Permutation_s": optimized_perm_time,
            "Speedup_Exploratory": (
                original_perm_time / optimized_perm_time
            )
        })

        print("Matriz de permutação exata:", exact_matrix)
        print("Valores divergentes:", divergent_values)
        print("Diferença máxima:", max_matrix_difference)
        print("CFIS completo exato:", exact_cfis)
        print("Ordem das features idêntica:", same_order)
        print(
            "Speedup exploratório da permutação:",
            f"{original_perm_time / optimized_perm_time:.2f}x"
        )

finally:
    cfis_globals["permutation_importance"] = (
        original_permutation
    )

summary = pd.DataFrame(records)
audit = pd.DataFrame(component_records)

summary.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False
)

audit.to_csv(
    OUTPUT_DIR / f"{run_id}_components.csv",
    index=False
)

all_exact = bool(
    summary["Permutation_Exact"].all()
    and summary["CFIS_Exact"].all()
)

metadata = {
    "datasets": available,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "permutation_matrix_and_cfis_exact": all_exact,
    "timing": "single-run exploratory"
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\n" + "=" * 72)
print("CFIS V3 — AUDITORIA MULTICLASSE DETALHADA")
print("=" * 72)

display(summary.round(6))

print("\nAUDITORIA DOS COMPONENTES")
display(audit)

print("\nEQUIVALÊNCIA EXATA GLOBAL:", all_exact)
print("Arquivos salvos em:", OUTPUT_DIR)
print("\nCÉLULA 14A CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 15
# REGRESSÃO V4 — AUDITORIA INICIAL DA PERMUTAÇÃO
# ============================================================

import time
import json
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.inspection import permutation_importance
from sklearn.metrics import r2_score
from sklearn.utils import check_random_state

SEED = 42
N_REPEATS = 10

DATASETS = [
    name for name, data in prepared_datasets.items()
    if data["task"] == "regression"
]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Regression_V4_Prototype"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")


def fast_regression_permutation_v4(
    estimator, X, y,
    n_repeats=10,
    random_state=42
):
    X = np.asarray(X)
    y = np.asarray(y).ravel()

    n_samples, n_features = X.shape

    coef = np.asarray(estimator.coef_).ravel()

    if coef.shape != (n_features,):
        raise ValueError("Coeficientes incompatíveis.")

    baseline_predictions = estimator.predict(X)
    baseline_r2 = r2_score(y, baseline_predictions)

    internal_seed = check_random_state(
        random_state
    ).randint(np.iinfo(np.int32).max)

    importances = np.empty(
        (n_features, n_repeats),
        dtype=float
    )

    for j in range(n_features):
        rng = check_random_state(internal_seed)
        indices = np.arange(n_samples)

        original_column = X[:, j]
        current_column = original_column.copy()

        for r in range(n_repeats):
            rng.shuffle(indices)

            # Mesma permutação cumulativa do sklearn.
            current_column = current_column[
                indices
            ].copy()

            predictions = (
                baseline_predictions
                + (current_column - original_column)
                * coef[j]
            )

            permuted_r2 = r2_score(y, predictions)

            importances[j, r] = (
                baseline_r2 - permuted_r2
            )

    return importances


records = []

print("Datasets de regressão encontrados:", DATASETS)
assert DATASETS, "Nenhum dataset de regressão disponível."

for name in DATASETS:
    data = prepared_datasets[name]

    X = data["X_train"]
    y = np.asarray(data["y_train"]).ravel()

    X_values = (
        X.to_numpy(dtype=float)
        if isinstance(X, pd.DataFrame)
        else np.asarray(X, dtype=float)
    )

    X_scaled = StandardScaler().fit_transform(
        X_values
    )

    # Exatamente os parâmetros do CFIS original.
    model = Lasso(
        alpha=0.01,
        random_state=SEED,
        max_iter=1000
    )
    model.fit(X_scaled, y)

    print("\n" + "=" * 68)
    print("DATASET:", name)
    print("=" * 68)

    start = time.perf_counter()

    reference = permutation_importance(
        model,
        X_scaled,
        y,
        n_repeats=N_REPEATS,
        random_state=SEED,
        scoring="r2",
        n_jobs=1
    )

    original_time = time.perf_counter() - start

    start = time.perf_counter()

    optimized = fast_regression_permutation_v4(
        model,
        X_scaled,
        y,
        n_repeats=N_REPEATS,
        random_state=SEED
    )

    optimized_time = time.perf_counter() - start

    original_values = reference.importances

    exact = bool(
        np.array_equal(original_values, optimized)
    )

    max_difference = float(
        np.max(np.abs(original_values - optimized))
    )

    divergent_values = int(
        np.count_nonzero(
            original_values != optimized
        )
    )

    equivalent_1e12 = bool(
        np.allclose(
            original_values,
            optimized,
            rtol=0,
            atol=1e-12
        )
    )

    means_exact = bool(
        np.array_equal(
            reference.importances_mean,
            optimized.mean(axis=1)
        )
    )

    records.append({
        "Dataset": name,
        "Samples": len(y),
        "Features": X_scaled.shape[1],
        "Original_Time_s": original_time,
        "Optimized_Time_s": optimized_time,
        "Speedup_Exploratory": (
            original_time / optimized_time
        ),
        "Matrix_Exact": exact,
        "Equivalent_1e12": equivalent_1e12,
        "Divergent_Values": divergent_values,
        "Max_Abs_Difference": max_difference,
        "Means_Exact": means_exact
    })

    print(f"Original: {original_time:.6f} s")
    print(f"Otimizado: {optimized_time:.6f} s")
    print(
        "Speedup exploratório:",
        f"{original_time / optimized_time:.2f}x"
    )
    print("Matriz exatamente igual:", exact)
    print("Equivalência com tolerância 1e-12:",
          equivalent_1e12)
    print("Valores divergentes:", divergent_values)
    print("Diferença máxima:", max_difference)
    print("Médias exatamente iguais:", means_exact)

    np.savez_compressed(
        OUTPUT_DIR / f"{name}_{run_id}_matrices.npz",
        original=original_values,
        optimized=optimized
    )

summary = pd.DataFrame(records)

summary.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False
)

metadata = {
    "datasets": DATASETS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "model": "Lasso(alpha=0.01, max_iter=1000)",
    "scoring": "r2",
    "scope": "permutation importance only",
    "timing": "single-run exploratory",
    "all_exact": bool(summary["Matrix_Exact"].all()),
    "all_equivalent_1e12": bool(
        summary["Equivalent_1e12"].all()
    )
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2)

print("\n" + "=" * 72)
print("CFIS V4 — AUDITORIA INICIAL DE REGRESSÃO")
print("=" * 72)

display(summary.round(12))

print(
    "\nMATRIZES EXATAS EM TODOS:",
    metadata["all_exact"]
)
print(
    "EQUIVALENTES A 1e-12 EM TODOS:",
    metadata["all_equivalent_1e12"]
)

print("\nResultados salvos em:", OUTPUT_DIR)
print("\nCÉLULA 15 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 16
# DIAGNÓSTICO DE PRECISÃO NUMÉRICA — REGRESSÃO V4
# ============================================================

import json
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score
from sklearn.inspection import permutation_importance
from sklearn.utils import check_random_state

SEED = 42
N_REPEATS = 10

DATASETS = ["California_Housing", "Diabetes_Reg"]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Regression_V4_Numerical_Diagnosis"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

records = []
examples = []

for name in DATASETS:

    data = prepared_datasets[name]

    X = data["X_train"]
    y = np.asarray(data["y_train"]).ravel()

    X_values = (
        X.to_numpy(dtype=float)
        if isinstance(X, pd.DataFrame)
        else np.asarray(X, dtype=float)
    )

    X_scaled = StandardScaler().fit_transform(X_values)

    model = Lasso(
        alpha=0.01,
        random_state=SEED,
        max_iter=1000
    )
    model.fit(X_scaled, y)

    baseline_pred = model.predict(X_scaled)
    baseline_r2 = r2_score(y, baseline_pred)

    n_samples, n_features = X_scaled.shape

    # Reproduzir a geração da semente interna do sklearn.
    internal_seed = check_random_state(
        SEED
    ).randint(np.iinfo(np.int32).max)

    max_pred_delta = 0.0
    max_pred_replacement = 0.0
    max_r2_delta = 0.0
    max_r2_replacement = 0.0

    divergent_pred_delta = 0
    divergent_pred_replacement = 0

    divergent_r2_delta = 0
    divergent_r2_replacement = 0

    total_predictions = 0
    total_permutations = 0

    # Referência oficial das importâncias.
    reference = permutation_importance(
        model,
        X_scaled,
        y,
        n_repeats=N_REPEATS,
        random_state=SEED,
        scoring="r2",
        n_jobs=1
    ).importances

    reconstructed = np.empty_like(reference)
    incremental = np.empty_like(reference)
    replacement = np.empty_like(reference)

    for j in range(n_features):

        rng = check_random_state(internal_seed)
        indices = np.arange(n_samples)

        current_column = X_scaled[:, j].copy()
        original_column = X_scaled[:, j]

        for r in range(N_REPEATS):

            rng.shuffle(indices)
            current_column = current_column[indices].copy()

            # A: referência — recálculo completo.
            X_permuted = X_scaled.copy()
            X_permuted[:, j] = current_column

            pred_full = model.predict(X_permuted)

            # B: atualização incremental V4.
            pred_delta = (
                baseline_pred
                + (current_column - original_column)
                * model.coef_[j]
            )

            # C: substituição da contribuição original.
            pred_replacement = (
                baseline_pred
                - original_column * model.coef_[j]
                + current_column * model.coef_[j]
            )

            # Diferenças nas previsões.
            diff_delta = np.abs(pred_full - pred_delta)
            diff_replacement = np.abs(
                pred_full - pred_replacement
            )

            max_pred_delta = max(
                max_pred_delta,
                float(diff_delta.max())
            )

            max_pred_replacement = max(
                max_pred_replacement,
                float(diff_replacement.max())
            )

            divergent_pred_delta += int(
                np.count_nonzero(pred_full != pred_delta)
            )

            divergent_pred_replacement += int(
                np.count_nonzero(
                    pred_full != pred_replacement
                )
            )

            total_predictions += n_samples
            total_permutations += 1

            # Comparar R² usando a mesma função.
            r2_full = r2_score(y, pred_full)
            r2_delta = r2_score(y, pred_delta)
            r2_replacement = r2_score(
                y, pred_replacement
            )

            max_r2_delta = max(
                max_r2_delta,
                abs(r2_full - r2_delta)
            )

            max_r2_replacement = max(
                max_r2_replacement,
                abs(r2_full - r2_replacement)
            )

            divergent_r2_delta += int(
                r2_full != r2_delta
            )

            divergent_r2_replacement += int(
                r2_full != r2_replacement
            )

            reconstructed[j, r] = (
                baseline_r2 - r2_full
            )

            incremental[j, r] = (
                baseline_r2 - r2_delta
            )

            replacement[j, r] = (
                baseline_r2 - r2_replacement
            )

            if r2_full != r2_delta and len(examples) < 12:
                examples.append({
                    "Dataset": name,
                    "Feature_Index": j,
                    "Repeat": r,
                    "R2_Full": r2_full,
                    "R2_Incremental": r2_delta,
                    "Difference": r2_full - r2_delta
                })

    reference_reconstructed_exact = bool(
        np.array_equal(reference, reconstructed)
    )

    records.append({
        "Dataset": name,
        "Total_Permutations": total_permutations,
        "Total_Predictions": total_predictions,
        "Reference_Reconstructed_Exact":
            reference_reconstructed_exact,
        "Max_Pred_Diff_Incremental": max_pred_delta,
        "Max_Pred_Diff_Replacement": max_pred_replacement,
        "Divergent_Pred_Incremental": divergent_pred_delta,
        "Divergent_Pred_Replacement":
            divergent_pred_replacement,
        "Max_R2_Diff_Incremental": max_r2_delta,
        "Max_R2_Diff_Replacement": max_r2_replacement,
        "Divergent_R2_Incremental": divergent_r2_delta,
        "Divergent_R2_Replacement":
            divergent_r2_replacement,
        "Incremental_Matrix_Exact": bool(
            np.array_equal(reference, incremental)
        ),
        "Replacement_Matrix_Exact": bool(
            np.array_equal(reference, replacement)
        ),
        "Incremental_Max_Importance_Diff": float(
            np.max(np.abs(reference - incremental))
        ),
        "Replacement_Max_Importance_Diff": float(
            np.max(np.abs(reference - replacement))
        )
    })

    print(f"\n{name}")
    print(
        "Referência reconstruída exatamente:",
        reference_reconstructed_exact
    )
    print(
        "Matriz incremental exata:",
        np.array_equal(reference, incremental)
    )
    print(
        "Matriz por substituição exata:",
        np.array_equal(reference, replacement)
    )

summary = pd.DataFrame(records)
examples_df = pd.DataFrame(examples)

summary.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False,
    float_format="%.17g"
)

examples_df.to_csv(
    OUTPUT_DIR / f"{run_id}_examples.csv",
    index=False,
    float_format="%.17g"
)

metadata = {
    "datasets": DATASETS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "model": "Lasso(alpha=0.01, max_iter=1000)",
    "purpose": "diagnose floating-point discrepancies",
    "original_cfis_modified": False
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2)

print("\n" + "=" * 72)
print("CFIS V4 — DIAGNÓSTICO NUMÉRICO")
print("=" * 72)

display(summary.T)

print("\nEXEMPLOS DE DIVERGÊNCIA NO R²")
display(examples_df)

print("\nResultados salvos em:", OUTPUT_DIR)
print("\nCÉLULA 16 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 17
# REGRESSÃO V4 — AUDITORIA DO CFIS COMPLETO
# ============================================================

import time
import json
from pathlib import Path
from datetime import datetime, timezone
from types import SimpleNamespace

import numpy as np
import pandas as pd

SEED = 42
N_REPEATS = 10
ATOL = 1e-12

DATASETS = [
    "California_Housing",
    "Diabetes_Reg"
]

COMPONENTS = [
    "Permutation_Importance",
    "Coefficient_Abs",
    "Correlation_Abs",
    "CFIS_Score",
    "CFIS_Percentage"
]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "05_Validation"
    / "Regression_V4_Full_CFIS"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

# ------------------------------------------------------------
# 1. ADAPTADOR COMPATÍVEL COM O CFIS ORIGINAL
# ------------------------------------------------------------

def optimized_regression_adapter_v4(
    estimator,
    X,
    y,
    n_repeats=10,
    random_state=42,
    scoring=None,
    n_jobs=None,
    **kwargs
):
    # Impedir uso fora do escopo validado.
    if scoring != "r2":
        raise ValueError("V4 exige scoring='r2'.")

    if kwargs:
        raise ValueError(
            f"Parâmetros não suportados: {list(kwargs)}"
        )

    if n_jobs not in (None, 1):
        raise ValueError("V4 exige n_jobs=1 ou None.")

    values = fast_regression_permutation_v4(
        estimator,
        X,
        y,
        n_repeats=n_repeats,
        random_state=random_state
    )

    return SimpleNamespace(
        importances=values,
        importances_mean=values.mean(axis=1),
        importances_std=values.std(axis=1)
    )


# ------------------------------------------------------------
# 2. AUDITORIA DOS DATASETS
# ------------------------------------------------------------

summary_records = []
component_records = []
topk_records = []

try:
    for name in DATASETS:

        data = prepared_datasets[name]

        assert data["task"] == "regression"

        X = data["X_train"]
        y = data["y_train"]

        print("\n" + "=" * 68)
        print("DATASET:", name)
        print("=" * 68)

        # ----------------------------------------
        # CFIS ORIGINAL
        # ----------------------------------------

        cfis_globals["permutation_importance"] = (
            original_permutation
        )

        start = time.perf_counter()

        original = cfis_function(
            X,
            y,
            task="regression",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        original_time = time.perf_counter() - start

        # ----------------------------------------
        # CFIS COM V4
        # ----------------------------------------

        try:
            cfis_globals["permutation_importance"] = (
                optimized_regression_adapter_v4
            )

            start = time.perf_counter()

            optimized = cfis_function(
                X,
                y,
                task="regression",
                model=None,
                n_repeats=N_REPEATS,
                random_state=SEED
            )

            optimized_time = time.perf_counter() - start

        finally:
            cfis_globals["permutation_importance"] = (
                original_permutation
            )

        # ----------------------------------------
        # AUDITORIA DOS COMPONENTES
        # ----------------------------------------

        original_indexed = original.set_index("Feature")
        optimized_indexed = optimized.set_index("Feature")

        same_feature_set = (
            set(original_indexed.index)
            == set(optimized_indexed.index)
        )

        assert same_feature_set, (
            f"Conjuntos de features diferentes em {name}."
        )

        optimized_aligned = optimized_indexed.reindex(
            original_indexed.index
        )

        same_order = bool(
            original["Feature"].tolist()
            == optimized["Feature"].tolist()
        )

        component_exact = []
        component_close = []

        for component in COMPONENTS:

            a = original_indexed[component].to_numpy(
                dtype=float
            )
            b = optimized_aligned[component].to_numpy(
                dtype=float
            )

            exact = bool(np.array_equal(a, b))

            close = bool(
                np.allclose(
                    a,
                    b,
                    rtol=0,
                    atol=ATOL,
                    equal_nan=False
                )
            )

            max_diff = float(
                np.max(np.abs(a - b))
            )

            divergent_count = int(
                np.count_nonzero(a != b)
            )

            component_exact.append(exact)
            component_close.append(close)

            component_records.append({
                "Dataset": name,
                "Component": component,
                "Exact": exact,
                "Equivalent_1e12": close,
                "Divergent_Values": divergent_count,
                "Max_Abs_Difference": max_diff
            })

        # ----------------------------------------
        # AUDITORIA TOP-K
        # ----------------------------------------

        n_features = len(original)
        k_values = sorted(set([
            max(1, int(np.ceil(n_features * 0.10))),
            max(1, int(np.ceil(n_features * 0.25))),
            max(1, int(np.ceil(n_features * 0.50)))
        ]))

        for k in k_values:

            original_topk = set(
                original["Feature"].iloc[:k]
            )
            optimized_topk = set(
                optimized["Feature"].iloc[:k]
            )

            intersection = len(
                original_topk & optimized_topk
            )
            union = len(
                original_topk | optimized_topk
            )

            topk_records.append({
                "Dataset": name,
                "K": k,
                "Identical_TopK": (
                    original_topk == optimized_topk
                ),
                "Jaccard": intersection / union
            })

        all_exact = bool(
            same_order and all(component_exact)
        )

        all_close = bool(
            all(component_close)
        )

        summary_records.append({
            "Dataset": name,
            "Samples": len(X),
            "Features": n_features,
            "Original_Time_s": original_time,
            "Optimized_Time_s": optimized_time,
            "Speedup_Exploratory": (
                original_time / optimized_time
            ),
            "Same_Feature_Order": same_order,
            "All_Components_Exact": all(component_exact),
            "All_Components_Equivalent_1e12": all_close,
            "Full_Exact_Equivalence": all_exact
        })

        # ----------------------------------------
        # SALVAR RANKINGS COMPLETOS
        # ----------------------------------------

        original.to_csv(
            OUTPUT_DIR /
            f"{name}_{run_id}_original.csv",
            index=False,
            float_format="%.17g"
        )

        optimized.to_csv(
            OUTPUT_DIR /
            f"{name}_{run_id}_optimized_v4.csv",
            index=False,
            float_format="%.17g"
        )

        print(f"Original:  {original_time:.6f} s")
        print(f"V4:        {optimized_time:.6f} s")
        print(
            "Speedup exploratório:",
            f"{original_time / optimized_time:.2f}x"
        )
        print("Mesma ordem:", same_order)
        print("Todos os componentes exatos:",
              all(component_exact))
        print("Todos equivalentes a 1e-12:", all_close)

finally:
    # Garantir que o CFIS original permaneça intacto.
    cfis_globals["permutation_importance"] = (
        original_permutation
    )


# ------------------------------------------------------------
# 3. CONSOLIDAÇÃO E PERSISTÊNCIA
# ------------------------------------------------------------

summary = pd.DataFrame(summary_records)
components = pd.DataFrame(component_records)
topk = pd.DataFrame(topk_records)

summary.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False,
    float_format="%.17g"
)

components.to_csv(
    OUTPUT_DIR / f"{run_id}_components.csv",
    index=False,
    float_format="%.17g"
)

topk.to_csv(
    OUTPUT_DIR / f"{run_id}_topk.csv",
    index=False,
    float_format="%.17g"
)

metadata = {
    "datasets": DATASETS,
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "atol": ATOL,
    "rtol": 0,
    "method": "CFIS original with temporary V4 adapter",
    "timing": "single-run exploratory",
    "all_component_values_close": bool(
        summary["All_Components_Equivalent_1e12"].all()
    ),
    "all_feature_orders_identical": bool(
        summary["Same_Feature_Order"].all()
    ),
    "all_topk_identical": bool(
        topk["Identical_TopK"].all()
    )
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=2,
        ensure_ascii=False
    )

print("\n" + "=" * 72)
print("CFIS V4 — AUDITORIA COMPLETA DE REGRESSÃO")
print("=" * 72)

print("\nRESUMO")
display(summary.round(6))

print("\nCOMPONENTES — diferenças em precisão completa")
display(components)

print("\nCONCORDÂNCIA TOP-K")
display(topk)

print(
    "\nTODOS OS COMPONENTES EQUIVALENTES A 1e-12:",
    metadata["all_component_values_close"]
)

print(
    "TODAS AS ORDENS IDÊNTICAS:",
    metadata["all_feature_orders_identical"]
)

print(
    "TODOS OS TOP-K IDÊNTICOS:",
    metadata["all_topk_identical"]
)

print("\nArquivos salvos em:", OUTPUT_DIR)
print("\nCÉLULA 17 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS OPTIMIZATION — CÉLULA 18
# BENCHMARK CONSOLIDADO V2 / V3 / V4
# 3 PARES, SEED=42, N_REPEATS=10
# ============================================================

import time
import json
import traceback
from pathlib import Path
from datetime import datetime, timezone
from types import SimpleNamespace

import numpy as np
import pandas as pd

SEED = 42
N_REPEATS = 10
N_PAIRS = 3
ATOL = 1e-12

COMPONENTS = [
    "Permutation_Importance",
    "Coefficient_Abs",
    "Correlation_Abs",
    "CFIS_Score",
    "CFIS_Percentage"
]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Optimization"
    / "04_Benchmarks"
    / "Consolidated_V2_V3_V4"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

run_id = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

cfis_globals = cfis_function.__globals__
original_permutation = cfis_globals["permutation_importance"]

# ------------------------------------------------------------
# ADAPTADOR CONSOLIDADO
# ------------------------------------------------------------

def consolidated_permutation_adapter(
    estimator, X, y,
    n_repeats=10,
    random_state=42,
    scoring=None,
    n_jobs=None,
    **kwargs
):
    if kwargs or n_jobs not in (None, 1):
        raise ValueError("Parâmetros fora do escopo validado.")

    X_array = np.asarray(X)

    if scoring == "accuracy":
        if not hasattr(estimator, "classes_"):
            raise ValueError("Estimador sem classes_.")

        n_classes = len(estimator.classes_)

        if n_classes == 2:
            values = fast_binary_permutation_importance_v2(
                estimator, X_array, y,
                n_repeats=n_repeats,
                random_state=random_state
            )
        elif n_classes > 2:
            values = fast_multiclass_permutation_importance_v3(
                estimator, X_array, y,
                n_repeats=n_repeats,
                random_state=random_state
            )
        else:
            raise ValueError("Número de classes inválido.")

    elif scoring == "r2":
        values = fast_regression_permutation_v4(
            estimator, X_array, y,
            n_repeats=n_repeats,
            random_state=random_state
        )

    else:
        raise ValueError(f"Scoring não suportado: {scoring}")

    return SimpleNamespace(
        importances=values,
        importances_mean=values.mean(axis=1),
        importances_std=values.std(axis=1)
    )


def execute_cfis(data, variant):
    if variant == "Original":
        adapter = original_permutation
    else:
        adapter = consolidated_permutation_adapter

    try:
        cfis_globals["permutation_importance"] = adapter

        wall_start = time.perf_counter()
        cpu_start = time.process_time()

        result = cfis_function(
            data["X_train"],
            data["y_train"],
            task=data["task"],
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

        cpu_elapsed = time.process_time() - cpu_start
        wall_elapsed = time.perf_counter() - wall_start

    finally:
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

    return result, wall_elapsed, cpu_elapsed


def audit_cfis(original, optimized, task):
    same_order = (
        original["Feature"].tolist()
        == optimized["Feature"].tolist()
    )

    a = original.set_index("Feature")
    b = optimized.set_index("Feature")

    if set(a.index) != set(b.index):
        return False, False, False, float("inf")

    b = b.reindex(a.index)

    exact_components = []
    close_components = []
    max_diff = 0.0

    for column in COMPONENTS:
        av = a[column].to_numpy(dtype=float)
        bv = b[column].to_numpy(dtype=float)

        exact_components.append(
            bool(np.array_equal(av, bv))
        )
        close_components.append(
            bool(np.allclose(
                av, bv, atol=ATOL, rtol=0
            ))
        )
        max_diff = max(
            max_diff,
            float(np.max(np.abs(av - bv)))
        )

    exact = same_order and all(exact_components)
    close = same_order and all(close_components)

    accepted = exact if task == "classification" else close

    return accepted, exact, same_order, max_diff


# ------------------------------------------------------------
# EXECUÇÃO COM SALVAMENTO INCREMENTAL
# ------------------------------------------------------------

measurements = []
audits = []
errors = []

dataset_names = sorted(prepared_datasets.keys())

print("Datasets:", len(dataset_names))
print("Pares por dataset:", N_PAIRS)

try:
    for name in dataset_names:
        data = prepared_datasets[name]

        task = data["task"]
        n_classes = (
            len(np.unique(data["y_train"]))
            if task == "classification" else 0
        )

        version = (
            "V4" if task == "regression"
            else "V2" if n_classes == 2
            else "V3"
        )

        print("\n" + "=" * 72)
        print(f"{name} | {task} | {version}")
        print("=" * 72)

        dataset_accepted = True

        for pair in range(1, N_PAIRS + 1):
            # Alternância da ordem em cada par.
            order = (
                ["Original", "Optimized"]
                if pair % 2 == 1
                else ["Optimized", "Original"]
            )

            pair_results = {}

            try:
                for variant in order:
                    result, wall_s, cpu_s = execute_cfis(
                        data, variant
                    )

                    pair_results[variant] = result

                    measurements.append({
                        "Dataset": name,
                        "Task": task,
                        "Version": version,
                        "Pair": pair,
                        "Variant": variant,
                        "Execution_Order": order.index(variant) + 1,
                        "Wall_Time_s": wall_s,
                        "CPU_Time_s": cpu_s
                    })

                accepted, exact, same_order, max_diff = (
                    audit_cfis(
                        pair_results["Original"],
                        pair_results["Optimized"],
                        task
                    )
                )

                audits.append({
                    "Dataset": name,
                    "Version": version,
                    "Pair": pair,
                    "Accepted": accepted,
                    "Exact": exact,
                    "Same_Order": same_order,
                    "Max_Abs_Difference": max_diff
                })

                print(
                    f"Par {pair}/{N_PAIRS} | "
                    f"Aceito: {accepted} | "
                    f"Exato: {exact} | "
                    f"MaxDiff: {max_diff:.3e}"
                )

                if not accepted:
                    dataset_accepted = False
                    print(
                        "DIVERGÊNCIA: interrompendo "
                        "este dataset."
                    )
                    break

            except Exception as exc:
                dataset_accepted = False
                errors.append({
                    "Dataset": name,
                    "Pair": pair,
                    "Error": repr(exc),
                    "Traceback": traceback.format_exc()
                })
                print("ERRO:", repr(exc))
                break

            finally:
                # Persistir resultados já obtidos.
                pd.DataFrame(measurements).to_csv(
                    OUTPUT_DIR / f"{run_id}_measurements.csv",
                    index=False
                )
                pd.DataFrame(audits).to_csv(
                    OUTPUT_DIR / f"{run_id}_audits.csv",
                    index=False
                )
                if errors:
                    pd.DataFrame(errors).to_csv(
                        OUTPUT_DIR / f"{run_id}_errors.csv",
                        index=False
                    )

        print(
            "Status:",
            "APROVADO" if dataset_accepted
            else "REQUER INVESTIGAÇÃO"
        )

finally:
    cfis_globals["permutation_importance"] = (
        original_permutation
    )

# ------------------------------------------------------------
# CONSOLIDAÇÃO
# ------------------------------------------------------------

measure_df = pd.DataFrame(measurements)
audit_df = pd.DataFrame(audits)

summary_rows = []

for name in dataset_names:
    subset = measure_df[
        measure_df["Dataset"] == name
    ]

    checked = audit_df[
        audit_df["Dataset"] == name
    ]

    complete = (
        len(checked) == N_PAIRS
        and bool(checked["Accepted"].all())
        and len(subset) == 2 * N_PAIRS
    )

    if not complete:
        summary_rows.append({
            "Dataset": name,
            "Status": "REQUIRES_INVESTIGATION"
        })
        continue

    original_rows = subset[
        subset["Variant"] == "Original"
    ]
    optimized_rows = subset[
        subset["Variant"] == "Optimized"
    ]

    paired = subset.pivot(
        index="Pair",
        columns="Variant",
        values=["Wall_Time_s", "CPU_Time_s"]
    )

    wall_ratios = (
        paired[("Wall_Time_s", "Original")]
        / paired[("Wall_Time_s", "Optimized")]
    )

    cpu_ratios = (
        paired[("CPU_Time_s", "Original")]
        / paired[("CPU_Time_s", "Optimized")]
    )

    summary_rows.append({
        "Dataset": name,
        "Version": checked["Version"].iloc[0],
        "Status": "VALIDATED",
        "Original_Wall_Median_s":
            original_rows["Wall_Time_s"].median(),
        "Optimized_Wall_Median_s":
            optimized_rows["Wall_Time_s"].median(),
        "Paired_Wall_Speedup_Median":
            wall_ratios.median(),
        "Original_CPU_Median_s":
            original_rows["CPU_Time_s"].median(),
        "Optimized_CPU_Median_s":
            optimized_rows["CPU_Time_s"].median(),
        "Paired_CPU_Speedup_Median":
            cpu_ratios.median(),
        "All_Pairs_Exact":
            bool(checked["Exact"].all()),
        "Max_Abs_Difference":
            checked["Max_Abs_Difference"].max()
    })

summary_df = pd.DataFrame(summary_rows)

summary_df.to_csv(
    OUTPUT_DIR / f"{run_id}_summary.csv",
    index=False,
    float_format="%.17g"
)

metadata = {
    "seed": SEED,
    "n_repeats": N_REPEATS,
    "n_pairs": N_PAIRS,
    "regression_atol": ATOL,
    "classification_requires_exact": True,
    "regression_requires_order_and_tolerance": True,
    "timing": "paired exploratory benchmark",
    "datasets_requested": dataset_names,
    "original_function_restored": (
        cfis_globals["permutation_importance"]
        is original_permutation
    )
}

with open(
    OUTPUT_DIR / f"{run_id}_metadata.json",
    "w", encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=2, ensure_ascii=False)

print("\n" + "=" * 72)
print("CFIS — BENCHMARK CONSOLIDADO")
print("=" * 72)

display(summary_df.round(6))

print("\nTOTAL VALIDADO:",
      int((summary_df["Status"] == "VALIDATED").sum()),
      "/", len(dataset_names))

print("ERROS:", len(errors))
print("Arquivos salvos em:", OUTPUT_DIR)
print("\nCÉLULA 18 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 19
# CONGELAMENTO DO PROTOCOLO EXPERIMENTAL
# ============================================================

import json
import hashlib
import platform
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import sklearn

SEED = 42
N_REPEATS_CFIS = 10
N_SPLITS_STABILITY = 10

ABSOLUTE_K = [3, 5, 10, 20]
RETENTION_RATES = [0.10, 0.25, 0.50, 0.75]

METHODS = [
    "CFIS",
    "Mutual_Information",
    "ANOVA_or_F_Regression",
    "mRMR",
    "L1",
    "CFS",
    "RFE",
    "PFI",
    "TreeSHAP",
    "MDI",
    "Random_TopK"
]

PREDICTIVE_MODELS = [
    "DecisionTree",
    "RandomForest"
]

METRICS = {
    "classification": [
        "Accuracy",
        "Balanced_Accuracy",
        "Macro_F1"
    ],
    "regression": [
        "R2",
        "RMSE",
        "MAE"
    ],
    "ranking": [
        "Jaccard_at_K",
        "Spearman",
        "Kendall"
    ],
    "stability": [
        "Pairwise_Jaccard",
        "Nogueira_Stability"
    ],
    "resources": [
        "Selection_Wall_Time_s",
        "Training_Wall_Time_s",
        "Total_Wall_Time_s",
        "CPU_Time_s",
        "Peak_RSS_MiB",
        "Peak_Python_Allocations_MiB"
    ]
}

# A avaliação energética só será ativada mediante
# identificação de uma fonte real e confiável.
ENERGY_MEASUREMENT = "NOT_YET_AVAILABLE"

# Este protocolo é registrado antes da comparação final.
# Métodos que não tenham implementação compatível serão
# sinalizados, nunca substituídos silenciosamente.

def fingerprint_array(value):
    arr = np.asarray(value)

    if arr.dtype.kind in "biufc":
        canonical = np.ascontiguousarray(
            arr.astype("<f8", copy=False)
        )
        payload = canonical.tobytes()
    else:
        payload = json.dumps(
            arr.tolist(),
            ensure_ascii=False,
            default=str
        ).encode("utf-8")

    h = hashlib.sha256()
    h.update(str(arr.shape).encode())
    h.update(str(arr.dtype).encode())
    h.update(payload)
    return h.hexdigest()


dataset_records = []

for name in sorted(prepared_datasets):
    data = prepared_datasets[name]

    X_train = data["X_train"]
    X_test = data["X_test"]
    y_train = data["y_train"]
    y_test = data["y_test"]

    task = data["task"]
    n_features = X_train.shape[1]

    assert X_test.shape[1] == n_features
    assert len(X_train) == len(y_train)
    assert len(X_test) == len(y_test)
    assert task in ("classification", "regression")

    k_absolute = [
        k for k in ABSOLUTE_K if k <= n_features
    ]

    k_proportional = [
        max(1, int(np.ceil(rate * n_features)))
        for rate in RETENTION_RATES
    ]

    k_all = sorted(
        set(k_absolute + k_proportional)
    )

    classes = (
        np.unique(np.asarray(y_train)).tolist()
        if task == "classification"
        else None
    )

    dataset_records.append({
        "Dataset": name,
        "Task": task,
        "Train_Samples": len(X_train),
        "Test_Samples": len(X_test),
        "Features": n_features,
        "Classes": (
            len(classes) if classes is not None else None
        ),
        "Absolute_K": k_absolute,
        "Proportional_K": k_proportional,
        "All_K": k_all,
        "X_Train_SHA256": fingerprint_array(X_train),
        "X_Test_SHA256": fingerprint_array(X_test),
        "Y_Train_SHA256": fingerprint_array(y_train),
        "Y_Test_SHA256": fingerprint_array(y_test)
    })

manifest_df = pd.DataFrame(dataset_records)

assert len(manifest_df) == 12, (
    "A quantidade de datasets mudou. "
    "Investigue antes de congelar o protocolo."
)

assert manifest_df["Dataset"].is_unique

# ------------------------------------------------------------
# DOCUMENTAÇÃO DO DESENHO EXPERIMENTAL
# ------------------------------------------------------------

protocol = {
    "protocol_name": "CFIS_Final_Scientific_Benchmark",
    "status": "FROZEN_DESIGN_PENDING_IMPLEMENTATION_AUDIT",
    "random_seed": SEED,
    "cfis_n_repeats": N_REPEATS_CFIS,
    "stability_splits": N_SPLITS_STABILITY,
    "absolute_k": ABSOLUTE_K,
    "retention_rates": RETENTION_RATES,
    "methods": METHODS,
    "predictive_models": PREDICTIVE_MODELS,
    "metrics": METRICS,
    "energy_measurement": ENERGY_MEASUREMENT,
    "rules": {
        "selection_fit": "training_data_only",
        "test_data": "evaluation_only",
        "preprocessing": "fit_on_training_only",
        "same_splits_for_all_methods": True,
        "same_predictive_models_for_all_methods": True,
        "cfis_formula_changes": False,
        "cfis_optimized_binary": "V2_exact",
        "cfis_optimized_multiclass": "V3_exact",
        "cfis_optimized_regression": "V4_atol_1e-12",
        "feature_count": "exactly_K",
        "topk_ties": "deterministic_feature_order",
        "repeated_evaluation": (
            "paired splits; no independent-fold assumption"
        ),
        "statistical_comparison": (
            "dataset-level Friedman and paired "
            "posthoc comparisons with Holm correction"
        ),
        "reference_importance_is_not_ground_truth": True,
        "random_topk": (
            "repeated seeded random selections"
        ),
        "full_feature_baseline": True,
        "resource_measurements": (
            "selection and downstream training "
            "measured separately"
        ),
        "energy_claims_without_sensor": False
    },
    "implementation_decisions_pending": [
        "Define exact selector implementations and versions",
        "Define model hyperparameters without outcome tuning",
        "Define nested or repeated evaluation splits",
        "Define feature preprocessing for each selector",
        "Define TreeSHAP multiclass aggregation",
        "Define CFS implementation and tie handling",
        "Define RSS measurement isolation",
        "Define Random Top-K repetition protocol"
    ],
    "environment": {
        "python": platform.python_version(),
        "numpy": np.__version__,
        "pandas": pd.__version__,
        "scikit_learn": sklearn.__version__
    },
    "datasets": dataset_records
}

output_dir = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "00_Protocol"
)
output_dir.mkdir(parents=True, exist_ok=True)

timestamp = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

manifest_path = (
    output_dir / f"{timestamp}_dataset_manifest.csv"
)
protocol_path = (
    output_dir / f"{timestamp}_protocol.json"
)

manifest_df.to_csv(
    manifest_path,
    index=False
)

with open(
    protocol_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        protocol,
        f,
        ensure_ascii=False,
        indent=2,
        default=str
    )

with open(protocol_path, "rb") as f:
    protocol_sha256 = hashlib.sha256(
        f.read()
    ).hexdigest()

print("=" * 72)
print("CFIS — PROTOCOLO EXPERIMENTAL FINAL")
print("=" * 72)

print("\nDATASETS E VALORES DE K")
display(
    manifest_df[
        [
            "Dataset",
            "Task",
            "Train_Samples",
            "Test_Samples",
            "Features",
            "Classes",
            "All_K"
        ]
    ]
)

print("\nMÉTODOS:", len(METHODS))
print(", ".join(METHODS))

print("\nMODELOS:", ", ".join(PREDICTIVE_MODELS))

print("\nSEED:", SEED)
print("CFIS N_REPEATS:", N_REPEATS_CFIS)
print("DIVISÕES DE ESTABILIDADE:", N_SPLITS_STABILITY)

print("\nSTATUS:", protocol["status"])
print("DECISÕES PENDENTES:",
      len(protocol["implementation_decisions_pending"]))

print("\nPROTOCOLO SHA256:", protocol_sha256)
print("Manifesto:", manifest_path)
print("Protocolo:", protocol_path)

print("\nCÉLULA 19 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 20
# AUDITORIA DE INTEGRIDADE E VIABILIDADE
# ============================================================

import json
import hashlib
import platform
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import sklearn
from scipy import sparse

SEED = 42
STABILITY_SPLITS = 10
N_METHODS = 11
N_MODELS = 2

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(
    timezone.utc
).strftime("%Y%m%dT%H%M%S%fZ")

records = []
class_records = []
issues = []

def as_numpy(value):
    if sparse.issparse(value):
        return value.toarray()
    if hasattr(value, "to_numpy"):
        return value.to_numpy()
    return np.asarray(value)

def numeric_matrix_info(value):
    arr = as_numpy(value)

    result = {
        "Shape": list(arr.shape),
        "Dtype": str(arr.dtype),
        "All_Numeric": False,
        "NaN_Count": None,
        "Inf_Count": None,
        "Constant_Features": None,
        "Zero_Variance_Features": None
    }

    try:
        numeric = np.asarray(arr, dtype=np.float64)
        result["All_Numeric"] = True
        result["NaN_Count"] = int(
            np.isnan(numeric).sum()
        )
        result["Inf_Count"] = int(
            np.isinf(numeric).sum()
        )

        if np.isfinite(numeric).all():
            min_values = np.min(numeric, axis=0)
            max_values = np.max(numeric, axis=0)

            result["Constant_Features"] = int(
                np.sum(min_values == max_values)
            )
            result["Zero_Variance_Features"] = int(
                np.sum(np.var(numeric, axis=0) == 0)
            )

    except (TypeError, ValueError, OverflowError):
        pass

    return result

for name in sorted(prepared_datasets):
    data = prepared_datasets[name]

    X_train = data["X_train"]
    X_test = data["X_test"]

    y_train = np.asarray(data["y_train"]).ravel()
    y_test = np.asarray(data["y_test"]).ravel()

    task = data["task"]

    train_info = numeric_matrix_info(X_train)
    test_info = numeric_matrix_info(X_test)

    n_train = len(y_train)
    n_test = len(y_test)
    n_features = X_train.shape[1]

    # Preservar nomes originais, se disponíveis.
    if isinstance(X_train, pd.DataFrame):
        feature_names = [
            str(c) for c in X_train.columns
        ]
    else:
        feature_names = [
            str(c) for c in data.get(
                "feature_names",
                range(n_features)
            )
        ]

    duplicated_names = (
        len(feature_names) - len(set(feature_names))
    )

    # Os valores de K são derivados do protocolo
    # registrado, não dos resultados preditivos.
    absolute_k = [
        k for k in [3, 5, 10, 20]
        if k <= n_features
    ]

    proportional_k = [
        max(1, int(np.ceil(rate * n_features)))
        for rate in [0.10, 0.25, 0.50, 0.75]
    ]

    all_k = sorted(
        set(absolute_k + proportional_k)
    )

    n_classes = None
    min_class_count = None
    train_target_finite = None
    test_target_finite = None

    if task == "classification":
        classes, counts = np.unique(
            y_train, return_counts=True
        )

        n_classes = len(classes)
        min_class_count = int(counts.min())

        test_classes, test_counts = np.unique(
            y_test, return_counts=True
        )

        for label, count in zip(classes, counts):
            class_records.append({
                "Dataset": name,
                "Partition": "Train",
                "Class": str(label),
                "Count": int(count)
            })

        for label, count in zip(
            test_classes, test_counts
        ):
            class_records.append({
                "Dataset": name,
                "Partition": "Test",
                "Class": str(label),
                "Count": int(count)
            })

        if not set(test_classes).issubset(set(classes)):
            issues.append({
                "Dataset": name,
                "Severity": "ERROR",
                "Issue": (
                    "Classe de teste ausente no treino"
                )
            })

        if min_class_count < 10:
            issues.append({
                "Dataset": name,
                "Severity": "WARNING",
                "Issue": (
                    "Classe minoritária com menos "
                    "de 10 exemplos de treino"
                )
            })

    else:
        try:
            train_target_finite = bool(
                np.isfinite(
                    y_train.astype(float)
                ).all()
            )
            test_target_finite = bool(
                np.isfinite(
                    y_test.astype(float)
                ).all()
            )
        except (TypeError, ValueError):
            train_target_finite = False
            test_target_finite = False

    if not train_info["All_Numeric"]:
        issues.append({
            "Dataset": name,
            "Severity": "ERROR",
            "Issue": "Matriz de treino não numérica"
        })

    for partition, info in [
        ("Train", train_info),
        ("Test", test_info)
    ]:
        if (
            info["NaN_Count"] is not None
            and info["NaN_Count"] > 0
        ):
            issues.append({
                "Dataset": name,
                "Severity": "WARNING",
                "Issue": (
                    f"{partition}: valores NaN"
                )
            })

        if (
            info["Inf_Count"] is not None
            and info["Inf_Count"] > 0
        ):
            issues.append({
                "Dataset": name,
                "Severity": "ERROR",
                "Issue": (
                    f"{partition}: valores infinitos"
                )
            })

    if duplicated_names:
        issues.append({
            "Dataset": name,
            "Severity": "ERROR",
            "Issue": (
                "Nomes de características duplicados"
            )
        })

    if (
        task == "regression"
        and (
            not train_target_finite
            or not test_target_finite
        )
    ):
        issues.append({
            "Dataset": name,
            "Severity": "ERROR",
            "Issue": "Alvo de regressão não finito"
        })

    records.append({
        "Dataset": name,
        "Task": task,
        "Train": n_train,
        "Test": n_test,
        "Features": n_features,
        "Classes": n_classes,
        "Min_Class_Count": min_class_count,
        "Train_NaN": train_info["NaN_Count"],
        "Test_NaN": test_info["NaN_Count"],
        "Train_Inf": train_info["Inf_Count"],
        "Test_Inf": test_info["Inf_Count"],
        "Constant_Train_Features":
            train_info["Constant_Features"],
        "Duplicate_Feature_Names": duplicated_names,
        "Distinct_K": len(all_k),
        "K_Values": all_k,
        "Estimated_Selector_Fits": (
            N_METHODS * STABILITY_SPLITS
        ),
        "Estimated_Downstream_Fits": (
            N_METHODS
            * STABILITY_SPLITS
            * len(all_k)
            * N_MODELS
        )
    })

audit_df = pd.DataFrame(records)
classes_df = pd.DataFrame(class_records)
issues_df = pd.DataFrame(
    issues,
    columns=["Dataset", "Severity", "Issue"]
)

# Estimativas de volume, não de duração.
total_selector_fits = int(
    audit_df["Estimated_Selector_Fits"].sum()
)

total_downstream_fits = int(
    audit_df["Estimated_Downstream_Fits"].sum()
)

# Essas estimativas não incluem ajustes internos de
# RFE, buscas de hiperparâmetros, repetições de Random
# Top-K ou ajustes para aferição de recursos.
# O baseline com 100% das features também será
# contabilizado separadamente.

audit_df.to_csv(
    OUTPUT_DIR / f"{RUN_ID}_datasets.csv",
    index=False
)

classes_df.to_csv(
    OUTPUT_DIR / f"{RUN_ID}_classes.csv",
    index=False
)

issues_df.to_csv(
    OUTPUT_DIR / f"{RUN_ID}_issues.csv",
    index=False
)

report = {
    "timestamp_utc": RUN_ID,
    "dataset_count": len(audit_df),
    "classification_count": int(
        (audit_df["Task"] == "classification").sum()
    ),
    "regression_count": int(
        (audit_df["Task"] == "regression").sum()
    ),
    "estimated_selector_fits":
        total_selector_fits,
    "estimated_downstream_fits":
        total_downstream_fits,
    "issue_count": len(issues),
    "important_limitations": [
        (
            "Prepared arrays alone cannot prove "
            "leakage-free preprocessing provenance."
        ),
        (
            "The saved train/test split cannot be "
            "assumed to represent repeated independent "
            "evaluation."
        ),
        (
            "Feature selection and preprocessing must "
            "be refitted inside each training split."
        ),
        (
            "Computational estimates omit internal "
            "selector fits and random-control repeats."
        ),
        (
            "No model or selector has been trained "
            "by this audit."
        )
    ],
    "environment": {
        "python": platform.python_version(),
        "numpy": np.__version__,
        "sklearn": sklearn.__version__
    }
}

with open(
    OUTPUT_DIR / f"{RUN_ID}_report.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        report,
        f,
        indent=2,
        ensure_ascii=False
    )

print("=" * 72)
print("CFIS — AUDITORIA DE VIABILIDADE")
print("=" * 72)

display(
    audit_df[
        [
            "Dataset",
            "Train",
            "Test",
            "Features",
            "Classes",
            "Min_Class_Count",
            "Constant_Train_Features",
            "Distinct_K",
            "Estimated_Downstream_Fits"
        ]
    ]
)

print("\nDISTRIBUIÇÃO DAS CLASSES")
display(classes_df)

print("\nPROBLEMAS DETECTADOS")
if issues_df.empty:
    print("Nenhum problema detectado nesta auditoria.")
else:
    display(issues_df)

print("\nESTIMATIVA INICIAL DE EXECUÇÕES")
print("Seleções:", total_selector_fits)
print("Treinamentos posteriores:",
      total_downstream_fits)

print("\nATENÇÃO:")
print(
    "A procedência do pré-processamento ainda "
    "precisa ser comprovada pelo pipeline original."
)
print(
    "Os totais são estimativas de volume, "
    "não estimativas de tempo."
)

print("\nResultados salvos em:", OUTPUT_DIR)
print("\nCÉLULA 20 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 21
# AUDITORIA DE PROVENIÊNCIA E PRÉ-PROCESSAMENTO
# ============================================================

import json
import inspect
from pathlib import Path
from datetime import datetime, timezone

import numpy as np
import pandas as pd

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Não expor valores individuais, caminhos privados de
# terceiros nem conteúdo textual dos documentos.
SOURCE_TERMS = [
    "source", "origin", "path", "dataset",
    "pipeline", "preprocess", "transform",
    "vectorizer", "tfidf", "imputer",
    "encoder", "scaler", "split"
]

EXCLUDED_GLOBALS = {
    "prepared_datasets",
    "dataset_records",
    "manifest_df",
    "audit_df",
    "classes_df",
    "issues_df"
}

def safe_type_name(obj):
    cls = type(obj)
    return f"{cls.__module__}.{cls.__name__}"

def safe_metadata(obj):
    result = {"type": safe_type_name(obj)}
    if isinstance(obj, (str, Path)):
        # Registrar apenas a existência de uma referência,
        # sem copiar seu valor potencialmente privado.
        result["kind"] = "path_or_text_reference"
    elif isinstance(obj, (list, tuple, dict)):
        result["length"] = len(obj)
    elif hasattr(obj, "get_params"):
        result["estimator_class"] = safe_type_name(obj)
        result["is_fitted_check"] = "not_performed"
    return result

global_candidates = {}

for name, obj in list(globals().items()):
    if name.startswith("_") or name in EXCLUDED_GLOBALS:
        continue

    if any(term in name.lower() for term in SOURCE_TERMS):
        global_candidates[name] = safe_metadata(obj)

dataset_audit = []

for name, data in sorted(prepared_datasets.items()):
    known_keys = sorted(map(str, data.keys()))

    provenance_keys = [
        key for key in known_keys
        if any(term in key.lower() for term in SOURCE_TERMS)
    ]

    xtr = data["X_train"]
    xte = data["X_test"]

    train_columns = (
        list(map(str, xtr.columns))
        if isinstance(xtr, pd.DataFrame)
        else None
    )
    test_columns = (
        list(map(str, xte.columns))
        if isinstance(xte, pd.DataFrame)
        else None
    )

    dataset_audit.append({
        "Dataset": name,
        "Task": data["task"],
        "Prepared_Keys": known_keys,
        "Provenance_Keys": provenance_keys,
        "Train_Type": safe_type_name(xtr),
        "Test_Type": safe_type_name(xte),
        "Same_Column_Order": (
            train_columns == test_columns
            if train_columns is not None
            and test_columns is not None
            else None
        ),
        "Train_Index_Unique": (
            bool(xtr.index.is_unique)
            if isinstance(xtr, pd.DataFrame)
            else None
        ),
        "Test_Index_Unique": (
            bool(xte.index.is_unique)
            if isinstance(xte, pd.DataFrame)
            else None
        ),
        "Train_Test_Index_Overlap": (
            len(set(xtr.index) & set(xte.index))
            if isinstance(xtr, pd.DataFrame)
            and isinstance(xte, pd.DataFrame)
            else None
        ),
        "Provenance_Proven": False
    })

report_df = pd.DataFrame(dataset_audit)

# Índices repetidos entre treino e teste não demonstram
# vazamento: muitos pipelines redefinem os índices.
# Da mesma forma, índices distintos não provam isolamento.

report = {
    "timestamp_utc": RUN_ID,
    "purpose": "Preprocessing provenance audit",
    "dataset_audit": dataset_audit,
    "global_candidate_names": global_candidates,
    "limitations": [
        "Prepared arrays cannot prove preprocessing order.",
        "Index overlap is not proof of sample leakage.",
        "Index separation is not proof of no leakage.",
        "A fitted transformer's presence does not prove "
        "that it was fitted on training data only.",
        "Original preparation code or execution history "
        "is required to establish provenance.",
        "No training, fitting, or transformation performed."
    ]
}

json_path = OUTPUT_DIR / (
    f"{RUN_ID}_provenance_audit.json"
)
csv_path = OUTPUT_DIR / (
    f"{RUN_ID}_provenance_audit.csv"
)

report_df.to_csv(csv_path, index=False)

with open(json_path, "w", encoding="utf-8") as f:
    json.dump(
        report, f,
        ensure_ascii=False,
        indent=2,
        default=str
    )

print("=" * 72)
print("CFIS — AUDITORIA DE PROVENIÊNCIA")
print("=" * 72)

display(report_df[[
    "Dataset",
    "Provenance_Keys",
    "Train_Type",
    "Same_Column_Order",
    "Train_Test_Index_Overlap",
    "Provenance_Proven"
]])

print("\nVARIÁVEIS GLOBAIS POTENCIALMENTE RELEVANTES")
print(
    json.dumps(
        global_candidates,
        indent=2,
        ensure_ascii=False
    )
)

print("\nIMPORTANTE:")
print(
    "Provenance_Proven=False significa ausência de "
    "comprovação nesta auditoria, não vazamento detectado."
)
print(
    "Para confirmar a proveniência, será necessário "
    "examinar o código original de preparação."
)

print("\nRelatório:", json_path)
print("Tabela:", csv_path)
print("\nCÉLULA 21 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 22
# RASTREAMENTO DA PROVENIÊNCIA DOS DATASETS
# ============================================================

import re
import json
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd

TERMS = [
    "train_test_split", "TfidfVectorizer",
    "CountVectorizer", "fit_transform",
    "SimpleImputer", "OneHotEncoder",
    "StandardScaler", "MinMaxScaler",
    "fetch_20newsgroups", "fetch_openml",
    "make_classification", "make_regression",
    "Credit_Approval", "20_Newsgroups",
    "prepared_datasets", "dataset_metadata"
]

OUTPUT_DIR = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

def short_repr(value, limit=900):
    try:
        result = repr(value)
        return result[:limit] + (
            " ...[truncado]" if len(result) > limit else ""
        )
    except Exception as exc:
        return f"<erro de representação: {type(exc).__name__}>"

print("=" * 72)
print("CFIS — RASTREAMENTO DE PROVENIÊNCIA")
print("=" * 72)

print("\n1. METADADOS EXISTENTES")

for variable in [
    "dataset_metadata",
    "dataset_dir",
    "SOURCE_PATH",
    "source_path",
    "PROJECT_DIR"
]:
    if variable in globals():
        value = globals()[variable]
        print(f"\n{variable}:")
        print(short_repr(value, 3000))

print("\n2. ARQUIVOS CANDIDATOS NO PROJETO")

root = Path(PROJECT_DIR)

candidate_files = []
allowed_extensions = {".py", ".ipynb", ".json", ".md", ".txt"}

if root.exists():
    for path in root.rglob("*"):
        if not path.is_file():
            continue

        if path.suffix.lower() not in allowed_extensions:
            continue

        # Evitar relatórios produzidos nas auditorias.
        if "CFIS_Final_Benchmark" in path.parts:
            continue

        try:
            size = path.stat().st_size
        except OSError:
            continue

        if size > 5_000_000:
            continue

        candidate_files.append(path)

print("Arquivos candidatos:", len(candidate_files))

matches = []

for path in candidate_files:
    try:
        content = path.read_text(
            encoding="utf-8",
            errors="replace"
        )
    except OSError:
        continue

    # Para notebooks, busca textual no JSON sem execução.
    lines = content.splitlines()

    for i, line in enumerate(lines):
        found = [
            term for term in TERMS
            if term.lower() in line.lower()
        ]

        if not found:
            continue

        matches.append({
            "File": str(path.relative_to(root)),
            "Line": i + 1,
            "Terms": ", ".join(found),
            "Excerpt": line.strip()[:350]
        })

matches_df = pd.DataFrame(
    matches,
    columns=["File", "Line", "Terms", "Excerpt"]
)

print("\n3. OCORRÊNCIAS ENCONTRADAS")
print("Total:", len(matches_df))

if not matches_df.empty:
    grouped = (
        matches_df.groupby("File")
        .size()
        .sort_values(ascending=False)
        .reset_index(name="Matches")
    )
    display(grouped.head(30))

    print("\nTRECHOS MAIS RELEVANTES")
    priority = [
        "train_test_split",
        "TfidfVectorizer",
        "SimpleImputer",
        "OneHotEncoder",
        "fit_transform",
        "fetch_20newsgroups",
        "fetch_openml"
    ]

    relevant = matches_df[
        matches_df["Terms"].apply(
            lambda text: any(
                term in text for term in priority
            )
        )
    ]

    display(relevant.head(60))
else:
    print("Nenhuma ocorrência localizada.")

csv_path = OUTPUT_DIR / (
    f"{RUN_ID}_source_code_matches.csv"
)

matches_df.to_csv(csv_path, index=False)

print("\n4. LIMITAÇÃO DA AUDITORIA")
print(
    "Ocorrências textuais indicam candidatos, "
    "não comprovam qual código foi executado."
)
print(
    "A próxima etapa será inspecionar a sequência "
    "de preparação e verificar o isolamento do teste."
)

print("\nArquivo:", csv_path)
print("\nCÉLULA 22 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 23
# AUDITORIA DOS METADADOS ORIGINAIS
# ============================================================

import json
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd

ROOT = Path(PROJECT_DIR) / "01_Datasets"
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

DATASET_NAMES = sorted(prepared_datasets.keys())

KEY_TERMS = (
    "source", "origin", "dataset", "seed",
    "random", "split", "train", "test",
    "fit", "transform", "preprocess",
    "imput", "encod", "scal", "tfidf",
    "vector", "leak", "synthetic",
    "feature", "class", "stratif"
)

def flatten_metadata(obj, prefix=""):
    rows = []

    if isinstance(obj, dict):
        for key, value in obj.items():
            path = f"{prefix}.{key}" if prefix else str(key)
            rows.extend(flatten_metadata(value, path))

    elif isinstance(obj, list):
        if len(obj) <= 15 and all(
            not isinstance(v, (dict, list))
            for v in obj
        ):
            rows.append((prefix, obj))
        else:
            rows.append((
                prefix,
                f"<lista com {len(obj)} elementos>"
            ))

    else:
        rows.append((prefix, obj))

    return rows

metadata_files = sorted(
    p for p in ROOT.rglob("*.json")
    if p.is_file()
    and any(
        section in p.parts
        for section in ("Raw", "Splits", "Processed", "Synthetic")
    )
)

records = []
errors = []

for path in metadata_files:
    try:
        if path.stat().st_size > 2_000_000:
            continue

        with open(path, "r", encoding="utf-8") as f:
            metadata = json.load(f)

    except (OSError, ValueError) as exc:
        errors.append({
            "File": str(path.relative_to(ROOT)),
            "Error": str(exc)[:200]
        })
        continue

    relative = str(path.relative_to(ROOT))
    flat = flatten_metadata(metadata)

    for key, value in flat:
        relevant = (
            any(term in key.lower() for term in KEY_TERMS)
            or any(
                name.lower() in relative.lower()
                for name in DATASET_NAMES
            )
        )

        if relevant:
            records.append({
                "File": relative,
                "Key": key,
                "Value": str(value)[:500]
            })

details_df = pd.DataFrame(
    records, columns=["File", "Key", "Value"]
)
errors_df = pd.DataFrame(
    errors, columns=["File", "Error"]
)

# Visão focada nos metadados de preparação.
priority_terms = (
    "fit_only_on_training",
    "fit_on_train",
    "train_only",
    "split",
    "seed",
    "source",
    "origin",
    "preprocess",
    "imput",
    "encod",
    "tfidf",
    "vector",
    "stratif"
)

if not details_df.empty:
    priority_df = details_df[
        details_df["Key"].str.lower().apply(
            lambda s: any(t in s for t in priority_terms)
        )
    ].copy()
else:
    priority_df = details_df.copy()

# Cobertura de metadados por dataset.
coverage = []

for name in DATASET_NAMES:
    matching = [
        str(p.relative_to(ROOT))
        for p in metadata_files
        if name.lower() in str(p).lower()
    ]

    coverage.append({
        "Dataset": name,
        "Metadata_Files": len(matching),
        "Files": matching
    })

coverage_df = pd.DataFrame(coverage)

details_path = OUT / f"{RUN_ID}_metadata_details.csv"
priority_path = OUT / f"{RUN_ID}_metadata_priority.csv"
coverage_path = OUT / f"{RUN_ID}_metadata_coverage.csv"

details_df.to_csv(details_path, index=False)
priority_df.to_csv(priority_path, index=False)
coverage_df.to_csv(coverage_path, index=False)

print("=" * 72)
print("CFIS — AUDITORIA DOS METADADOS")
print("=" * 72)

print("\n1. COBERTURA POR DATASET")
display(coverage_df[["Dataset", "Metadata_Files"]])

print("\n2. EVIDÊNCIAS PRIORITÁRIAS")
display(priority_df.head(100))

print("\n3. METADADOS DE DATASETS CRÍTICOS")

critical_names = [
    "20_Newsgroups",
    "Credit_Approval",
    "Gisette",
    "Synthetic_120",
    "Synthetic_60",
    "Synthetic_30"
]

if not details_df.empty:
    critical_df = details_df[
        details_df["File"].apply(
            lambda s: any(
                name.lower() in s.lower()
                for name in critical_names
            )
        )
    ]
    display(critical_df.head(100))

print("\n4. ERROS DE LEITURA")
if errors_df.empty:
    print("Nenhum.")
else:
    display(errors_df)

print("\nLIMITAÇÃO:")
print(
    "Metadados declarados não comprovam, sozinhos, "
    "a ordem real das operações executadas."
)
print(
    "A confirmação definitiva exige o código "
    "ou um registro verificável da preparação."
)

print("\nArquivos gerados:")
print(details_path)
print(priority_path)
print(coverage_path)

print("\nCÉLULA 23 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 24
# VERIFICAÇÃO CRUZADA DE METADADOS
# ============================================================

import json
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd

ROOT = Path(PROJECT_DIR) / "01_Datasets"
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

def read_json(path):
    if not path.exists():
        return None
    try:
        with open(path, encoding="utf-8") as f:
            return json.load(f)
    except (OSError, ValueError) as exc:
        return {"_read_error": str(exc)}

def flatten(obj, prefix=""):
    if isinstance(obj, dict):
        result = {}
        for key, value in obj.items():
            new_key = (
                f"{prefix}.{key}" if prefix else str(key)
            )
            result.update(flatten(value, new_key))
        return result
    return {prefix: obj}

rows = []
details = []
flags = []

for name, data in sorted(prepared_datasets.items()):
    paths = {
        "processed": (
            ROOT / "Processed" / name / "metadata.json"
        ),
        "split": (
            ROOT / "Splits" / f"{name}_split_metadata.json"
        ),
        "raw": (
            ROOT / "Raw" / f"{name}_metadata.json"
        ),
        "synthetic": (
            ROOT / "Synthetic" / f"{name}_metadata.json"
        )
    }

    metadata = {
        kind: read_json(path)
        for kind, path in paths.items()
    }

    n_train = len(data["X_train"])
    n_test = len(data["X_test"])
    n_features = data["X_train"].shape[1]

    processed = metadata["processed"] or {}
    split = metadata["split"] or {}

    def compare(field, observed, expected):
        if expected is None:
            return "NOT_DOCUMENTED"
        try:
            return (
                "MATCH"
                if int(expected) == int(observed)
                else "MISMATCH"
            )
        except (TypeError, ValueError):
            return "NOT_COMPARABLE"

    checks = {
        "Train_Check": compare(
            "train_samples",
            n_train,
            processed.get("train_samples")
        ),
        "Test_Check": compare(
            "test_samples",
            n_test,
            processed.get("test_samples")
        ),
        "Feature_Check": compare(
            "processed_features",
            n_features,
            processed.get("processed_features")
        ),
        "Split_Train_Check": compare(
            "n_train",
            n_train,
            split.get("n_train")
        ),
        "Split_Test_Check": compare(
            "n_test",
            n_test,
            split.get("n_test")
        )
    }

    rows.append({
        "Dataset": name,
        "Train": n_train,
        "Test": n_test,
        "Features": n_features,
        "Fit_Only_On_Training_Declared":
            processed.get("fit_only_on_training"),
        "Split_Strategy": split.get("strategy"),
        **checks
    })

    for kind, obj in metadata.items():
        if obj is None:
            continue

        for key, value in flatten(obj).items():
            details.append({
                "Dataset": name,
                "Source": kind,
                "Key": key,
                "Value": json.dumps(
                    value,
                    ensure_ascii=False,
                    default=str
                )
            })

    for check_name, result in checks.items():
        if result == "MISMATCH":
            flags.append({
                "Dataset": name,
                "Issue": check_name,
                "Severity": "ERROR"
            })

    if processed.get("fit_only_on_training") is not True:
        flags.append({
            "Dataset": name,
            "Issue": (
                "Train-only preprocessing not declared"
            ),
            "Severity": "REVIEW"
        })

summary_df = pd.DataFrame(rows)
details_df = pd.DataFrame(details)
flags_df = pd.DataFrame(
    flags,
    columns=["Dataset", "Issue", "Severity"]
)

summary_path = OUT / f"{RUN_ID}_crosscheck.csv"
details_path = OUT / f"{RUN_ID}_metadata_full.csv"
flags_path = OUT / f"{RUN_ID}_flags.csv"

summary_df.to_csv(summary_path, index=False)
details_df.to_csv(details_path, index=False)
flags_df.to_csv(flags_path, index=False)

print("=" * 72)
print("CFIS — VERIFICAÇÃO CRUZADA")
print("=" * 72)

print("\n1. CONSISTÊNCIA DOS 12 DATASETS")
print(summary_df.to_string(index=False))

print("\n2. CAMPOS CRÍTICOS DOS METADADOS")
critical = details_df[
    details_df["Dataset"].isin([
        "20_Newsgroups",
        "Credit_Approval",
        "Gisette",
        "Synthetic_30",
        "Synthetic_60",
        "Synthetic_120"
    ])
]

print(critical.to_string(index=False, max_colwidth=120))

print("\n3. INCONSISTÊNCIAS")
if flags_df.empty:
    print("Nenhuma inconsistência detectada.")
else:
    print(flags_df.to_string(index=False))

print("\n4. INTERPRETAÇÃO")
print(
    "MATCH confirma coerência de metadados, "
    "não ausência de vazamento."
)
print(
    "A ordem real de fit/transform exige "
    "código de preparação ou evidência equivalente."
)

print("\nArquivos salvos:")
print(summary_path)
print(details_path)
print(flags_path)
print("\nCÉLULA 24 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 25
# LOCALIZAÇÃO DOS PIPELINES ORIGINAIS
# ============================================================

import json
from pathlib import Path
from datetime import datetime, timezone
import pandas as pd

ROOT = Path(PROJECT_DIR)
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

SEARCH_TERMS = {
    "split": [
        "train_test_split", "GroupShuffleSplit",
        "StratifiedShuffleSplit", "ShuffleSplit"
    ],
    "preprocessing": [
        "TfidfVectorizer", "CountVectorizer",
        "SimpleImputer", "OneHotEncoder",
        "ColumnTransformer", "StandardScaler",
        "VarianceThreshold", "Pipeline("
    ],
    "fit": [
        ".fit(", ".fit_transform(", ".transform("
    ],
    "saving": [
        "to_csv(", "to_parquet(",
        "joblib.dump(", "pickle.dump("
    ]
}

def extract_code(path):
    if path.suffix == ".py":
        return [{
            "cell": 0,
            "code": path.read_text(
                encoding="utf-8", errors="replace"
            )
        }]

    with open(path, encoding="utf-8") as f:
        notebook = json.load(f)

    return [
        {
            "cell": index,
            "code": "".join(cell.get("source", []))
        }
        for index, cell in enumerate(
            notebook.get("cells", [])
        )
        if cell.get("cell_type") == "code"
    ]

files = [
    p for p in ROOT.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".py", ".ipynb"}
    and "CFIS_Final_Benchmark" not in p.parts
    and p.stat().st_size <= 10_000_000
]

records = []
errors = []

for path in sorted(files):
    try:
        cells = extract_code(path)
    except (OSError, ValueError, TypeError) as exc:
        errors.append({
            "File": str(path.relative_to(ROOT)),
            "Error": str(exc)[:180]
        })
        continue

    for cell in cells:
        code = cell["code"]
        found = {
            group: [
                term for term in terms
                if term.lower() in code.lower()
            ]
            for group, terms in SEARCH_TERMS.items()
        }

        found = {
            key: value
            for key, value in found.items()
            if value
        }

        if not found:
            continue

        # Identificar células com maior relevância
        # para a reconstrução da preparação.
        score = (
            4 * int("split" in found)
            + 4 * int("preprocessing" in found)
            + 2 * int("fit" in found)
            + 2 * int("saving" in found)
        )

        lines = code.splitlines()
        relevant_lines = []

        for i, line in enumerate(lines):
            if any(
                term.lower() in line.lower()
                for terms in SEARCH_TERMS.values()
                for term in terms
            ):
                relevant_lines.append(
                    f"L{i+1}: {line.strip()[:180]}"
                )

        records.append({
            "File": str(path.relative_to(ROOT)),
            "Cell": cell["cell"],
            "Score": score,
            "Categories": ", ".join(found.keys()),
            "Matches": "; ".join(
                term
                for terms in found.values()
                for term in terms
            ),
            "Relevant_Lines": "\n".join(
                relevant_lines[:18]
            ),
            "Code": code
        })

df = pd.DataFrame(records)
errors_df = pd.DataFrame(errors)

if not df.empty:
    df = df.sort_values(
        ["Score", "File", "Cell"],
        ascending=[False, True, True]
    ).reset_index(drop=True)

    # Guardar código completo para auditoria posterior.
    full_path = OUT / f"{RUN_ID}_pipeline_cells.json"
    with open(full_path, "w", encoding="utf-8") as f:
        json.dump(
            df.to_dict(orient="records"),
            f,
            ensure_ascii=False,
            indent=2
        )

    summary_path = OUT / f"{RUN_ID}_pipeline_summary.csv"
    df.drop(columns=["Code"]).to_csv(
        summary_path, index=False
    )

print("=" * 72)
print("CFIS — LOCALIZAÇÃO DOS PIPELINES")
print("=" * 72)

print("Arquivos de código examinados:", len(files))
print("Células/trechos relevantes:", len(df))
print("Erros de leitura:", len(errors_df))

if not df.empty:
    print("\nARQUIVOS MAIS RELEVANTES")
    display(
        df.groupby("File")
        .agg(
            Max_Score=("Score", "max"),
            Relevant_Cells=("Cell", "count")
        )
        .sort_values(
            ["Max_Score", "Relevant_Cells"],
            ascending=False
        )
        .head(20)
    )

    print("\nTRECHOS PRIORITÁRIOS")
    for _, row in df.head(12).iterrows():
        print("\n" + "-" * 60)
        print("Arquivo:", row["File"])
        print("Célula:", row["Cell"])
        print("Categorias:", row["Categories"])
        print(row["Relevant_Lines"][:2500])

    print("\nRelatório completo:", full_path)
    print("Resumo:", summary_path)
else:
    print(
        "\nNenhum pipeline identificado nos arquivos "
        "Python/notebooks examinados."
    )
    print(
        "Nesse caso, precisaremos localizar o "
        "notebook original de preparação."
    )

if not errors_df.empty:
    print("\nERROS DE LEITURA")
    display(errors_df.head(10))

print("\nCÉLULA 25 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 27
# INSPEÇÃO DOS TRANSFORMADORES SERIALIZADOS
# ============================================================

import json
import joblib
import pandas as pd
import numpy as np

from pathlib import Path
from datetime import datetime, timezone
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

ROOT = Path(PROJECT_DIR) / "01_Datasets"
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

def class_name(obj):
    cls = type(obj)
    return f"{cls.__module__}.{cls.__name__}"

def describe_estimator(estimator, prefix="root", depth=0):
    """Inspeciona a estrutura sem chamar fit ou transform."""
    rows = [{
        "Component": prefix,
        "Class": class_name(estimator),
        "N_Features_In": getattr(
            estimator, "n_features_in_", None
        ),
        "N_Output_Features": None,
        "Fitted_Attributes": ", ".join(
            sorted(
                key for key in vars(estimator)
                if key.endswith("_")
                and not key.startswith("__")
            )
        )[:350] if hasattr(estimator, "__dict__") else ""
    }]

    if depth >= 5:
        return rows

    if isinstance(estimator, Pipeline):
        for name, step in estimator.steps:
            rows.extend(describe_estimator(
                step, f"{prefix}.{name}", depth + 1
            ))

    elif isinstance(estimator, ColumnTransformer):
        components = getattr(
            estimator,
            "transformers_",
            estimator.transformers
        )

        for name, transformer, columns in components:
            if isinstance(transformer, str):
                rows.append({
                    "Component": f"{prefix}.{name}",
                    "Class": transformer,
                    "N_Features_In": None,
                    "N_Output_Features": None,
                    "Fitted_Attributes": ""
                })
            else:
                rows.extend(describe_estimator(
                    transformer,
                    f"{prefix}.{name}",
                    depth + 1
                ))

    return rows

dataset_rows = []
component_rows = []
errors = []

for dataset_name in sorted(prepared_datasets):
    folder = ROOT / "Processed" / dataset_name

    joblib_files = sorted(folder.glob("*.joblib"))
    parquet_files = sorted(folder.glob("*.parquet"))

    dataset_rows.append({
        "Dataset": dataset_name,
        "Joblib_Files": ", ".join(
            p.name for p in joblib_files
        ),
        "Parquet_Files": ", ".join(
            p.name for p in parquet_files
        ),
        "N_Joblib": len(joblib_files),
        "N_Parquet": len(parquet_files)
    })

    for path in joblib_files:
        try:
            transformer = joblib.load(path)

            if isinstance(transformer, dict):
                # Alguns projetos salvam um dicionário
                # contendo vários transformadores.
                for key, obj in transformer.items():
                    if hasattr(obj, "transform"):
                        components = describe_estimator(
                            obj, str(key)
                        )
                        for item in components:
                            component_rows.append({
                                "Dataset": dataset_name,
                                "File": path.name,
                                **item
                            })
                    else:
                        component_rows.append({
                            "Dataset": dataset_name,
                            "File": path.name,
                            "Component": str(key),
                            "Class": class_name(obj),
                            "N_Features_In": None,
                            "N_Output_Features": None,
                            "Fitted_Attributes": ""
                        })

            else:
                components = describe_estimator(transformer)
                for item in components:
                    component_rows.append({
                        "Dataset": dataset_name,
                        "File": path.name,
                        **item
                    })

        except Exception as exc:
            errors.append({
                "Dataset": dataset_name,
                "File": path.name,
                "Error": (
                    f"{type(exc).__name__}: {str(exc)[:300]}"
                )
            })

files_df = pd.DataFrame(dataset_rows)
components_df = pd.DataFrame(component_rows)
errors_df = pd.DataFrame(
    errors, columns=["Dataset", "File", "Error"]
)

files_path = OUT / f"{RUN_ID}_transformer_files.csv"
components_path = OUT / f"{RUN_ID}_transformer_components.csv"
errors_path = OUT / f"{RUN_ID}_transformer_errors.csv"

files_df.to_csv(files_path, index=False)
components_df.to_csv(components_path, index=False)
errors_df.to_csv(errors_path, index=False)

print("=" * 72)
print("CFIS — INSPEÇÃO DOS TRANSFORMADORES")
print("=" * 72)

print("\n1. ARQUIVOS POR DATASET")
print(files_df.to_string(index=False))

print("\n2. COMPONENTES DOS TRANSFORMADORES")
if not components_df.empty:
    print(
        components_df[
            [
                "Dataset", "Component",
                "Class", "N_Features_In",
                "Fitted_Attributes"
            ]
        ].to_string(
            index=False,
            max_colwidth=85
        )
    )
else:
    print("Nenhum componente identificado.")

print("\n3. ERROS DE CARREGAMENTO")
if errors_df.empty:
    print("Nenhum.")
else:
    print(errors_df.to_string(index=False))

print("\n4. INTERPRETAÇÃO")
print(
    "Atributos de ajuste indicam que o objeto "
    "contém estado aprendido."
)
print(
    "Isso não comprova em quais amostras "
    "o transformador foi ajustado."
)
print(
    "Nenhum fit ou transform foi executado "
    "por esta célula."
)

print("\nRelatórios:")
print(files_path)
print(components_path)
print(errors_path)

print("\nCÉLULA 27 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 28
# RECONSTRUÇÃO TRAIN-ONLY DOS DATASETS CRÍTICOS
# ============================================================

import json
import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone
from sklearn.base import clone
from scipy import sparse

ROOT = Path(PROJECT_DIR) / "01_Datasets"
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

TARGETS = [
    "20_Newsgroups",
    "Credit_Approval",
    "Gisette"
]

def dense_array(value):
    if sparse.issparse(value):
        return value.toarray()
    return np.asarray(value)

def transform_input(X, name):
    if name == "20_Newsgroups":
        if "text" not in X.columns:
            raise ValueError("Coluna 'text' ausente.")
        return X["text"].fillna("").astype(str)
    return X

def numeric_comparison(a, b):
    a = dense_array(a)
    b = dense_array(b)

    if a.shape != b.shape:
        return {
            "Match": False,
            "Max_Abs_Diff": np.nan,
            "Shape_A": str(a.shape),
            "Shape_B": str(b.shape)
        }

    try:
        a = np.asarray(a, dtype=np.float64)
        b = np.asarray(b, dtype=np.float64)
        difference = np.abs(a - b)

        return {
            "Match": bool(np.allclose(
                a, b, rtol=1e-10, atol=1e-10,
                equal_nan=True
            )),
            "Max_Abs_Diff": float(
                np.nanmax(difference)
            ) if difference.size else 0.0,
            "Shape_A": str(a.shape),
            "Shape_B": str(b.shape)
        }
    except (TypeError, ValueError):
        return {
            "Match": bool(np.array_equal(a, b)),
            "Max_Abs_Diff": np.nan,
            "Shape_A": str(a.shape),
            "Shape_B": str(b.shape)
        }

results = []
errors = []

for name in TARGETS:
    print(f"\n{'=' * 64}")
    print("DATASET:", name)

    try:
        raw_folder = (
            ROOT / "Synthetic"
            if name.startswith("Synthetic_")
            else ROOT / "Raw"
        )

        raw = pd.read_parquet(
            raw_folder / f"{name}.parquet"
        )

        train_idx = np.load(
            ROOT / "Splits"
            / f"{name}_train_indices.npy",
            allow_pickle=False
        )
        test_idx = np.load(
            ROOT / "Splits"
            / f"{name}_test_indices.npy",
            allow_pickle=False
        )

        train_idx = np.asarray(train_idx)
        test_idx = np.asarray(test_idx)

        if (
            train_idx.ndim != 1
            or test_idx.ndim != 1
            or not np.issubdtype(
                train_idx.dtype, np.integer
            )
            or not np.issubdtype(
                test_idx.dtype, np.integer
            )
        ):
            raise ValueError("Índices inválidos.")

        if (
            np.any(train_idx < 0)
            or np.any(test_idx < 0)
            or np.any(train_idx >= len(raw))
            or np.any(test_idx >= len(raw))
        ):
            raise ValueError("Índices fora dos limites.")

        if (
            len(np.unique(train_idx)) != len(train_idx)
            or len(np.unique(test_idx)) != len(test_idx)
            or len(np.intersect1d(
                train_idx, test_idx
            )) > 0
        ):
            raise ValueError(
                "Duplicação ou sobreposição de índices."
            )

        with open(
            raw_folder / f"{name}_metadata.json",
            encoding="utf-8"
        ) as f:
            raw_meta = json.load(f)

        target_column = raw_meta["target_column"]

        X_raw = raw.drop(columns=[target_column])
        y_raw = raw[target_column]

        X_train_raw = X_raw.iloc[train_idx]
        X_test_raw = X_raw.iloc[test_idx]

        y_train_raw = y_raw.iloc[train_idx]
        y_test_raw = y_raw.iloc[test_idx]

        folder = ROOT / "Processed" / name
        artifact = joblib.load(
            folder / "preprocessor.joblib"
        )

        if not isinstance(artifact, dict):
            raise TypeError(
                "Formato inesperado do artefato: "
                f"{type(artifact).__name__}"
            )

        original_preprocessor = artifact["preprocessor"]
        original_filter = artifact["variance_filter"]

        # Clones sem os parâmetros aprendidos.
        new_preprocessor = clone(original_preprocessor)
        new_filter = clone(original_filter)

        train_input = transform_input(
            X_train_raw, name
        )
        test_input = transform_input(
            X_test_raw, name
        )

        # Ajuste EXCLUSIVAMENTE no treinamento.
        train_transformed = (
            new_preprocessor.fit_transform(train_input)
        )
        test_transformed = (
            new_preprocessor.transform(test_input)
        )

        train_rebuilt = new_filter.fit_transform(
            train_transformed
        )
        test_rebuilt = new_filter.transform(
            test_transformed
        )

        # Dados históricos preservados.
        X_train_saved = pd.read_parquet(
            folder / "X_train.parquet"
        )
        X_test_saved = pd.read_parquet(
            folder / "X_test.parquet"
        )
        y_train_saved = pd.read_parquet(
            folder / "y_train.parquet"
        )
        y_test_saved = pd.read_parquet(
            folder / "y_test.parquet"
        )

        train_check = numeric_comparison(
            train_rebuilt,
            X_train_saved.to_numpy()
        )
        test_check = numeric_comparison(
            test_rebuilt,
            X_test_saved.to_numpy()
        )

        # Comparação do estado aprendido pelo
        # filtro de variância.
        variance_check = numeric_comparison(
            new_filter.variances_,
            original_filter.variances_
        )

        support_match = np.array_equal(
            new_filter.get_support(),
            original_filter.get_support()
        )

        # Comparação dos alvos na ordem dos índices.
        y_train_check = np.array_equal(
            np.asarray(y_train_raw).ravel(),
            y_train_saved.to_numpy().ravel()
        )
        y_test_check = np.array_equal(
            np.asarray(y_test_raw).ravel(),
            y_test_saved.to_numpy().ravel()
        )

        # Inspecionar a quantidade de amostras usada
        # pelo scaler original, quando disponível.
        scaler_sample_counts = []

        def inspect_scalers(obj):
            if hasattr(obj, "n_samples_seen_"):
                scaler_sample_counts.append(
                    str(obj.n_samples_seen_)
                )

            if hasattr(obj, "steps"):
                for _, child in obj.steps:
                    inspect_scalers(child)

            if hasattr(obj, "transformers_"):
                for _, child, _ in obj.transformers_:
                    if not isinstance(child, str):
                        inspect_scalers(child)

        inspect_scalers(original_preprocessor)

        record = {
            "Dataset": name,
            "Raw_Samples": len(raw),
            "Train_Samples": len(train_idx),
            "Test_Samples": len(test_idx),
            "Train_Match": train_check["Match"],
            "Test_Match": test_check["Match"],
            "Train_Max_Abs_Diff":
                train_check["Max_Abs_Diff"],
            "Test_Max_Abs_Diff":
                test_check["Max_Abs_Diff"],
            "Variance_State_Match":
                variance_check["Match"],
            "Variance_Max_Abs_Diff":
                variance_check["Max_Abs_Diff"],
            "Support_Match": support_match,
            "Y_Train_Match": y_train_check,
            "Y_Test_Match": y_test_check,
            "Original_Scaler_Sample_Counts":
                "; ".join(scaler_sample_counts),
            "Train_Shape": train_check["Shape_A"],
            "Test_Shape": test_check["Shape_A"]
        }

        results.append(record)

        for key, value in record.items():
            print(f"{key}: {value}")

    except Exception as exc:
        errors.append({
            "Dataset": name,
            "Error": (
                f"{type(exc).__name__}: {str(exc)}"
            )
        })
        print("ERRO:", type(exc).__name__, str(exc))

results_df = pd.DataFrame(results)
errors_df = pd.DataFrame(
    errors, columns=["Dataset", "Error"]
)

results_path = OUT / f"{RUN_ID}_reconstruction.csv"
errors_path = OUT / f"{RUN_ID}_reconstruction_errors.csv"

results_df.to_csv(results_path, index=False)
errors_df.to_csv(errors_path, index=False)

print("\n" + "=" * 72)
print("RESUMO DA RECONSTRUÇÃO")
print("=" * 72)

if not results_df.empty:
    print(results_df.to_string(index=False))

if not errors_df.empty:
    print("\nERROS:")
    print(errors_df.to_string(index=False))

print("\nINTERPRETAÇÃO:")
print(
    "A coincidência dos dados e estados aprendidos "
    "é evidência forte de reprodução train-only."
)
print(
    "A comparação não recupera o histórico de "
    "execução do pipeline original."
)

print("\nRelatórios:")
print(results_path)
print(errors_path)
print("\nCÉLULA 28 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 29
# RECONSTRUÇÃO TRAIN-ONLY DOS 9 DATASETS RESTANTES
# ============================================================

import json
import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone
from sklearn.base import clone
from scipy import sparse

ROOT = Path(PROJECT_DIR) / "01_Datasets"
OUT = (
    Path(DIRS["efficiency"])
    / "CFIS_Final_Benchmark"
    / "01_Implementation_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

ALREADY_CHECKED = {
    "20_Newsgroups",
    "Credit_Approval",
    "Gisette"
}

TARGETS = sorted(
    set(prepared_datasets) - ALREADY_CHECKED
)

def as_array(value):
    if sparse.issparse(value):
        return value.toarray()
    return np.asarray(value)

def compare_numeric(a, b, atol=1e-10):
    a = np.asarray(as_array(a), dtype=np.float64)
    b = np.asarray(as_array(b), dtype=np.float64)

    if a.shape != b.shape:
        return False, np.nan

    if a.size == 0:
        return True, 0.0

    difference = np.abs(a - b)

    return (
        bool(np.allclose(
            a, b,
            rtol=1e-10,
            atol=atol,
            equal_nan=True
        )),
        float(np.nanmax(difference))
    )

def fitted_state(obj):
    """Extrai atributos numéricos aprendidos dos componentes."""
    states = {}

    def walk(component, prefix):
        for attr in (
            "mean_", "scale_", "var_", "statistics_",
            "variances_", "n_samples_seen_"
        ):
            if hasattr(component, attr):
                states[f"{prefix}.{attr}"] = getattr(
                    component, attr
                )

        if hasattr(component, "steps"):
            for key, child in component.steps:
                walk(child, f"{prefix}.{key}")

        if hasattr(component, "transformers_"):
            for key, child, _ in component.transformers_:
                if not isinstance(child, str):
                    walk(child, f"{prefix}.{key}")

    walk(obj, "preprocessor")
    return states

results = []

for name in TARGETS:
    record = {
        "Dataset": name,
        "Status": "ERROR",
        "Train_Match": False,
        "Test_Match": False,
        "Y_Train_Match": False,
        "Y_Test_Match": False,
        "Variance_Match": False,
        "Support_Match": False,
        "Fitted_State_Match": False,
        "Train_Max_Diff": np.nan,
        "Test_Max_Diff": np.nan,
        "State_Max_Diff": np.nan,
        "Error": ""
    }

    try:
        raw_folder = (
            ROOT / "Synthetic"
            if name.startswith("Synthetic_")
            else ROOT / "Raw"
        )

        raw = pd.read_parquet(
            raw_folder / f"{name}.parquet"
        )

        with open(
            raw_folder / f"{name}_metadata.json",
            encoding="utf-8"
        ) as f:
            meta = json.load(f)

        target = meta["target_column"]

        train_idx = np.load(
            ROOT / "Splits"
            / f"{name}_train_indices.npy",
            allow_pickle=False
        )
        test_idx = np.load(
            ROOT / "Splits"
            / f"{name}_test_indices.npy",
            allow_pickle=False
        )

        if (
            train_idx.ndim != 1
            or test_idx.ndim != 1
            or not np.issubdtype(train_idx.dtype, np.integer)
            or not np.issubdtype(test_idx.dtype, np.integer)
        ):
            raise ValueError("Formato de índices inválido.")

        combined = np.concatenate(
            [train_idx, test_idx]
        )

        if (
            len(combined) != len(raw)
            or len(np.unique(combined)) != len(raw)
            or np.any(combined < 0)
            or np.any(combined >= len(raw))
        ):
            raise ValueError(
                "Índices não formam uma partição válida."
            )

        X = raw.drop(columns=[target])
        y = raw[target]

        X_train = X.iloc[train_idx]
        X_test = X.iloc[test_idx]

        folder = ROOT / "Processed" / name

        artifact = joblib.load(
            folder / "preprocessor.joblib"
        )

        original = artifact["preprocessor"]
        original_filter = artifact["variance_filter"]

        rebuilt = clone(original)
        rebuilt_filter = clone(original_filter)

        # O ajuste ocorre somente no treinamento.
        Z_train = rebuilt.fit_transform(X_train)
        Z_test = rebuilt.transform(X_test)

        Z_train = rebuilt_filter.fit_transform(
            Z_train
        )
        Z_test = rebuilt_filter.transform(Z_test)

        saved_train = pd.read_parquet(
            folder / "X_train.parquet"
        )
        saved_test = pd.read_parquet(
            folder / "X_test.parquet"
        )
        saved_y_train = pd.read_parquet(
            folder / "y_train.parquet"
        )
        saved_y_test = pd.read_parquet(
            folder / "y_test.parquet"
        )

        train_ok, train_diff = compare_numeric(
            Z_train, saved_train.to_numpy()
        )
        test_ok, test_diff = compare_numeric(
            Z_test, saved_test.to_numpy()
        )

        variance_ok, _ = compare_numeric(
            rebuilt_filter.variances_,
            original_filter.variances_
        )

        support_ok = bool(np.array_equal(
            rebuilt_filter.get_support(),
            original_filter.get_support()
        ))

        original_state = fitted_state(original)
        rebuilt_state = fitted_state(rebuilt)

        state_ok = (
            original_state.keys() == rebuilt_state.keys()
        )
        max_state_diff = 0.0

        if state_ok:
            for key in original_state:
                equal, difference = compare_numeric(
                    original_state[key],
                    rebuilt_state[key]
                )

                state_ok = state_ok and equal

                if np.isfinite(difference):
                    max_state_diff = max(
                        max_state_diff, difference
                    )

        y_train_ok = bool(np.array_equal(
            y.iloc[train_idx].to_numpy().ravel(),
            saved_y_train.to_numpy().ravel()
        ))
        y_test_ok = bool(np.array_equal(
            y.iloc[test_idx].to_numpy().ravel(),
            saved_y_test.to_numpy().ravel()
        ))

        checks = [
            train_ok, test_ok,
            y_train_ok, y_test_ok,
            variance_ok, support_ok,
            state_ok
        ]

        record.update({
            "Status": (
                "PASS" if all(checks) else "REVIEW"
            ),
            "Train_Match": train_ok,
            "Test_Match": test_ok,
            "Y_Train_Match": y_train_ok,
            "Y_Test_Match": y_test_ok,
            "Variance_Match": variance_ok,
            "Support_Match": support_ok,
            "Fitted_State_Match": state_ok,
            "Train_Max_Diff": train_diff,
            "Test_Max_Diff": test_diff,
            "State_Max_Diff": max_state_diff
        })

    except Exception as exc:
        record["Error"] = (
            f"{type(exc).__name__}: {exc}"
        )

    results.append(record)

results_df = pd.DataFrame(results)

report_path = OUT / f"{RUN_ID}_remaining_reconstruction.csv"
results_df.to_csv(report_path, index=False)

print("=" * 72)
print("CFIS — RECONSTRUÇÃO DOS DATASETS RESTANTES")
print("=" * 72)

print("\n1. RESULTADOS")
print(results_df.to_string(index=False))

print("\n2. RESUMO")
print("Datasets verificados:", len(results_df))
print(
    "Aprovados:",
    int((results_df["Status"] == "PASS").sum())
)
print(
    "Requerem revisão:",
    int((results_df["Status"] != "PASS").sum())
)

print("\n3. CONCLUSÃO")
if (results_df["Status"] == "PASS").all():
    print(
        "Todos os nove datasets foram reproduzidos "
        "por transformações ajustadas somente no treino."
    )
    print(
        "A auditoria de reprodução dos 12 datasets "
        "pode ser consolidada."
    )
else:
    print(
        "Existem divergências que precisam ser "
        "investigadas antes da consolidação."
    )

print("\nRelatório:", report_path)
print("\nCÉLULA 29 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 30
# INVENTÁRIO E AUDITORIA DOS MÉTODOS COMPARATIVOS
# Somente leitura dos resultados históricos.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import pandas as pd

ROOT = Path(PROJECT_DIR)
RANKINGS = ROOT / "02_Rankings"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

METHODS = {
    "CFIS": ["CFIS"],
    "Mutual_Information": ["Mutual_Information"],
    "ANOVA_or_F_Regression": ["ANOVA", "F_Regression"],
    "mRMR": ["mRMR"],
    "L1": ["L1_LASSO", "L1"],
    "CFS": ["CFS"],
    "RFE": ["RFE"],
    "PFI": ["PFI", "Permutation_Importance"],
    "TreeSHAP": ["TreeSHAP", "SHAP"],
    "MDI": ["MDI"],
    "Random_TopK": ["Random_TopK"]
}

DATASETS = sorted(prepared_datasets.keys())

def match_method(relative_path, aliases):
    parts = relative_path.parts
    return any(
        alias.lower() == part.lower()
        for alias in aliases
        for part in parts
    )

def safe_metadata(path):
    try:
        with open(path, encoding="utf-8") as f:
            return json.load(f)
    except (OSError, ValueError) as exc:
        return {"_read_error": str(exc)}

def flatten(obj, prefix=""):
    if isinstance(obj, dict):
        result = {}
        for key, value in obj.items():
            key_path = (
                f"{prefix}.{key}" if prefix else str(key)
            )
            result.update(flatten(value, key_path))
        return result
    return {prefix: obj}

files = sorted(
    p for p in RANKINGS.rglob("*") if p.is_file()
)

inventory = []
metadata_rows = []
coverage = []

for method, aliases in METHODS.items():
    method_files = [
        p for p in files
        if match_method(
            p.relative_to(RANKINGS), aliases
        )
    ]

    for dataset in DATASETS:
        matching = [
            p for p in method_files
            if dataset.lower() in str(
                p.relative_to(RANKINGS)
            ).lower()
        ]

        ranking_files = [
            p for p in matching
            if p.suffix.lower() in {
                ".csv", ".parquet", ".npy"
            }
        ]

        metadata_files = [
            p for p in matching
            if p.suffix.lower() == ".json"
        ]

        coverage.append({
            "Method": method,
            "Dataset": dataset,
            "Ranking_Files": len(ranking_files),
            "Metadata_Files": len(metadata_files),
            "Historical_Artifacts_Present":
                bool(ranking_files)
        })

        for path in matching:
            inventory.append({
                "Method": method,
                "Dataset": dataset,
                "File": str(
                    path.relative_to(RANKINGS)
                ),
                "Extension": path.suffix.lower(),
                "Size_KiB": round(
                    path.stat().st_size / 1024, 2
                )
            })

        for path in metadata_files:
            obj = safe_metadata(path)
            for key, value in flatten(obj).items():
                metadata_rows.append({
                    "Method": method,
                    "Dataset": dataset,
                    "File": str(
                        path.relative_to(RANKINGS)
                    ),
                    "Key": key,
                    "Value": str(value)[:500]
                })

coverage_df = pd.DataFrame(coverage)
inventory_df = pd.DataFrame(inventory)
metadata_df = pd.DataFrame(metadata_rows)

coverage_path = OUT / f"{RUN_ID}_method_coverage.csv"
inventory_path = OUT / f"{RUN_ID}_method_files.csv"
metadata_path = OUT / f"{RUN_ID}_method_metadata.csv"

coverage_df.to_csv(coverage_path, index=False)
inventory_df.to_csv(inventory_path, index=False)
metadata_df.to_csv(metadata_path, index=False)

print("=" * 72)
print("CFIS — AUDITORIA DOS MÉTODOS COMPARATIVOS")
print("=" * 72)

print("\n1. COBERTURA HISTÓRICA")
matrix = coverage_df.pivot(
    index="Method",
    columns="Dataset",
    values="Historical_Artifacts_Present"
).fillna(False)

print(
    matrix.astype(int).to_string()
)

print("\n2. RESUMO POR MÉTODO")
summary = coverage_df.groupby("Method").agg(
    Datasets_With_Rankings=(
        "Historical_Artifacts_Present", "sum"
    ),
    Ranking_Files=("Ranking_Files", "sum"),
    Metadata_Files=("Metadata_Files", "sum")
)
print(summary.to_string())

print("\n3. CAMPOS DE METADADOS RELEVANTES")
terms = (
    "seed", "random", "repeat", "model",
    "parameter", "version", "scor",
    "time", "method", "implementation",
    "train", "feature", "split"
)

if not metadata_df.empty:
    relevant = metadata_df[
        metadata_df["Key"].str.lower().apply(
            lambda s: any(t in s for t in terms)
        )
    ]

    for method in METHODS:
        subset = relevant[
            relevant["Method"] == method
        ].head(8)

        if not subset.empty:
            print(f"\n{method}:")
            print(
                subset[
                    ["Dataset", "Key", "Value"]
                ].to_string(
                    index=False,
                    max_colwidth=100
                )
            )
else:
    print("Nenhum metadado localizado.")

print("\n4. DIRETÓRIOS DE RANKING")
if RANKINGS.exists():
    for folder in sorted(RANKINGS.iterdir()):
        if folder.is_dir():
            print("-", folder.name)

print("\n5. LIMITAÇÃO")
print(
    "A presença de rankings históricos não "
    "comprova equivalência de protocolo."
)
print(
    "Não foram executados seletores, "
    "modelos ou treinamentos."
)

print("\nRelatórios:")
print(coverage_path)
print(inventory_path)
print(metadata_path)
print("\nCÉLULA 30 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 31
# INTEGRIDADE DOS RANKINGS E INVENTÁRIO DOS MODELOS
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import re
import numpy as np
import pandas as pd

ROOT = Path(PROJECT_DIR)
RANKINGS = ROOT / "02_Rankings"
MODELS = ROOT / "10_Models"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

DATASETS = sorted(prepared_datasets)

METHOD_DIRS = {
    "CFIS": "CFIS",
    "Mutual_Information": "Mutual_Information",
    "ANOVA_or_F_Regression": "ANOVA",
    "mRMR": "mRMR",
    "L1": "L1_LASSO",
    "CFS": "CFS",
    "TreeSHAP": "TreeSHAP",
    "Random_TopK": "Random_TopK"
}

ranking_rows = []
model_rows = []
errors = []

for method, dirname in METHOD_DIRS.items():
    folder = RANKINGS / dirname

    for name in DATASETS:
        candidates = [
            p for p in folder.rglob("*")
            if p.is_file()
            and p.suffix.lower() in (".csv", ".parquet")
            and name.lower() in str(
                p.relative_to(folder)
            ).lower()
        ] if folder.exists() else []

        # Limitar a inspeção para evitar centenas
        # de impressões do Random Top-K.
        for path in sorted(candidates):
            record = {
                "Method": method,
                "Dataset": name,
                "File": str(path.relative_to(ROOT)),
                "Rows": None,
                "Columns": "",
                "Missing_Values": None,
                "Duplicate_Features": None,
                "Expected_Features": len(
                    prepared_datasets[name]["feature_names"]
                ),
                "Error": ""
            }

            try:
                if path.suffix.lower() == ".csv":
                    df = pd.read_csv(path)
                else:
                    df = pd.read_parquet(path)

                record["Rows"] = len(df)
                record["Columns"] = ", ".join(
                    map(str, df.columns)
                )[:400]
                record["Missing_Values"] = int(
                    df.isna().sum().sum()
                )

                feature_candidates = [
                    c for c in df.columns
                    if str(c).lower() in (
                        "feature", "feature_name",
                        "feature_names"
                    )
                ]

                if feature_candidates:
                    record["Duplicate_Features"] = int(
                        df[
                            feature_candidates[0]
                        ].duplicated().sum()
                    )

            except Exception as exc:
                record["Error"] = (
                    f"{type(exc).__name__}: {exc}"
                )
                errors.append(record.copy())

            ranking_rows.append(record)

# Apenas inventariar os arquivos dos modelos.
# Não desserializar modelos nesta célula.
if MODELS.exists():
    for path in sorted(MODELS.rglob("*")):
        if not path.is_file():
            continue

        if path.suffix.lower() not in (
            ".joblib", ".pkl", ".pickle"
        ):
            continue

        rel = str(path.relative_to(ROOT))
        basename = path.stem.lower()

        if re.search(
            r"(^|[_\-])(rf|randomforest|random_forest)"
            r"($|[_\-])",
            basename
        ):
            model_type = "RandomForest"
        elif re.search(
            r"(^|[_\-])(dt|decisiontree|decision_tree)"
            r"($|[_\-])",
            basename
        ):
            model_type = "DecisionTree"
        else:
            model_type = "UNKNOWN"

        matched = [
            name for name in DATASETS
            if name.lower() in rel.lower()
        ]

        model_rows.append({
            "Dataset": (
                ", ".join(matched)
                if matched else "UNMATCHED"
            ),
            "Model_Type_Inferred": model_type,
            "File": rel,
            "Size_MiB": round(
                path.stat().st_size / 1024**2, 3
            )
        })

rankings_df = pd.DataFrame(ranking_rows)
models_df = pd.DataFrame(
    model_rows,
    columns=[
        "Dataset", "Model_Type_Inferred",
        "File", "Size_MiB"
    ]
)

rankings_path = OUT / f"{RUN_ID}_ranking_integrity.csv"
models_path = OUT / f"{RUN_ID}_model_inventory.csv"

rankings_df.to_csv(rankings_path, index=False)
models_df.to_csv(models_path, index=False)

print("=" * 72)
print("CFIS — AUDITORIA DE RANKINGS E MODELOS")
print("=" * 72)

print("\n1. FORMATOS DOS RANKINGS")
if not rankings_df.empty:
    summary = rankings_df.groupby("Method").agg(
        Files=("File", "count"),
        Min_Rows=("Rows", "min"),
        Max_Rows=("Rows", "max"),
        Files_With_Errors=(
            "Error",
            lambda s: int((s != "").sum())
        )
    )
    print(summary.to_string())
else:
    print("Nenhum ranking encontrado.")

print("\n2. COLUNAS POR MÉTODO")
for method in METHOD_DIRS:
    subset = rankings_df[
        rankings_df["Method"] == method
    ]
    if subset.empty:
        continue

    print(f"\n{method}:")
    print(
        subset[
            ["Dataset", "Rows", "Columns"]
        ].drop_duplicates(
            subset=["Columns"]
        ).head(5).to_string(
            index=False,
            max_colwidth=120
        )
    )

print("\n3. ARQUIVOS COM POSSÍVEIS PROBLEMAS")
if not rankings_df.empty:
    issues = rankings_df[
        (rankings_df["Error"] != "")
        | (rankings_df["Missing_Values"].fillna(0) > 0)
        | (
            rankings_df[
                "Duplicate_Features"
            ].fillna(0) > 0
        )
    ]
    if issues.empty:
        print("Nenhum problema estrutural identificado.")
    else:
        print(
            issues[
                [
                    "Method", "Dataset", "File",
                    "Missing_Values",
                    "Duplicate_Features", "Error"
                ]
            ].head(25).to_string(
                index=False,
                max_colwidth=100
            )
        )

print("\n4. INVENTÁRIO DOS MODELOS")
if models_df.empty:
    print("Nenhum modelo serializado encontrado.")
else:
    print(
        models_df.to_string(
            index=False,
            max_colwidth=100
        )
    )

    print("\nModelos por tipo inferido:")
    print(
        models_df.groupby(
            "Model_Type_Inferred"
        ).size().to_string()
    )

print("\n5. LIMITAÇÕES")
print(
    "A integridade estrutural não demonstra "
    "equivalência dos protocolos."
)
print(
    "Os tipos de modelo foram inferidos dos "
    "nomes dos arquivos, sem desserialização."
)
print(
    "A análise de valores e ordenação dos "
    "rankings será feita separadamente."
)

print("\nRelatórios:")
print(rankings_path)
print(models_path)
print("\nCÉLULA 31 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 32
# AUDITORIA DOS MODELOS RANDOM FOREST HISTÓRICOS
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import joblib
import json
import numpy as np
import pandas as pd

from sklearn.ensemble import (
    RandomForestClassifier,
    RandomForestRegressor
)
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    r2_score,
    mean_squared_error
)

ROOT = Path(PROJECT_DIR)
MODEL_DIR = ROOT / "10_Models" / "External_Baselines"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

rows = []
errors = []

for name in sorted(prepared_datasets):
    record = {
        "Dataset": name,
        "Status": "ERROR",
        "Model_Class": "",
        "Task": "",
        "N_Features_Model": np.nan,
        "N_Features_Data": np.nan,
        "N_Estimators": np.nan,
        "Max_Depth": "",
        "Min_Samples_Leaf": np.nan,
        "Max_Features": "",
        "Random_State": "",
        "Bootstrap": "",
        "OOB_Score_Enabled": "",
        "Train_Metric": np.nan,
        "Test_Metric": np.nan,
        "Metric_Name": "",
        "Feature_Names_Match": "",
        "Error": ""
    }

    try:
        data = prepared_datasets[name]
        X_train = data["X_train"]
        X_test = data["X_test"]
        y_train = np.asarray(data["y_train"]).ravel()
        y_test = np.asarray(data["y_test"]).ravel()

        path = MODEL_DIR / f"{name}_RF.joblib"

        if not path.exists():
            raise FileNotFoundError(str(path))

        # Carregar somente modelos históricos
        # produzidos pelo próprio projeto.
        model = joblib.load(path)

        if not isinstance(
            model,
            (RandomForestClassifier, RandomForestRegressor)
        ):
            raise TypeError(
                "Objeto inesperado: "
                f"{type(model).__name__}"
            )

        task = data["task"]
        expected_class = (
            RandomForestClassifier
            if task == "classification"
            else RandomForestRegressor
        )

        if not isinstance(model, expected_class):
            raise TypeError(
                "Tipo do modelo incompatível "
                "com a tarefa."
            )

        n_features = int(model.n_features_in_)
        dimensions_match = (
            n_features == X_train.shape[1]
            and n_features == X_test.shape[1]
        )

        if not dimensions_match:
            raise ValueError(
                "Dimensionalidade incompatível."
            )

        feature_names_match = "NOT_AVAILABLE"
        if hasattr(model, "feature_names_in_"):
            feature_names_match = str(
                np.array_equal(
                    model.feature_names_in_,
                    X_train.columns.to_numpy()
                )
            )

            if feature_names_match != "True":
                raise ValueError(
                    "Nomes/ordem de features incompatíveis."
                )

        pred_train = model.predict(X_train)
        pred_test = model.predict(X_test)

        if task == "classification":
            train_metric = f1_score(
                y_train, pred_train,
                average="macro",
                zero_division=0
            )
            test_metric = f1_score(
                y_test, pred_test,
                average="macro",
                zero_division=0
            )
            metric_name = "Macro-F1"
        else:
            train_metric = r2_score(
                y_train, pred_train
            )
            test_metric = r2_score(
                y_test, pred_test
            )
            metric_name = "R2"

        params = model.get_params(deep=False)

        record.update({
            "Status": "PASS",
            "Model_Class": type(model).__name__,
            "Task": task,
            "N_Features_Model": n_features,
            "N_Features_Data": X_train.shape[1],
            "N_Estimators": params["n_estimators"],
            "Max_Depth": str(params["max_depth"]),
            "Min_Samples_Leaf": params["min_samples_leaf"],
            "Max_Features": str(params["max_features"]),
            "Random_State": str(params["random_state"]),
            "Bootstrap": str(params["bootstrap"]),
            "OOB_Score_Enabled": str(params["oob_score"]),
            "Train_Metric": float(train_metric),
            "Test_Metric": float(test_metric),
            "Metric_Name": metric_name,
            "Feature_Names_Match": feature_names_match
        })

    except Exception as exc:
        record["Error"] = (
            f"{type(exc).__name__}: {exc}"
        )
        errors.append({
            "Dataset": name,
            "Error": record["Error"]
        })

    rows.append(record)

results_df = pd.DataFrame(rows)

report_path = OUT / f"{RUN_ID}_rf_model_audit.csv"
results_df.to_csv(report_path, index=False)

print("=" * 72)
print("CFIS — AUDITORIA DOS MODELOS RANDOM FOREST")
print("=" * 72)

print("\n1. INTEGRIDADE E PARÂMETROS")
print(
    results_df[
        [
            "Dataset", "Status", "Model_Class",
            "N_Features_Model", "N_Estimators",
            "Max_Depth", "Min_Samples_Leaf",
            "Max_Features", "Random_State",
            "Feature_Names_Match"
        ]
    ].to_string(index=False)
)

print("\n2. DESEMPENHO HISTÓRICO DOS MODELOS")
print(
    results_df[
        [
            "Dataset", "Metric_Name",
            "Train_Metric", "Test_Metric"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

print("\n3. RESUMO")
print(
    "Modelos aprovados:",
    int((results_df["Status"] == "PASS").sum()),
    "/",
    len(results_df)
)

if errors:
    print("\n4. ERROS")
    for item in errors:
        print(item["Dataset"], ":", item["Error"])

print("\n5. LIMITAÇÕES")
print(
    "Compatibilidade dimensional e desempenho "
    "não comprovam a origem do treinamento."
)
print(
    "A equivalência do protocolo e a ausência "
    "de exposição ao teste ainda precisam "
    "ser verificadas."
)
print(
    "Nenhum modelo foi treinado nesta célula."
)

print("\nRelatório:", report_path)
print("\nCÉLULA 32 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 33
# REPRODUÇÃO INDEPENDENTE DAS RANDOM FOREST
# ============================================================

import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from datetime import datetime, timezone
from sklearn.base import clone

ROOT = Path(PROJECT_DIR)
MODEL_DIR = ROOT / "10_Models" / "External_Baselines"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

RUN_ID = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

def compare_arrays(a, b, atol=1e-12):
    a = np.asarray(a)
    b = np.asarray(b)

    if a.shape != b.shape:
        return False, np.nan

    if a.size == 0:
        return True, 0.0

    if np.issubdtype(a.dtype, np.number) and \
       np.issubdtype(b.dtype, np.number):

        a = a.astype(np.float64)
        b = b.astype(np.float64)

        difference = np.abs(a - b)

        return (
            bool(np.allclose(
                a, b,
                rtol=1e-12,
                atol=atol,
                equal_nan=True
            )),
            float(np.nanmax(difference))
        )

    return bool(np.array_equal(a, b)), np.nan

rows = []

for name in sorted(prepared_datasets):
    record = {
        "Dataset": name,
        "Status": "ERROR",
        "Pred_Train_Match": False,
        "Pred_Test_Match": False,
        "Prob_Train_Match": None,
        "Prob_Test_Match": None,
        "MDI_Match": False,
        "MDI_Max_Diff": np.nan,
        "Pred_Train_Max_Diff": np.nan,
        "Pred_Test_Max_Diff": np.nan,
        "Prob_Train_Max_Diff": np.nan,
        "Prob_Test_Max_Diff": np.nan,
        "Error": ""
    }

    try:
        data = prepared_datasets[name]
        X_train = data["X_train"]
        X_test = data["X_test"]
        y_train = np.asarray(data["y_train"]).ravel()

        historical = joblib.load(
            MODEL_DIR / f"{name}_RF.joblib"
        )

        # clone preserva os hiperparâmetros e
        # descarta todo o estado previamente ajustado.
        reconstructed = clone(historical)

        # Somente dados de treinamento.
        reconstructed.fit(X_train, y_train)

        historical_train = historical.predict(X_train)
        rebuilt_train = reconstructed.predict(X_train)

        historical_test = historical.predict(X_test)
        rebuilt_test = reconstructed.predict(X_test)

        train_ok, train_diff = compare_arrays(
            historical_train, rebuilt_train
        )
        test_ok, test_diff = compare_arrays(
            historical_test, rebuilt_test
        )

        mdi_ok, mdi_diff = compare_arrays(
            historical.feature_importances_,
            reconstructed.feature_importances_
        )

        prob_train_ok = None
        prob_test_ok = None
        prob_train_diff = np.nan
        prob_test_diff = np.nan

        if hasattr(historical, "predict_proba"):
            prob_train_ok, prob_train_diff = compare_arrays(
                historical.predict_proba(X_train),
                reconstructed.predict_proba(X_train)
            )
            prob_test_ok, prob_test_diff = compare_arrays(
                historical.predict_proba(X_test),
                reconstructed.predict_proba(X_test)
            )

        checks = [
            train_ok,
            test_ok,
            mdi_ok
        ]

        if prob_train_ok is not None:
            checks.extend([
                prob_train_ok,
                prob_test_ok
            ])

        record.update({
            "Status": (
                "PASS" if all(checks) else "REVIEW"
            ),
            "Pred_Train_Match": train_ok,
            "Pred_Test_Match": test_ok,
            "Prob_Train_Match": prob_train_ok,
            "Prob_Test_Match": prob_test_ok,
            "MDI_Match": mdi_ok,
            "MDI_Max_Diff": mdi_diff,
            "Pred_Train_Max_Diff": train_diff,
            "Pred_Test_Max_Diff": test_diff,
            "Prob_Train_Max_Diff": prob_train_diff,
            "Prob_Test_Max_Diff": prob_test_diff
        })

    except Exception as exc:
        record["Error"] = (
            f"{type(exc).__name__}: {exc}"
        )

    rows.append(record)

results_df = pd.DataFrame(rows)

report_path = OUT / f"{RUN_ID}_rf_reproduction.csv"
results_df.to_csv(report_path, index=False)

print("=" * 72)
print("CFIS — REPRODUÇÃO DAS RANDOM FOREST")
print("=" * 72)

print("\n1. RESULTADOS")
print(
    results_df[
        [
            "Dataset", "Status",
            "Pred_Train_Match",
            "Pred_Test_Match",
            "Prob_Train_Match",
            "Prob_Test_Match",
            "MDI_Match",
            "MDI_Max_Diff"
        ]
    ].to_string(index=False)
)

print("\n2. RESUMO")
print(
    "Aprovados:",
    int((results_df["Status"] == "PASS").sum()),
    "/",
    len(results_df)
)

reviews = results_df[
    results_df["Status"] != "PASS"
]

if not reviews.empty:
    print("\n3. CASOS PARA REVISÃO")
    print(
        reviews[
            [
                "Dataset", "Status",
                "Pred_Train_Max_Diff",
                "Pred_Test_Max_Diff",
                "MDI_Max_Diff",
                "Error"
            ]
        ].to_string(index=False)
    )

print("\n4. INTERPRETAÇÃO")
print(
    "Coincidência das previsões e MDI oferece "
    "evidência de reprodutibilidade do modelo "
    "a partir do treinamento histórico."
)
print(
    "Não comprova por si só o histórico de "
    "execução do treinamento original."
)
print(
    "Os dados de teste foram usados apenas "
    "para comparação das previsões."
)

print("\nRelatório:", report_path)
print("\nCÉLULA 33 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 34
# MANIFESTO DO BENCHMARK FINAL E CHECKPOINTS
# Não executa modelos nem seletores.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import itertools
import json
import os
import pandas as pd

PROJECT_DIR = Path(
    "/content/drive/MyDrive/CFIS_Artigo_Diogo"
)

ROOT = (
    PROJECT_DIR / "04_Efficiency"
    / "CFIS_Final_Benchmark"
)

PROTOCOL_DIR = ROOT / "00_Protocol"
EXEC_DIR = ROOT / "03_Final_Execution"
MANIFEST_DIR = EXEC_DIR / "Manifests"
CHECKPOINT_DIR = EXEC_DIR / "Checkpoints"
RESULT_DIR = EXEC_DIR / "Results"
ERROR_DIR = EXEC_DIR / "Errors"

for directory in (
    PROTOCOL_DIR, MANIFEST_DIR,
    CHECKPOINT_DIR, RESULT_DIR, ERROR_DIR
):
    directory.mkdir(parents=True, exist_ok=True)

SEED = 42
CFIS_N_REPEATS = 10
STABILITY_SPLITS = 10

K_VALUES = {
    "20_Newsgroups": [3, 5, 10, 20, 50, 125, 250, 375],
    "Breast_Cancer": [3, 5, 8, 10, 15, 20, 23],
    "California_Housing": [1, 2, 3, 4, 5, 6],
    "Credit_Approval": [3, 5, 7, 10, 16, 20, 31, 46],
    "Diabetes_Reg": [1, 3, 5, 8, 10],
    "Gisette": [3, 5, 10, 20, 47, 117, 234, 351],
    "Iris": [1, 2, 3],
    "Pima_Diabetes": [1, 2, 3, 4, 5, 6],
    "Synthetic_120": [3, 5, 10, 12, 20, 30, 60, 90],
    "Synthetic_30": [3, 5, 8, 10, 15, 20, 23],
    "Synthetic_60": [3, 5, 6, 10, 15, 20, 30, 45],
    "Wine": [2, 3, 4, 5, 7, 10]
}

METHODS = [
    "CFIS",
    "Mutual_Information",
    "ANOVA_or_F_Regression",
    "mRMR",
    "L1",
    "CFS",
    "RFE",
    "PFI",
    "TreeSHAP",
    "MDI",
    "Random_TopK"
]

MODELS = ["DecisionTree", "RandomForest"]

if "prepared_datasets" not in globals():
    raise RuntimeError(
        "prepared_datasets não está disponível. "
        "Execute primeiro as células de preparação."
    )

available = set(prepared_datasets)
expected = set(K_VALUES)

if available != expected:
    raise RuntimeError(
        f"Datasets faltantes: {sorted(expected - available)}; "
        f"inesperados: {sorted(available - expected)}"
    )

for name, ks in K_VALUES.items():
    n_features = len(
        prepared_datasets[name]["feature_names"]
    )
    if not all(1 <= k <= n_features for k in ks):
        raise ValueError(
            f"Top-K inválido para {name}: "
            f"{ks}; features={n_features}"
        )

# Manifesto das avaliações externas.
# Uma linha representa uma combinação de
# dataset, seletor, modelo e Top-K.
rows = []

for name, method, model in itertools.product(
    sorted(K_VALUES), METHODS, MODELS
):
    for k in K_VALUES[name]:
        task = prepared_datasets[name]["task"]

        key = f"{name}|{method}|{model}|{k}|{SEED}"
        job_id = hashlib.sha256(
            key.encode("utf-8")
        ).hexdigest()[:20]

        rows.append({
            "job_id": job_id,
            "dataset": name,
            "task": task,
            "method": method,
            "model": model,
            "top_k": k,
            "seed": SEED,
            "status": "PENDING",
            "evaluation": "outer_holdout",
            "checkpoint": str(
                CHECKPOINT_DIR / f"{job_id}.json"
            )
        })

manifest = pd.DataFrame(rows)

if manifest["job_id"].duplicated().any():
    raise RuntimeError(
        "Foram encontrados IDs de experimento duplicados."
    )

# Verificações pendentes antes da execução.
pending_decisions = [
    "Implementações e versões dos 11 seletores",
    "Hiperparâmetros finais de DecisionTree e RandomForest",
    "Estratégia de avaliação interna e estabilidade",
    "Reajuste do pré-processamento em cada divisão interna",
    "Agregação TreeSHAP multiclasse e regressão",
    "Definição exata de PFI e MDI por modelo",
    "Desempates, CFS e RFE",
    "Repetições e agregação de Random_TopK",
    "Medição isolada de tempo e memória",
    "Regra de reutilização de rankings históricos",
    "Política de uso único do teste externo"
]

protocol = {
    "status": "PENDING_IMPLEMENTATION_VALIDATION",
    "seed": SEED,
    "cfis_n_repeats": CFIS_N_REPEATS,
    "stability_splits": STABILITY_SPLITS,
    "datasets": sorted(K_VALUES),
    "methods": METHODS,
    "models": MODELS,
    "top_k": K_VALUES,
    "pending_decisions": pending_decisions,
    "outer_test_policy": (
        "Não usar o teste externo para ajustar "
        "seletores, hiperparâmetros ou escolher K."
    ),
    "historical_policy": (
        "Não sobrescrever resultados históricos. "
        "Reutilização condicionada à equivalência "
        "do protocolo e à ausência de vazamento."
    )
}

# Arquivos imutáveis por identificador de execução.
timestamp = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

manifest_path = (
    MANIFEST_DIR / f"{timestamp}_manifest.csv"
)
protocol_path = (
    MANIFEST_DIR / f"{timestamp}_protocol.json"
)

with open(protocol_path, "x", encoding="utf-8") as f:
    json.dump(
        protocol, f,
        ensure_ascii=False,
        indent=2
    )

manifest.to_csv(
    manifest_path,
    index=False,
    mode="x"
)

print("=" * 72)
print("CFIS — BENCHMARK FINAL: MANIFESTO")
print("=" * 72)

print("\n1. ESCOPO")
print("Datasets:", len(K_VALUES))
print("Métodos:", len(METHODS))
print("Modelos:", len(MODELS))
print("Combinações externas:", len(manifest))
print("Divisões internas previstas:", STABILITY_SPLITS)

print("\n2. COMBINAÇÕES POR DATASET")
print(
    manifest.groupby("dataset")
    .size()
    .rename("Experimentos")
    .to_string()
)

print("\n3. COMBINAÇÕES POR MÉTODO")
print(
    manifest.groupby("method")
    .size()
    .rename("Experimentos")
    .to_string()
)

print("\n4. CONTROLE")
print("Semente:", SEED)
print("CFIS n_repeats:", CFIS_N_REPEATS)
print("Status:", protocol["status"])
print("Treinamentos executados: 0")
print("Arquivos históricos modificados: 0")

print("\n5. PENDÊNCIAS PARA LIBERAR EXECUÇÃO")
for i, decision in enumerate(
    pending_decisions, start=1
):
    print(f"{i:02d}. {decision}")

print("\n6. ARQUIVOS")
print("Manifesto:", manifest_path)
print("Protocolo:", protocol_path)
print("Checkpoints:", CHECKPOINT_DIR)
print("Resultados:", RESULT_DIR)

print("\nCÉLULA 34 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 35
# CONSOLIDAÇÃO DOS PARÂMETROS HISTÓRICOS
# Apenas leitura dos experimentos anteriores.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from collections import defaultdict
import json
import joblib
import pandas as pd
import sklearn

PROJECT_DIR = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
RANKINGS = PROJECT_DIR / "02_Rankings"
MODELS_DIR = PROJECT_DIR / "10_Models" / "External_Baselines"
OUT = (
    PROJECT_DIR / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

STAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

METHOD_DIRS = {
    "CFIS": "CFIS",
    "Mutual_Information": "Mutual_Information",
    "ANOVA_or_F_Regression": "ANOVA",
    "mRMR": "mRMR",
    "L1": "L1_LASSO",
    "CFS": "CFS",
    "TreeSHAP": "TreeSHAP",
    "Random_TopK": "Random_TopK",
}

DATASETS = sorted(prepared_datasets)

# Campos relevantes para comparar protocolos.
KEY_TERMS = (
    "method", "version", "seed", "random",
    "repeat", "parameter", "alpha", "scoring",
    "model", "estimator", "criterion",
    "max_depth", "min_samples", "max_features",
    "bootstrap", "perturbation", "aggregation",
    "background", "train_only", "discrete",
    "rebuild", "selection", "class_weight",
    "solver", "normaliz", "standard",
    "fit_intercept", "test", "split"
)

def flatten_dict(obj, prefix=""):
    result = {}
    if isinstance(obj, dict):
        for k, v in obj.items():
            path = f"{prefix}.{k}" if prefix else str(k)
            result.update(flatten_dict(v, path))
    else:
        result[prefix] = obj
    return result

metadata_rows = []
issues = []

for method, folder_name in METHOD_DIRS.items():
    folder = RANKINGS / folder_name

    for dataset in DATASETS:
        if not folder.exists():
            issues.append({
                "Method": method,
                "Dataset": dataset,
                "Issue": "Diretório inexistente"
            })
            continue

        candidates = sorted(
            p for p in folder.rglob("*.json")
            if dataset.lower() in str(
                p.relative_to(folder)
            ).lower()
        )

        if not candidates:
            issues.append({
                "Method": method,
                "Dataset": dataset,
                "Issue": "Metadados não localizados"
            })

        for path in candidates:
            try:
                with open(path, encoding="utf-8") as f:
                    obj = json.load(f)

                for key, value in flatten_dict(obj).items():
                    if any(
                        term in key.lower()
                        for term in KEY_TERMS
                    ):
                        metadata_rows.append({
                            "Method": method,
                            "Dataset": dataset,
                            "Key": key,
                            "Value": json.dumps(
                                value,
                                ensure_ascii=False,
                                default=str
                            ),
                            "File": str(
                                path.relative_to(PROJECT_DIR)
                            )
                        })
            except Exception as exc:
                issues.append({
                    "Method": method,
                    "Dataset": dataset,
                    "Issue": (
                        f"{type(exc).__name__}: {exc}"
                    )
                })

# Auditar os parâmetros efetivos dos modelos RF.
rf_rows = []

for dataset in DATASETS:
    path = MODELS_DIR / f"{dataset}_RF.joblib"

    try:
        model = joblib.load(path)
        params = model.get_params(deep=False)

        for key, value in sorted(params.items()):
            rf_rows.append({
                "Dataset": dataset,
                "Parameter": key,
                "Value": repr(value),
                "Model_Class": type(model).__name__
            })
    except Exception as exc:
        issues.append({
            "Method": "RandomForest",
            "Dataset": dataset,
            "Issue": (
                f"{type(exc).__name__}: {exc}"
            )
        })

metadata_df = pd.DataFrame(
    metadata_rows,
    columns=[
        "Method", "Dataset", "Key",
        "Value", "File"
    ]
)

rf_df = pd.DataFrame(
    rf_rows,
    columns=[
        "Dataset", "Parameter",
        "Value", "Model_Class"
    ]
)

issues_df = pd.DataFrame(
    issues,
    columns=["Method", "Dataset", "Issue"]
)

# Detectar campos cujos valores diferem entre datasets.
variation_rows = []

if not metadata_df.empty:
    grouped = metadata_df.groupby(
        ["Method", "Key"], dropna=False
    )

    for (method, key), group in grouped:
        values = sorted(group["Value"].unique())
        variation_rows.append({
            "Method": method,
            "Key": key,
            "Datasets": group["Dataset"].nunique(),
            "Distinct_Values": len(values),
            "Examples": " | ".join(values[:3])[:300]
        })

variations_df = pd.DataFrame(
    variation_rows,
    columns=[
        "Method", "Key", "Datasets",
        "Distinct_Values", "Examples"
    ]
)

for label, df in (
    ("metadata", metadata_df),
    ("rf_parameters", rf_df),
    ("variations", variations_df),
    ("issues", issues_df)
):
    df.to_csv(
        OUT / f"{STAMP}_{label}.csv",
        index=False
    )

print("=" * 72)
print("CFIS — PARÂMETROS HISTÓRICOS")
print("=" * 72)

print("\n1. VERSÕES DO AMBIENTE")
print("scikit-learn:", sklearn.__version__)

try:
    import shap
    print("SHAP:", shap.__version__)
except ImportError:
    print("SHAP: não instalado")

print("\n2. METADADOS POR MÉTODO")
if not metadata_df.empty:
    print(
        metadata_df.groupby("Method")
        .agg(
            Datasets=("Dataset", "nunique"),
            Fields=("Key", "nunique")
        )
        .to_string()
    )

print("\n3. PARÂMETROS E DEFINIÇÕES PRINCIPAIS")
important = (
    "method", "version", "seed",
    "random_state", "n_repeats",
    "scoring", "solver", "alpha",
    "parameters.C", "parameters.alpha",
    "model", "feature_perturbation",
    "model_output", "aggregation",
    "rebuild_version", "train_only"
)

for method in METHOD_DIRS:
    sub = metadata_df[
        metadata_df["Method"] == method
    ]

    if sub.empty:
        continue

    print(f"\n{method}")

    selected = sub[
        sub["Key"].apply(
            lambda k: any(
                k == x or k.endswith("." + x)
                for x in important
            )
        )
    ]

    if selected.empty:
        print("Sem campos principais identificados.")
        continue

    # Mostrar exemplos distintos, evitando repetir
    # 12 vezes o mesmo parâmetro.
    print(
        selected[
            ["Key", "Value"]
        ].drop_duplicates()
        .head(20)
        .to_string(
            index=False,
            max_colwidth=95
        )
    )

print("\n4. PARÂMETROS RF QUE VARIAM ENTRE DATASETS")
if not rf_df.empty:
    rf_variations = (
        rf_df.groupby("Parameter")["Value"]
        .nunique()
    )
    varying = rf_variations[
        rf_variations > 1
    ].index

    if len(varying):
        print(
            rf_df[
                rf_df["Parameter"].isin(varying)
            ][
                ["Dataset", "Parameter", "Value"]
            ].to_string(
                index=False,
                max_colwidth=80
            )
        )
    else:
        print("Nenhuma variação.")

print("\n5. CAMPOS DE SELETORES QUE VARIAM")
if not variations_df.empty:
    differing = variations_df[
        (variations_df["Distinct_Values"] > 1)
        & (
            ~variations_df["Key"].str.lower()
            .str.contains(
                "time|duration|path|n_train|n_features",
                regex=True
            )
        )
    ]
    print(
        differing[
            [
                "Method", "Key",
                "Distinct_Values", "Examples"
            ]
        ].head(40).to_string(
            index=False,
            max_colwidth=85
        )
        if not differing.empty
        else "Nenhuma variação relevante identificada."
    )

print("\n6. PROBLEMAS DE LEITURA")
if issues_df.empty:
    print("Nenhum.")
else:
    print(issues_df.to_string(index=False))

print("\n7. ESTADO DO BENCHMARK")
print(
    "Configuração ainda pendente de validação; "
    "nenhum treinamento foi iniciado."
)
print("Relatórios salvos em:", OUT)
print("\nCÉLULA 35 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 36
# VALIDAÇÃO SEMÂNTICA DOS RANKINGS HISTÓRICOS
# Somente leitura; não executa seletores nem modelos.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import numpy as np
import pandas as pd

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
RANKINGS = ROOT / "02_Rankings"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

STAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

CONFIG = {
    "CFIS": ("CFIS", "CFIS_Score", "descending"),
    "Mutual_Information": (
        "Mutual_Information", "MI_Score", "descending"
    ),
    "ANOVA_or_F_Regression": (
        "ANOVA", "ANOVA_Score", "descending"
    ),
    "mRMR": ("mRMR", "mRMR_Position", "ascending"),
    "L1": ("L1_LASSO", "L1_LASSO_Score", "descending"),
    "CFS": ("CFS", "CFS_Position", "ascending"),
    "TreeSHAP": (
        "TreeSHAP", "TreeSHAP_Score", "descending"
    ),
    "Random_TopK": (
        "Random_TopK", "Random_Position", "ascending"
    ),
}

def read_ranking(path):
    if path.suffix.lower() == ".csv":
        return pd.read_csv(path)
    if path.suffix.lower() == ".parquet":
        return pd.read_parquet(path)
    raise ValueError(f"Formato não suportado: {path}")

def monotonic_check(values, direction):
    values = np.asarray(values, dtype=float)
    if not np.isfinite(values).all():
        return False
    diff = np.diff(values)
    tolerance = 1e-10
    if direction == "ascending":
        return bool(np.all(diff >= -tolerance))
    return bool(np.all(diff <= tolerance))

rows = []

for method, (folder_name, score_col, direction) in CONFIG.items():
    folder = RANKINGS / folder_name

    for dataset in sorted(prepared_datasets):
        expected_names = list(
            map(
                str,
                prepared_datasets[dataset]["feature_names"]
            )
        )
        expected_set = set(expected_names)

        paths = sorted(
            p for p in folder.rglob("*")
            if p.is_file()
            and p.suffix.lower() in (".csv", ".parquet")
            and dataset.lower() in str(
                p.relative_to(folder)
            ).lower()
        ) if folder.exists() else []

        if not paths:
            rows.append({
                "Method": method,
                "Dataset": dataset,
                "File": "",
                "Status": "MISSING",
                "Expected": len(expected_names),
                "Observed": 0,
                "Missing_Features": len(expected_names),
                "Unknown_Features": 0,
                "Duplicates": 0,
                "Order_Valid": False,
                "Error": "Ranking não encontrado"
            })
            continue

        for path in paths:
            record = {
                "Method": method,
                "Dataset": dataset,
                "File": str(path.relative_to(ROOT)),
                "Status": "ERROR",
                "Expected": len(expected_names),
                "Observed": 0,
                "Missing_Features": np.nan,
                "Unknown_Features": np.nan,
                "Duplicates": np.nan,
                "Order_Valid": False,
                "Error": ""
            }

            try:
                df = read_ranking(path)

                if "Feature" not in df.columns:
                    raise ValueError("Coluna Feature ausente")

                if score_col not in df.columns:
                    raise ValueError(
                        f"Coluna {score_col} ausente"
                    )

                observed = df["Feature"].astype(str).tolist()
                observed_set = set(observed)

                missing = expected_set - observed_set
                unknown = observed_set - expected_set
                duplicates = len(observed) - len(observed_set)

                scores = pd.to_numeric(
                    df[score_col], errors="coerce"
                ).to_numpy(dtype=float)

                order_ok = monotonic_check(
                    scores, direction
                )

                valid = (
                    len(observed) == len(expected_names)
                    and not missing
                    and not unknown
                    and duplicates == 0
                    and order_ok
                )

                record.update({
                    "Status": "PASS" if valid else "REVIEW",
                    "Observed": len(observed),
                    "Missing_Features": len(missing),
                    "Unknown_Features": len(unknown),
                    "Duplicates": duplicates,
                    "Order_Valid": order_ok
                })

                if not valid:
                    record["Error"] = (
                        f"Missing={sorted(missing)[:3]}; "
                        f"Unknown={sorted(unknown)[:3]}"
                    )

            except Exception as exc:
                record["Error"] = (
                    f"{type(exc).__name__}: {exc}"
                )

            rows.append(record)

results = pd.DataFrame(rows)

report_path = OUT / f"{STAMP}_ranking_semantic_audit.csv"
results.to_csv(report_path, index=False)

print("=" * 72)
print("CFIS — VALIDAÇÃO SEMÂNTICA DOS RANKINGS")
print("=" * 72)

print("\n1. RESUMO POR MÉTODO")
summary = results.groupby("Method").agg(
    Files=("File", "count"),
    Approved=("Status", lambda x: int((x == "PASS").sum())),
    Review=("Status", lambda x: int((x == "REVIEW").sum())),
    Errors=("Status", lambda x: int((x == "ERROR").sum())),
    Missing=("Status", lambda x: int((x == "MISSING").sum()))
)
print(summary.to_string())

print("\n2. VERIFICAÇÃO RANDOM TOP-K")
random_rows = results[
    results["Method"] == "Random_TopK"
]
print(
    random_rows.groupby("Dataset")
    .agg(
        Files=("File", "count"),
        Approved=("Status", lambda x: int((x == "PASS").sum()))
    ).to_string()
)

print("\n3. CASOS QUE EXIGEM REVISÃO")
problems = results[results["Status"] != "PASS"]

if problems.empty:
    print("Nenhum problema identificado.")
else:
    print(
        problems[
            [
                "Method", "Dataset", "Status",
                "Missing_Features", "Unknown_Features",
                "Duplicates", "Order_Valid", "Error"
            ]
        ].head(35).to_string(
            index=False,
            max_colwidth=100
        )
    )

print("\n4. RESUMO GERAL")
print("Arquivos examinados:", len(results))
print(
    "Aprovados:",
    int((results["Status"] == "PASS").sum())
)
print(
    "Requerem revisão:",
    int((results["Status"] != "PASS").sum())
)

print("\n5. LIMITAÇÕES")
print(
    "A validação confirma integridade e ordenação "
    "dos rankings, não equivalência estatística "
    "ou ausência de vazamento."
)
print(
    "A reutilização nas divisões internas continua "
    "proibida sem reajuste dos seletores."
)

print("\nRelatório:", report_path)
print("\nCÉLULA 36 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 37
# PILOTO DO EXECUTOR: SELEÇÃO + ÁRVORES + CHECKPOINT
# Apenas treino interno e validação interna.
# NÃO utiliza o teste externo.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import time

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_selection import (
    mutual_info_classif, f_classif
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score,
    f1_score
)

PROJECT_DIR = Path(
    "/content/drive/MyDrive/CFIS_Artigo_Diogo"
)
OUT = (
    PROJECT_DIR / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Pilot"
)
OUT.mkdir(parents=True, exist_ok=True)

DATASET = "Breast_Cancer"
SEED = 42
K_VALUES_PILOT = [3, 5, 8, 10, 15, 20, 23]

data = prepared_datasets[DATASET]

# Usar exclusivamente o treinamento externo.
X = data["X_train"].copy()
y = pd.Series(
    np.asarray(data["y_train"]).ravel(),
    index=X.index
)

assert X.shape[1] == len(data["feature_names"])
assert X.columns.is_unique
assert len(X) == len(y)
assert not X.isna().any().any()

indices = np.arange(len(X))

idx_fit, idx_val = train_test_split(
    indices,
    test_size=0.25,
    random_state=SEED,
    stratify=y.to_numpy()
)

X_fit = X.iloc[idx_fit].copy()
X_val = X.iloc[idx_val].copy()
y_fit = y.iloc[idx_fit].to_numpy()
y_val = y.iloc[idx_val].to_numpy()

# ATENÇÃO:
# Os dados aqui já foram pré-processados no treino
# externo. Este piloto verifica o executor, mas
# NÃO é uma avaliação interna definitiva livre
# de vazamento do pré-processamento.
# Na execução científica definitiva, cada divisão
# deverá reconstruir o pré-processamento usando
# somente os dados brutos de seu treino interno.

selectors = {
    "Mutual_Information": lambda: mutual_info_classif(
        X_fit, y_fit,
        random_state=SEED,
        discrete_features=False
    ),
    "ANOVA": lambda: f_classif(
        X_fit, y_fit
    )[0]
}

models = {
    "DecisionTree": lambda: DecisionTreeClassifier(
        random_state=SEED
    ),
    "RandomForest": lambda: RandomForestClassifier(
        n_estimators=100,
        random_state=SEED,
        n_jobs=1
    )
}

rows = []
rankings = {}

for method, selector in selectors.items():
    start = time.perf_counter()
    scores = np.asarray(
        selector(), dtype=np.float64
    ).ravel()
    selection_seconds = time.perf_counter() - start

    if len(scores) != X_fit.shape[1]:
        raise RuntimeError(
            f"Dimensão inválida para {method}"
        )

    scores = np.nan_to_num(
        scores,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    # Desempate determinístico pela posição original.
    ordered = np.lexsort((
        np.arange(len(scores)),
        -scores
    ))

    rankings[method] = [
        str(X_fit.columns[i]) for i in ordered
    ]

    for k in K_VALUES_PILOT:
        selected = rankings[method][:k]

        for model_name, factory in models.items():
            model = factory()

            t0 = time.perf_counter()
            model.fit(
                X_fit[selected], y_fit
            )
            fit_seconds = (
                time.perf_counter() - t0
            )

            t1 = time.perf_counter()
            predictions = model.predict(
                X_val[selected]
            )
            predict_seconds = (
                time.perf_counter() - t1
            )

            rows.append({
                "Dataset": DATASET,
                "Selector": method,
                "Model": model_name,
                "TopK": k,
                "Accuracy": accuracy_score(
                    y_val, predictions
                ),
                "Balanced_Accuracy":
                    balanced_accuracy_score(
                        y_val, predictions
                    ),
                "Macro_F1": f1_score(
                    y_val, predictions,
                    average="macro",
                    zero_division=0
                ),
                "Selection_Seconds":
                    selection_seconds,
                "Fit_Seconds": fit_seconds,
                "Predict_Seconds":
                    predict_seconds,
                "Total_Seconds":
                    selection_seconds
                    + fit_seconds
                    + predict_seconds,
                "Evaluation": "SMOKE_TEST_ONLY"
            })

results = pd.DataFrame(rows)

stamp = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

csv_path = OUT / f"{stamp}_pilot_results.csv"
json_path = OUT / f"{stamp}_pilot_metadata.json"

results.to_csv(
    csv_path, index=False, mode="x"
)

metadata = {
    "dataset": DATASET,
    "seed": SEED,
    "status": "SMOKE_TEST_ONLY",
    "n_fit": len(X_fit),
    "n_validation": len(X_val),
    "outer_test_used": False,
    "historical_rankings_used": False,
    "inner_preprocessing_refitted": False,
    "scientific_results_eligible": False,
    "selectors": list(selectors),
    "models": list(models),
    "top_k": K_VALUES_PILOT,
    "rankings": rankings
}

with open(
    json_path, "x", encoding="utf-8"
) as f:
    json.dump(
        metadata, f,
        indent=2,
        ensure_ascii=False
    )

print("=" * 72)
print("CFIS — PILOTO DO EXECUTOR")
print("=" * 72)

print("\n1. DIVISÃO INTERNA")
print("Dataset:", DATASET)
print("Treinamento interno:", len(X_fit))
print("Validação interna:", len(X_val))
print("Teste externo utilizado: NÃO")

print("\n2. RESULTADOS DO TESTE FUNCIONAL")
print(
    results[
        [
            "Selector", "Model", "TopK",
            "Macro_F1", "Total_Seconds"
        ]
    ].to_string(
        index=False,
        float_format=lambda v: f"{v:.5f}"
    )
)

print("\n3. RESUMO")
print("Execuções:", len(results))
print("Erros:", 0)
print("Seletores testados:", len(selectors))
print("Modelos testados:", len(models))

print("\n4. LIMITAÇÃO METODOLÓGICA")
print(
    "Este é apenas um teste funcional. "
    "O pré-processamento foi herdado do "
    "treinamento externo."
)
print(
    "Os resultados NÃO devem ser utilizados "
    "nas tabelas científicas."
)

print("\n5. ARQUIVOS")
print(csv_path)
print(json_path)
print("\nCÉLULA 37 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 38B
# PILOTO CORRIGIDO COM PRÉ-PROCESSAMENTO TRAIN-ONLY
# Não acessa o teste externo.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import time
import joblib

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import (
    mutual_info_classif, f_classif
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score
)

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA_ROOT = ROOT / "01_Datasets"
NAME = "Breast_Cancer"
SEED = 42
TOP_K = [3, 5, 8, 10, 15, 20, 23]

OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Pilot"
)
OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. CARREGAR DADOS BRUTOS E METADADOS CORRETOS
# ------------------------------------------------------------

raw_path = DATA_ROOT / "Raw" / f"{NAME}.parquet"
metadata_path = (
    DATA_ROOT / "Raw" / f"{NAME}_metadata.json"
)
train_indices_path = (
    DATA_ROOT / "Splits"
    / f"{NAME}_train_indices.npy"
)
preprocessor_path = (
    DATA_ROOT / "Processed"
    / NAME / "preprocessor.joblib"
)

for path in (
    raw_path,
    metadata_path,
    train_indices_path,
    preprocessor_path
):
    if not path.is_file():
        raise FileNotFoundError(path)

with open(metadata_path, encoding="utf-8") as f:
    metadata = json.load(f)

target_column = metadata["target_column"]
raw_df = pd.read_parquet(raw_path)

if target_column not in raw_df.columns:
    raise ValueError(
        f"Variável-alvo não encontrada: {target_column}"
    )

outer_train_indices = np.load(
    train_indices_path,
    allow_pickle=False
)

if (
    len(np.unique(outer_train_indices))
    != len(outer_train_indices)
):
    raise ValueError("Índices de treino duplicados.")

if (
    outer_train_indices.min() < 0
    or outer_train_indices.max() >= len(raw_df)
):
    raise ValueError("Índices de treino inválidos.")

# Somente amostras do treinamento externo.
outer_train_raw = raw_df.iloc[
    outer_train_indices
].copy()

X_raw = outer_train_raw.drop(
    columns=[target_column]
)
y_raw = outer_train_raw[
    target_column
].to_numpy().ravel()

assert len(X_raw) == 455
assert X_raw.shape[1] == 30

# ------------------------------------------------------------
# 2. DIVISÃO INTERNA ANTES DO PRÉ-PROCESSAMENTO
# ------------------------------------------------------------

fit_idx, val_idx = train_test_split(
    np.arange(len(X_raw)),
    test_size=0.25,
    random_state=SEED,
    stratify=y_raw
)

assert len(set(fit_idx) & set(val_idx)) == 0

X_fit_raw = X_raw.iloc[fit_idx].copy()
X_val_raw = X_raw.iloc[val_idx].copy()

y_fit = y_raw[fit_idx]
y_val = y_raw[val_idx]

# ------------------------------------------------------------
# 3. REAJUSTAR PRÉ-PROCESSADOR E FILTRO DE VARIÂNCIA
# ------------------------------------------------------------

historical_artifact = joblib.load(
    preprocessor_path
)

preprocessor = clone(
    historical_artifact["preprocessor"]
)

variance_template = historical_artifact[
    "variance_filter"
]

t0 = time.perf_counter()

X_fit_trans = preprocessor.fit_transform(
    X_fit_raw, y_fit
)
X_val_trans = preprocessor.transform(
    X_val_raw
)

if variance_template is not None:
    variance_filter = clone(variance_template)

    X_fit = variance_filter.fit_transform(
        X_fit_trans
    )
    X_val = variance_filter.transform(
        X_val_trans
    )
else:
    variance_filter = None
    X_fit = X_fit_trans
    X_val = X_val_trans

preprocess_seconds = time.perf_counter() - t0

if hasattr(X_fit, "toarray"):
    X_fit = X_fit.toarray()

if hasattr(X_val, "toarray"):
    X_val = X_val.toarray()

X_fit = np.asarray(X_fit, dtype=np.float64)
X_val = np.asarray(X_val, dtype=np.float64)

assert X_fit.shape[1] == X_val.shape[1]
assert X_fit.shape[0] == len(y_fit)
assert X_val.shape[0] == len(y_val)
assert np.isfinite(X_fit).all()
assert np.isfinite(X_val).all()
assert max(TOP_K) <= X_fit.shape[1]

# ------------------------------------------------------------
# 4. SELETORES AJUSTADOS NO TREINAMENTO INTERNO
# ------------------------------------------------------------

selectors = {
    "Mutual_Information": lambda: mutual_info_classif(
        X_fit,
        y_fit,
        random_state=SEED,
        discrete_features=False
    ),
    "ANOVA": lambda: f_classif(
        X_fit, y_fit
    )[0]
}

models = {
    "DecisionTree": lambda: DecisionTreeClassifier(
        random_state=SEED
    ),
    "RandomForest": lambda: RandomForestClassifier(
        n_estimators=100,
        random_state=SEED,
        n_jobs=1
    )
}

results = []
rankings = {}

for selector_name, selector in selectors.items():

    start = time.perf_counter()

    scores = np.asarray(
        selector(),
        dtype=np.float64
    ).ravel()

    selection_seconds = (
        time.perf_counter() - start
    )

    if len(scores) != X_fit.shape[1]:
        raise RuntimeError(
            f"Ranking inválido: {selector_name}"
        )

    scores = np.nan_to_num(
        scores,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    # Ordenação decrescente; empate por índice original.
    ranking = np.lexsort((
        np.arange(len(scores)),
        -scores
    ))

    rankings[selector_name] = ranking.tolist()

    for k in TOP_K:
        chosen = ranking[:k]

        for model_name, model_factory in models.items():

            model = model_factory()

            start_fit = time.perf_counter()
            model.fit(
                X_fit[:, chosen],
                y_fit
            )
            fit_seconds = (
                time.perf_counter() - start_fit
            )

            start_predict = time.perf_counter()
            predictions = model.predict(
                X_val[:, chosen]
            )
            predict_seconds = (
                time.perf_counter() - start_predict
            )

            results.append({
                "Dataset": NAME,
                "Selector": selector_name,
                "Model": model_name,
                "TopK": k,
                "Accuracy": accuracy_score(
                    y_val, predictions
                ),
                "Balanced_Accuracy":
                    balanced_accuracy_score(
                        y_val, predictions
                    ),
                "Macro_F1": f1_score(
                    y_val,
                    predictions,
                    average="macro",
                    zero_division=0
                ),
                "Preprocess_Seconds":
                    preprocess_seconds,
                "Selection_Seconds":
                    selection_seconds,
                "Fit_Seconds":
                    fit_seconds,
                "Predict_Seconds":
                    predict_seconds,
                "Evaluation":
                    "INNER_TRAIN_ONLY_PILOT"
            })

results_df = pd.DataFrame(results)

# ------------------------------------------------------------
# 5. SALVAR SEM SOBRESCREVER RESULTADOS HISTÓRICOS
# ------------------------------------------------------------

stamp = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

csv_path = OUT / (
    f"{stamp}_breast_cancer_train_only.csv"
)
json_path = OUT / (
    f"{stamp}_breast_cancer_train_only.json"
)

results_df.to_csv(
    csv_path,
    index=False,
    mode="x"
)

audit = {
    "dataset": NAME,
    "seed": SEED,
    "target_column": target_column,
    "raw_dataset_shape": list(raw_df.shape),
    "outer_train_samples": len(X_raw),
    "inner_train_samples": len(X_fit_raw),
    "inner_validation_samples": len(X_val_raw),
    "inner_features": X_fit.shape[1],
    "outer_test_used": False,
    "historical_rankings_used": False,
    "inner_preprocessing_refitted": True,
    "inner_variance_filter_refitted":
        variance_filter is not None,
    "status": "INNER_TRAIN_ONLY_PILOT",
    "scientific_final_result": False,
    "rankings": rankings
}

with open(json_path, "x", encoding="utf-8") as f:
    json.dump(
        audit,
        f,
        ensure_ascii=False,
        indent=2
    )

# ------------------------------------------------------------
# 6. RELATÓRIO
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — PILOTO TRAIN-ONLY CORRIGIDO")
print("=" * 72)

print("\n1. ISOLAMENTO")
print("Dados brutos:", raw_df.shape)
print("Treino externo:", len(X_raw))
print("Treino interno:", len(X_fit_raw))
print("Validação interna:", len(X_val_raw))
print("Características processadas:", X_fit.shape[1])
print("Teste externo utilizado: NÃO")
print("Pré-processamento interno: REAJUSTADO")
print(
    "Filtro de variância:",
    "REAJUSTADO" if variance_filter is not None else "N/A"
)

print("\n2. RESULTADOS")
print(
    results_df[
        ["Selector", "Model", "TopK", "Macro_F1"]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)

print("\n3. RESUMO")
print("Execuções:", len(results_df))
print("Erros:", 0)
print(
    "Tempo de pré-processamento:",
    f"{preprocess_seconds:.4f} s"
)

print("\n4. ARQUIVOS")
print(csv_path)
print(json_path)

print("\nCÉLULA 38B CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 39
# AUDITORIA TRAIN-ONLY DOS 12 DATASETS
# Não acessa o teste externo.
# Não treina modelos nem executa seletores.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import time
import warnings
import joblib

import numpy as np
import pandas as pd

from scipy import sparse
from sklearn.base import clone
from sklearn.model_selection import train_test_split

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA_ROOT = ROOT / "01_Datasets"
OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

SEED = 42
INNER_TEST_SIZE = 0.25

# Recuperar a configuração do manifesto da Célula 34,
# sem depender da variável K_VALUES do notebook.
manifest_dir = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Manifests"
)

protocol_files = sorted(
    manifest_dir.glob("*_protocol.json")
)

if not protocol_files:
    raise FileNotFoundError(
        "Protocolo da Célula 34 não encontrado."
    )

with open(
    protocol_files[-1], encoding="utf-8"
) as f:
    protocol = json.load(f)

K_BY_DATASET = protocol["top_k"]
DATASETS = sorted(K_BY_DATASET)

def finite_matrix(X):
    values = X.data if sparse.issparse(X) else np.asarray(X)
    return bool(np.isfinite(values).all())

def run_dataset_audit(name):
    start = time.perf_counter()

    result = {
        "Dataset": name,
        "Status": "ERROR",
        "Task": "",
        "Outer_Train": None,
        "Inner_Train": None,
        "Inner_Validation": None,
        "Inner_Features": None,
        "Historical_Features": None,
        "Sparse": None,
        "Max_K": max(K_BY_DATASET[name]),
        "TopK_Compatible": False,
        "Preprocessor_Refitted": False,
        "Variance_Refitted": False,
        "Seconds": None,
        "Error": ""
    }

    try:
        source_dir = (
            "Synthetic" if name.startswith("Synthetic_")
            else "Raw"
        )

        raw_path = (
            DATA_ROOT / source_dir / f"{name}.parquet"
        )
        metadata_path = (
            DATA_ROOT / source_dir
            / f"{name}_metadata.json"
        )
        indices_path = (
            DATA_ROOT / "Splits"
            / f"{name}_train_indices.npy"
        )
        artifact_path = (
            DATA_ROOT / "Processed"
            / name / "preprocessor.joblib"
        )

        for path in (
            raw_path, metadata_path,
            indices_path, artifact_path
        ):
            if not path.is_file():
                raise FileNotFoundError(str(path))

        with open(
            metadata_path, encoding="utf-8"
        ) as f:
            metadata = json.load(f)

        target = metadata["target_column"]
        task = metadata["task"]

        if task not in ("classification", "regression"):
            raise ValueError(f"Tarefa inválida: {task}")

        df = pd.read_parquet(raw_path)
        indices = np.load(
            indices_path, allow_pickle=False
        )

        if (
            len(indices) == 0
            or len(np.unique(indices)) != len(indices)
            or indices.min() < 0
            or indices.max() >= len(df)
        ):
            raise ValueError(
                "Índices de treinamento inválidos."
            )

        if target not in df.columns:
            raise ValueError(
                f"Variável-alvo ausente: {target}"
            )

        # Somente os índices do treino externo.
        outer = df.iloc[indices].copy()

        X = outer.drop(columns=[target])
        y = outer[target].to_numpy().ravel()

        if task == "classification":
            fit_idx, val_idx = train_test_split(
                np.arange(len(X)),
                test_size=INNER_TEST_SIZE,
                random_state=SEED,
                stratify=y
            )
        else:
            fit_idx, val_idx = train_test_split(
                np.arange(len(X)),
                test_size=INNER_TEST_SIZE,
                random_state=SEED
            )

        if set(fit_idx) & set(val_idx):
            raise RuntimeError(
                "Sobreposição entre treino e validação."
            )

        X_fit_raw = X.iloc[fit_idx]
        X_val_raw = X.iloc[val_idx]
        y_fit = y[fit_idx]

        artifact = joblib.load(artifact_path)

        preprocessor = clone(
            artifact["preprocessor"]
        )

        variance_template = artifact.get(
            "variance_filter"
        )

        # Fit apenas nas amostras internas de treino.
        X_fit_trans = preprocessor.fit_transform(
            X_fit_raw, y_fit
        )
        X_val_trans = preprocessor.transform(
            X_val_raw
        )

        if variance_template is not None:
            variance = clone(variance_template)

            X_fit_final = variance.fit_transform(
                X_fit_trans
            )
            X_val_final = variance.transform(
                X_val_trans
            )
        else:
            variance = None
            X_fit_final = X_fit_trans
            X_val_final = X_val_trans

        if (
            X_fit_final.shape[0] != len(fit_idx)
            or X_val_final.shape[0] != len(val_idx)
            or X_fit_final.shape[1]
                != X_val_final.shape[1]
        ):
            raise RuntimeError(
                "Dimensões inconsistentes."
            )

        if not finite_matrix(X_fit_final):
            raise RuntimeError(
                "Valores não finitos no treino."
            )

        if not finite_matrix(X_val_final):
            raise RuntimeError(
                "Valores não finitos na validação."
            )

        n_features = X_fit_final.shape[1]
        compatible = (
            max(K_BY_DATASET[name]) <= n_features
        )

        historical_features = len(
            prepared_datasets[name]["feature_names"]
        )

        result.update({
            "Status": (
                "PASS" if compatible else "REVIEW"
            ),
            "Task": task,
            "Outer_Train": len(X),
            "Inner_Train": len(fit_idx),
            "Inner_Validation": len(val_idx),
            "Inner_Features": n_features,
            "Historical_Features":
                historical_features,
            "Sparse": bool(
                sparse.issparse(X_fit_final)
            ),
            "TopK_Compatible": compatible,
            "Preprocessor_Refitted": True,
            "Variance_Refitted":
                variance is not None
        })

        if not compatible:
            result["Error"] = (
                f"Top-K máximo {max(K_BY_DATASET[name])} "
                f"excede {n_features} características."
            )

    except Exception as exc:
        result["Error"] = (
            f"{type(exc).__name__}: {exc}"
        )

    result["Seconds"] = (
        time.perf_counter() - start
    )
    return result

rows = []

for name in DATASETS:
    print(f"Auditando {name}...", flush=True)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", FutureWarning)
        rows.append(run_dataset_audit(name))

results = pd.DataFrame(rows)

stamp = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)
report_path = OUT / (
    f"{stamp}_inner_preprocessing_audit.csv"
)

results.to_csv(
    report_path,
    index=False,
    mode="x"
)

print("\n" + "=" * 72)
print("CFIS — AUDITORIA TRAIN-ONLY DOS 12 DATASETS")
print("=" * 72)

print("\n1. RESULTADOS")
print(
    results[
        [
            "Dataset", "Status", "Task",
            "Inner_Train", "Inner_Validation",
            "Inner_Features", "Historical_Features",
            "Sparse", "TopK_Compatible",
            "Seconds"
        ]
    ].to_string(index=False)
)

print("\n2. RESUMO")
print(
    results["Status"].value_counts()
    .to_string()
)
print(
    "Tempo total:",
    f"{results['Seconds'].sum():.2f} s"
)

print("\n3. CASOS PENDENTES")
problems = results[
    results["Status"] != "PASS"
]

if problems.empty:
    print("Nenhum problema identificado.")
else:
    print(
        problems[
            ["Dataset", "Status", "Error"]
        ].to_string(
            index=False,
            max_colwidth=140
        )
    )

print("\n4. GARANTIAS DESTA AUDITORIA")
print("Teste externo acessado: NÃO")
print("Modelos treinados: 0")
print("Seletores executados: 0")
print("Pré-processamento: ajustado no treino interno")
print("Resultados históricos modificados: 0")

print("\nRelatório:", report_path)
print("\nCÉLULA 39 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 40
# DIAGNÓSTICO DO TF-IDF — 20_NEWSGROUPS
# Somente leitura. Nenhum treinamento.
# ============================================================

from pathlib import Path
import json
import joblib
import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA_ROOT = ROOT / "01_Datasets"
NAME = "20_Newsgroups"

raw_path = DATA_ROOT / "Raw" / f"{NAME}.parquet"
metadata_path = (
    DATA_ROOT / "Raw" / f"{NAME}_metadata.json"
)
indices_path = (
    DATA_ROOT / "Splits"
    / f"{NAME}_train_indices.npy"
)
artifact_path = (
    DATA_ROOT / "Processed"
    / NAME / "preprocessor.joblib"
)

for path in (
    raw_path, metadata_path,
    indices_path, artifact_path
):
    if not path.is_file():
        raise FileNotFoundError(path)

with open(metadata_path, encoding="utf-8") as f:
    metadata = json.load(f)

raw_df = pd.read_parquet(raw_path)
indices = np.load(
    indices_path, allow_pickle=False
)
artifact = joblib.load(artifact_path)

target = metadata["target_column"]
X_outer = raw_df.iloc[indices].drop(
    columns=[target]
)
y_outer = raw_df.iloc[indices][target]

print("=" * 72)
print("CFIS — DIAGNÓSTICO TF-IDF")
print("=" * 72)

print("\n1. DADOS BRUTOS")
print("Formato total:", raw_df.shape)
print("Treino externo:", X_outer.shape)
print("Colunas:", X_outer.columns.tolist())
print("Tipos:")
print(X_outer.dtypes.to_string())

print("\n2. METADADOS")
print(
    json.dumps(
        metadata,
        ensure_ascii=False,
        indent=2,
        default=str
    )[:4000]
)

print("\n3. PRÉ-PROCESSADOR")
preprocessor = artifact["preprocessor"]

print("Tipo:", type(preprocessor).__name__)
print("Representação:")
print(repr(preprocessor)[:5000])

print("\n4. VETORIZADORES TF-IDF ENCONTRADOS")

found = []

def inspect_estimator(obj, path="root", visited=None):
    if visited is None:
        visited = set()

    if id(obj) in visited:
        return

    visited.add(id(obj))

    if isinstance(obj, TfidfVectorizer):
        found.append((path, obj))

        print("\nCaminho:", path)
        print("min_df:", obj.min_df)
        print("max_df:", obj.max_df)
        print("max_features:", obj.max_features)
        print("ngram_range:", obj.ngram_range)
        print("stop_words:", obj.stop_words)
        print("vocabulary configurado:",
              obj.vocabulary is not None)
        print("vocabulary_ ajustado:",
              hasattr(obj, "vocabulary_"))

    if isinstance(obj, Pipeline):
        for name, child in obj.steps:
            inspect_estimator(
                child,
                f"{path}.{name}",
                visited
            )

    if isinstance(obj, ColumnTransformer):
        for name, child, columns in obj.transformers:
            print(
                "\nTransformador:",
                name,
                "| Colunas:",
                columns
            )
            if child not in ("drop", "passthrough"):
                inspect_estimator(
                    child,
                    f"{path}.{name}",
                    visited
                )

inspect_estimator(preprocessor)

if not found:
    print(
        "\nNenhum TfidfVectorizer identificado "
        "pela inspeção da estrutura."
    )

print("\n5. AMOSTRAS DO CAMPO TEXTUAL")

for column in X_outer.columns:
    series = X_outer[column]

    if (
        pd.api.types.is_object_dtype(series)
        or pd.api.types.is_string_dtype(series)
    ):
        lengths = series.fillna("").astype(str).str.len()

        print("\nColuna:", column)
        print("Valores não nulos:", series.notna().sum())
        print("Comprimento mínimo:", lengths.min())
        print("Comprimento mediano:", lengths.median())
        print("Comprimento máximo:", lengths.max())
        print("Exemplo:", repr(str(series.iloc[0])[:180]))

print("\n6. DIAGNÓSTICO DOS LIMITES")

n_outer = len(X_outer)
n_inner = round(n_outer * 0.75)

print("Documentos no treino externo:", n_outer)
print("Treino interno aproximado:", n_inner)

for path, vectorizer in found:
    min_df = vectorizer.min_df
    max_df = vectorizer.max_df

    min_count = (
        min_df
        if isinstance(min_df, (int, np.integer))
        else min_df * n_inner
    )
    max_count = (
        max_df
        if isinstance(max_df, (int, np.integer))
        else max_df * n_inner
    )

    print("\nVetorizador:", path)
    print("Limite mínimo aproximado:", min_count)
    print("Limite máximo aproximado:", max_count)
    print(
        "Limites incompatíveis:",
        max_count < min_count
    )

print("\n7. PRÓXIMO PASSO")
print(
    "Diagnóstico concluído. Nenhum parâmetro "
    "foi alterado."
)
print("\nCÉLULA 40 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 41
# CORREÇÃO DO TF-IDF — 20_NEWSGROUPS
# Sem alteração de hiperparâmetros.
# Sem acesso ao teste externo.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import time
import joblib

import numpy as np
import pandas as pd
from scipy import sparse

from sklearn.base import clone
from sklearn.model_selection import train_test_split

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA_ROOT = ROOT / "01_Datasets"
NAME = "20_Newsgroups"
SEED = 42

OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "02_Method_Audit"
)
OUT.mkdir(parents=True, exist_ok=True)

raw_path = DATA_ROOT / "Raw" / f"{NAME}.parquet"
metadata_path = (
    DATA_ROOT / "Raw" / f"{NAME}_metadata.json"
)
indices_path = (
    DATA_ROOT / "Splits"
    / f"{NAME}_train_indices.npy"
)
artifact_path = (
    DATA_ROOT / "Processed"
    / NAME / "preprocessor.joblib"
)

for path in (
    raw_path, metadata_path,
    indices_path, artifact_path
):
    if not path.is_file():
        raise FileNotFoundError(path)

with open(metadata_path, encoding="utf-8") as f:
    metadata = json.load(f)

df = pd.read_parquet(raw_path)
outer_indices = np.load(
    indices_path,
    allow_pickle=False
)

if (
    len(np.unique(outer_indices)) != len(outer_indices)
    or outer_indices.min() < 0
    or outer_indices.max() >= len(df)
):
    raise ValueError(
        "Índices do treinamento externo inválidos."
    )

outer_train = df.iloc[outer_indices].copy()

text_column = metadata["feature_columns"][0]
target_column = metadata["target_column"]

texts = outer_train[
    text_column
].fillna("").astype(str).to_numpy()

labels = outer_train[
    target_column
].to_numpy().ravel()

fit_idx, val_idx = train_test_split(
    np.arange(len(texts)),
    test_size=0.25,
    random_state=SEED,
    stratify=labels
)

if set(fit_idx) & set(val_idx):
    raise RuntimeError(
        "Sobreposição entre treino e validação."
    )

train_texts = texts[fit_idx]
validation_texts = texts[val_idx]
train_labels = labels[fit_idx]

artifact = joblib.load(artifact_path)

historical_vectorizer = artifact["preprocessor"]
historical_variance = artifact["variance_filter"]

# Clone preserva os hiperparâmetros, mas não
# reutiliza vocabulário ou IDF previamente ajustados.
vectorizer = clone(historical_vectorizer)

if vectorizer.get_params() != (
    historical_vectorizer.get_params()
):
    raise RuntimeError(
        "Hiperparâmetros do TF-IDF foram alterados."
    )

start = time.perf_counter()

# CORREÇÃO: sequência 1D de documentos.
X_fit_tfidf = vectorizer.fit_transform(
    train_texts
)

X_val_tfidf = vectorizer.transform(
    validation_texts
)

if historical_variance is not None:
    variance_filter = clone(historical_variance)

    X_fit = variance_filter.fit_transform(
        X_fit_tfidf
    )
    X_val = variance_filter.transform(
        X_val_tfidf
    )
else:
    variance_filter = None
    X_fit = X_fit_tfidf
    X_val = X_val_tfidf

elapsed = time.perf_counter() - start

def all_finite(X):
    values = (
        X.data if sparse.issparse(X)
        else np.asarray(X)
    )
    return bool(np.isfinite(values).all())

assert X_fit.shape[0] == len(fit_idx)
assert X_val.shape[0] == len(val_idx)
assert X_fit.shape[1] == X_val.shape[1]
assert all_finite(X_fit)
assert all_finite(X_val)

# Maior K previsto para este dataset.
MAX_K = 375

if X_fit.shape[1] < MAX_K:
    raise RuntimeError(
        f"Características insuficientes: "
        f"{X_fit.shape[1]} < {MAX_K}"
    )

# Registrar os parâmetros relevantes.
params = {
    "min_df": vectorizer.min_df,
    "max_df": vectorizer.max_df,
    "max_features": vectorizer.max_features,
    "ngram_range": list(vectorizer.ngram_range),
    "sublinear_tf": vectorizer.sublinear_tf
}

stamp = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S%fZ"
)

report_path = OUT / (
    f"{stamp}_newsgroups_tfidf_train_only.json"
)

report = {
    "dataset": NAME,
    "status": "PASS",
    "seed": SEED,
    "outer_train_samples": len(texts),
    "inner_train_samples": len(train_texts),
    "inner_validation_samples":
        len(validation_texts),
    "historical_features":
        len(artifact["feature_names"]),
    "inner_features": X_fit.shape[1],
    "max_k": MAX_K,
    "top_k_compatible": True,
    "tfidf_refitted": True,
    "variance_refitted":
        variance_filter is not None,
    "outer_test_accessed": False,
    "parameters_unchanged": True,
    "tfidf_parameters": params,
    "seconds": elapsed
}

with open(report_path, "x", encoding="utf-8") as f:
    json.dump(
        report,
        f,
        ensure_ascii=False,
        indent=2
    )

print("=" * 72)
print("CFIS — TF-IDF TRAIN-ONLY CORRIGIDO")
print("=" * 72)

print("\n1. ISOLAMENTO")
print("Treino externo:", len(texts))
print("Treino interno:", len(train_texts))
print("Validação interna:", len(validation_texts))
print("Teste externo utilizado: NÃO")

print("\n2. TRANSFORMAÇÃO")
print("TF-IDF:", type(vectorizer).__name__)
print("Parâmetros:", params)
print("Matriz de treino:", X_fit.shape)
print("Matriz de validação:", X_val.shape)
print(
    "Matriz esparsa:",
    sparse.issparse(X_fit)
)
print(
    "Filtro de variância reajustado:",
    variance_filter is not None
)

print("\n3. VALIDAÇÃO")
print("Valores finitos: SIM")
print("Top-K máximo:", MAX_K)
print("Top-K compatível: SIM")
print("Hiperparâmetros preservados: SIM")
print("Status: PASS")

print("\n4. DESEMPENHO")
print(f"Tempo de pré-processamento: {elapsed:.4f} s")

print("\n5. RELATÓRIO")
print(report_path)
print("\nCÉLULA 41 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 42
# EXECUTOR CHECKPOINTADO — PRIMEIRO LOTE
#
# 2 datasets × 2 seletores × 2 modelos × Top-K
# 60 avaliações; 1 divisão interna por dataset.
#
# Não utiliza o teste externo.
# Não altera os arquivos históricos.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os
import time
import warnings

import joblib
import numpy as np
import pandas as pd

from scipy import sparse
from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import (
    mutual_info_classif,
    f_classif
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score
)

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA_ROOT = ROOT / "01_Datasets"

BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

# Identificador estável: permite retomada.
RUN_ID = "inner_v1_smoke_seed42"

RUN_DIR = BASE / "Checkpoints" / RUN_ID
RESULT_DIR = BASE / "Results" / RUN_ID

RUN_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
INNER_TEST_SIZE = 0.25

K_BY_DATASET = {
    "Breast_Cancer": [3, 5, 8, 10, 15, 20, 23],
    "20_Newsgroups": [3, 5, 10, 20, 50, 125, 250, 375]
}

METHODS = (
    "Mutual_Information",
    "ANOVA_or_F_Regression"
)

MODELS = (
    "DecisionTree",
    "RandomForest"
)

def utc_now():
    return datetime.now(timezone.utc).isoformat()

def write_json_atomic(path, data):
    """Grava em arquivo temporário e substitui atomicamente."""
    temp = path.with_name(path.name + ".tmp")

    with open(temp, "w", encoding="utf-8") as f:
        json.dump(
            data, f,
            ensure_ascii=False,
            indent=2,
            default=str
        )
        f.flush()
        os.fsync(f.fileno())

    os.replace(temp, path)

def sha256_json(data):
    encoded = json.dumps(
        data,
        sort_keys=True,
        ensure_ascii=False
    ).encode("utf-8")
    return hashlib.sha256(encoded).hexdigest()

# ------------------------------------------------------------
# 1. FIXAR CONFIGURAÇÃO DA EXECUÇÃO
# ------------------------------------------------------------

configuration = {
    "run_id": RUN_ID,
    "seed": SEED,
    "inner_test_size": INNER_TEST_SIZE,
    "datasets": K_BY_DATASET,
    "methods": METHODS,
    "models": MODELS,
    "rf_n_estimators": 100,
    "rf_n_jobs": 1,
    "purpose": "CHECKPOINT_INFRASTRUCTURE_SMOKE",
    "outer_test_accessed": False,
    "final_scientific_result": False
}

config_hash = sha256_json(configuration)
config_path = RUN_DIR / "configuration.json"

if config_path.exists():
    with open(config_path, encoding="utf-8") as f:
        existing = json.load(f)

    if sha256_json(existing) != config_hash:
        raise RuntimeError(
            "A configuração do RUN_ID mudou. "
            "Não é seguro reutilizar os checkpoints."
        )
else:
    write_json_atomic(config_path, configuration)

# ------------------------------------------------------------
# 2. PREPARAÇÃO TRAIN-ONLY
# ------------------------------------------------------------

def prepare_inner_dataset(name):
    raw_dir = DATA_ROOT / "Raw"

    with open(
        raw_dir / f"{name}_metadata.json",
        encoding="utf-8"
    ) as f:
        metadata = json.load(f)

    df = pd.read_parquet(
        raw_dir / f"{name}.parquet"
    )

    outer_indices = np.load(
        DATA_ROOT / "Splits"
        / f"{name}_train_indices.npy",
        allow_pickle=False
    )

    if (
        len(outer_indices) == 0
        or len(np.unique(outer_indices))
            != len(outer_indices)
        or outer_indices.min() < 0
        or outer_indices.max() >= len(df)
    ):
        raise RuntimeError(
            f"Índices externos inválidos: {name}"
        )

    target = metadata["target_column"]

    # Acessa apenas linhas do treinamento externo.
    outer = df.iloc[outer_indices].copy()
    y = outer[target].to_numpy().ravel()

    fit_idx, val_idx = train_test_split(
        np.arange(len(outer)),
        test_size=INNER_TEST_SIZE,
        random_state=SEED,
        stratify=y
    )

    if set(fit_idx) & set(val_idx):
        raise RuntimeError(
            "Sobreposição entre treino e validação."
        )

    artifact = joblib.load(
        DATA_ROOT / "Processed"
        / name / "preprocessor.joblib"
    )

    transformer = clone(
        artifact["preprocessor"]
    )

    variance_template = artifact.get(
        "variance_filter"
    )

    if name == "20_Newsgroups":
        text_col = metadata["feature_columns"][0]

        raw_inputs = outer[text_col].fillna(
            ""
        ).astype(str).to_numpy()

        train_input = raw_inputs[fit_idx]
        val_input = raw_inputs[val_idx]

    else:
        raw_inputs = outer.drop(columns=[target])

        train_input = raw_inputs.iloc[fit_idx]
        val_input = raw_inputs.iloc[val_idx]

    start = time.perf_counter()

    X_train_trans = transformer.fit_transform(
        train_input,
        y[fit_idx]
    )

    X_val_trans = transformer.transform(
        val_input
    )

    if variance_template is not None:
        variance = clone(variance_template)

        X_train = variance.fit_transform(
            X_train_trans
        )
        X_val = variance.transform(
            X_val_trans
        )
    else:
        variance = None
        X_train = X_train_trans
        X_val = X_val_trans

    preprocess_seconds = (
        time.perf_counter() - start
    )

    if name == "20_Newsgroups":
        names = np.asarray(
            transformer.get_feature_names_out(),
            dtype=str
        )
    else:
        try:
            names = np.asarray(
                transformer.get_feature_names_out(),
                dtype=str
            )
        except (AttributeError, ValueError):
            names = np.asarray(
                raw_inputs.columns,
                dtype=str
            )

    if variance is not None:
        names = names[variance.get_support()]

    if len(names) != X_train.shape[1]:
        raise RuntimeError(
            f"Características inconsistentes: {name}"
        )

    if len(set(names)) != len(names):
        raise RuntimeError(
            f"Nomes de características duplicados: {name}"
        )

    if max(K_BY_DATASET[name]) > X_train.shape[1]:
        raise RuntimeError(
            f"Top-K incompatível: {name}"
        )

    return {
        "X_train": X_train,
        "X_val": X_val,
        "y_train": y[fit_idx],
        "y_val": y[val_idx],
        "feature_names": names,
        "preprocess_seconds": preprocess_seconds,
        "n_outer_train": len(outer),
        "n_inner_train": len(fit_idx),
        "n_inner_val": len(val_idx)
    }

# ------------------------------------------------------------
# 3. SELETORES E CLASSIFICADORES
# ------------------------------------------------------------

def select_features(method, X, y):
    # MI com discrete_features=False preserva
    # a configuração histórica auditada.
    if method == "Mutual_Information":
        scores = mutual_info_classif(
            X,
            y,
            random_state=SEED,
            discrete_features=False
        )

    elif method == "ANOVA_or_F_Regression":
        scores = f_classif(X, y)[0]

    else:
        raise ValueError(method)

    scores = np.asarray(
        scores, dtype=np.float64
    ).ravel()

    scores = np.nan_to_num(
        scores,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    return np.lexsort((
        np.arange(len(scores)),
        -scores
    )), scores

def make_model(name):
    if name == "DecisionTree":
        return DecisionTreeClassifier(
            random_state=SEED
        )

    if name == "RandomForest":
        return RandomForestClassifier(
            n_estimators=100,
            random_state=SEED,
            n_jobs=1
        )

    raise ValueError(name)

def job_path(dataset, method, model, k):
    return (
        RUN_DIR
        / f"{dataset}__{method}__{model}__K{k}.json"
    )

# ------------------------------------------------------------
# 4. EXECUÇÃO CHECKPOINTADA
# ------------------------------------------------------------

expected_jobs = sum(
    len(ks) * len(METHODS) * len(MODELS)
    for ks in K_BY_DATASET.values()
)

completed = 0
skipped = 0
failures = []

print("=" * 72)
print("CFIS — EXECUTOR CHECKPOINTADO")
print("=" * 72)
print("RUN_ID:", RUN_ID)
print("Jobs esperados:", expected_jobs)
print("Teste externo: NÃO")
print("Resultados finais do artigo: NÃO")

for dataset, ks in K_BY_DATASET.items():

    print(f"\nDATASET: {dataset}", flush=True)

    try:
        data = prepare_inner_dataset(dataset)

        X_train = data["X_train"]
        X_val = data["X_val"]

        # Para MI contínua, converter a matriz esparsa
        # em densa somente quando necessário.
        # O dataset deste lote tem no máximo 500 colunas.
        X_mi = (
            X_train.toarray()
            if sparse.issparse(X_train)
            else X_train
        )

        print(
            "Treino:",
            data["n_inner_train"],
            "| Validação:",
            data["n_inner_val"],
            "| Features:",
            X_train.shape[1],
            flush=True
        )

    except Exception as exc:
        failures.append(
            (dataset, "PREPROCESS", repr(exc))
        )
        print("ERRO:", repr(exc), flush=True)
        continue

    for method in METHODS:

        pending = [
            (model_name, k)
            for k in ks
            for model_name in MODELS
            if not job_path(
                dataset, method, model_name, k
            ).exists()
        ]

        if not pending:
            count = len(ks) * len(MODELS)
            skipped += count
            print(
                f"{method}: {count} checkpoints "
                "já existentes.",
                flush=True
            )
            continue

        try:
            start = time.perf_counter()

            with warnings.catch_warnings():
                warnings.simplefilter(
                    "ignore",
                    UserWarning
                )

                ranking, scores = select_features(
                    method,
                    X_mi if method == "Mutual_Information"
                    else X_train,
                    data["y_train"]
                )

            selection_seconds = (
                time.perf_counter() - start
            )

            ranking_path = (
                RUN_DIR
                / f"{dataset}__{method}__ranking.json"
            )

            ranking_record = {
                "dataset": dataset,
                "method": method,
                "config_hash": config_hash,
                "feature_names": (
                    data["feature_names"].tolist()
                ),
                "ranking_indices": ranking.tolist(),
                "ranking_names": (
                    data["feature_names"][ranking].tolist()
                ),
                "scores": scores.tolist(),
                "selection_seconds": selection_seconds,
                "inner_train_only": True
            }

            if not ranking_path.exists():
                write_json_atomic(
                    ranking_path,
                    ranking_record
                )

        except Exception as exc:
            failures.append(
                (dataset, method, repr(exc))
            )
            print(
                "ERRO NO SELETOR:",
                repr(exc),
                flush=True
            )
            continue

        for k in ks:
            selected = ranking[:k]

            for model_name in MODELS:
                path = job_path(
                    dataset, method, model_name, k
                )

                if path.exists():
                    skipped += 1
                    continue

                try:
                    model = make_model(model_name)

                    start = time.perf_counter()
                    model.fit(
                        X_train[:, selected],
                        data["y_train"]
                    )
                    fit_seconds = (
                        time.perf_counter() - start
                    )

                    start = time.perf_counter()
                    prediction = model.predict(
                        X_val[:, selected]
                    )
                    predict_seconds = (
                        time.perf_counter() - start
                    )

                    record = {
                        "Run_ID": RUN_ID,
                        "Config_Hash": config_hash,
                        "Dataset": dataset,
                        "Method": method,
                        "Model": model_name,
                        "TopK": int(k),
                        "Inner_Train": data["n_inner_train"],
                        "Inner_Validation": data["n_inner_val"],
                        "Inner_Features": X_train.shape[1],
                        "Accuracy": float(
                            accuracy_score(
                                data["y_val"],
                                prediction
                            )
                        ),
                        "Balanced_Accuracy": float(
                            balanced_accuracy_score(
                                data["y_val"],
                                prediction
                            )
                        ),
                        "Macro_F1": float(
                            f1_score(
                                data["y_val"],
                                prediction,
                                average="macro",
                                zero_division=0
                            )
                        ),
                        "Preprocess_Seconds":
                            data["preprocess_seconds"],
                        "Selection_Seconds":
                            selection_seconds,
                        "Fit_Seconds": fit_seconds,
                        "Predict_Seconds":
                            predict_seconds,
                        "Outer_Test_Used": False,
                        "Final_Result": False,
                        "Created_UTC": utc_now()
                    }

                    write_json_atomic(path, record)
                    completed += 1

                    print(
                        f"  {method} | {model_name} "
                        f"| K={k} | "
                        f"F1={record['Macro_F1']:.4f}",
                        flush=True
                    )

                except Exception as exc:
                    failures.append(
                        (
                            dataset,
                            f"{method}/{model_name}/K{k}",
                            repr(exc)
                        )
                    )
                    print(
                        "  ERRO:",
                        repr(exc),
                        flush=True
                    )

# ------------------------------------------------------------
# 5. CONSOLIDAR CHECKPOINTS
# ------------------------------------------------------------

records = []

for dataset, ks in K_BY_DATASET.items():
    for method in METHODS:
        for k in ks:
            for model_name in MODELS:
                path = job_path(
                    dataset, method, model_name, k
                )

                if path.exists():
                    with open(
                        path, encoding="utf-8"
                    ) as f:
                        record = json.load(f)

                    if (
                        record.get("Config_Hash")
                        != config_hash
                    ):
                        raise RuntimeError(
                            "Checkpoint com configuração "
                            f"incompatível: {path}"
                        )

                    records.append(record)

results = pd.DataFrame(records)

summary_path = (
    RESULT_DIR / "smoke_results.csv"
)

# O consolidado é derivado dos checkpoints.
temp_csv = summary_path.with_name(
    summary_path.name + ".tmp"
)
results.to_csv(temp_csv, index=False)
os.replace(temp_csv, summary_path)

print("\n" + "=" * 72)
print("RESUMO DO LOTE")
print("=" * 72)

print("Jobs esperados:", expected_jobs)
print("Concluídos nesta execução:", completed)
print("Ignorados por checkpoint:", skipped)
print("Checkpoints disponíveis:", len(results))
print("Falhas nesta execução:", len(failures))

if not results.empty:
    print("\nMELHOR MACRO-F1 POR DATASET/MÉTODO/MODELO")
    best = (
        results.sort_values(
            "Macro_F1", ascending=False
        )
        .drop_duplicates(
            ["Dataset", "Method", "Model"]
        )
        [
            [
                "Dataset", "Method",
                "Model", "TopK", "Macro_F1"
            ]
        ]
    )

    print(
        best.to_string(
            index=False,
            float_format=lambda x: f"{x:.5f}"
        )
    )

if failures:
    print("\nFALHAS")
    for failure in failures:
        print(failure)

print("\nArquivo consolidado:")
print(summary_path)

print("\nStatus:", (
    "COMPLETE"
    if len(results) == expected_jobs
    else "INCOMPLETE"
))

print("\nCÉLULA 42 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 43
# ESTABILIDADE DOS RANKINGS — BREAST_CANCER
#
# 10 divisões internas, MI e ANOVA.
# Jaccard médio entre todos os pares de divisões.
# Não acessa o teste externo.
# ============================================================

from pathlib import Path
from itertools import combinations
from datetime import datetime, timezone
import json
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.feature_selection import (
    mutual_info_classif, f_classif
)

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
NAME = "Breast_Cancer"
SEED = 42
N_SPLITS = 10
TEST_SIZE = 0.25
TOP_K = [3, 5, 8, 10, 15, 20, 23]

OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Checkpoints"
    / "stability_v1_breast_cancer"
)
OUT.mkdir(parents=True, exist_ok=True)

raw_path = (
    ROOT / "01_Datasets" / "Raw"
    / f"{NAME}.parquet"
)
metadata_path = (
    ROOT / "01_Datasets" / "Raw"
    / f"{NAME}_metadata.json"
)
indices_path = (
    ROOT / "01_Datasets" / "Splits"
    / f"{NAME}_train_indices.npy"
)
artifact_path = (
    ROOT / "01_Datasets" / "Processed"
    / NAME / "preprocessor.joblib"
)

with open(metadata_path, encoding="utf-8") as f:
    metadata = json.load(f)

raw_df = pd.read_parquet(raw_path)
outer_indices = np.load(
    indices_path, allow_pickle=False
)

outer = raw_df.iloc[outer_indices].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(artifact_path)

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS,
    test_size=TEST_SIZE,
    random_state=SEED
)

def save_atomic(path, payload):
    tmp = path.with_name(path.name + ".tmp")
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(
            payload, f,
            ensure_ascii=False,
            indent=2
        )
        f.flush()
        os.fsync(f.fileno())
    os.replace(tmp, path)

rankings = {
    "Mutual_Information": [],
    "ANOVA_or_F_Regression": []
}

print("=" * 72)
print("CFIS — ESTABILIDADE EM 10 DIVISÕES")
print("=" * 72)

for split_number, (fit_idx, val_idx) in enumerate(
    splitter.split(X_raw, y), start=1
):
    print(
        f"\nDivisão {split_number}/{N_SPLITS}",
        flush=True
    )

    preprocessor = clone(
        artifact["preprocessor"]
    )
    variance = clone(
        artifact["variance_filter"]
    )

    X_fit_raw = X_raw.iloc[fit_idx]
    y_fit = y[fit_idx]

    X_trans = preprocessor.fit_transform(
        X_fit_raw, y_fit
    )
    X_fit = variance.fit_transform(X_trans)

    X_fit = np.asarray(X_fit, dtype=np.float64)

    try:
        names = np.asarray(
            preprocessor.get_feature_names_out(),
            dtype=str
        )
    except (AttributeError, ValueError):
        names = np.asarray(
            X_fit_raw.columns, dtype=str
        )

    names = names[variance.get_support()]

    if len(names) != X_fit.shape[1]:
        raise RuntimeError(
            "Nomes e características inconsistentes."
        )

    methods = {
        "Mutual_Information":
            mutual_info_classif(
                X_fit,
                y_fit,
                random_state=SEED,
                discrete_features=False
            ),
        "ANOVA_or_F_Regression":
            f_classif(X_fit, y_fit)[0]
    }

    for method, raw_scores in methods.items():
        path = OUT / (
            f"split_{split_number:02d}__"
            f"{method}.json"
        )

        if path.exists():
            with open(path, encoding="utf-8") as f:
                saved = json.load(f)

            ranking_names = saved["ranking_names"]
            print(
                f"  {method}: checkpoint existente"
            )
        else:
            scores = np.nan_to_num(
                np.asarray(
                    raw_scores, dtype=float
                ),
                nan=0.0,
                posinf=0.0,
                neginf=0.0
            )

            order = np.lexsort((
                np.arange(len(scores)),
                -scores
            ))

            ranking_names = names[order].tolist()

            save_atomic(path, {
                "dataset": NAME,
                "method": method,
                "split_number": split_number,
                "seed": SEED,
                "n_fit": len(fit_idx),
                "n_validation": len(val_idx),
                "ranking_names": ranking_names,
                "outer_test_used": False,
                "created_utc": datetime.now(
                    timezone.utc
                ).isoformat()
            })

            print(
                f"  {method}: ranking salvo "
                f"({len(ranking_names)} features)"
            )

        rankings[method].append(ranking_names)

def jaccard(a, b):
    a, b = set(a), set(b)
    union = a | b
    return (
        len(a & b) / len(union)
        if union else 1.0
    )

rows = []

for method, method_rankings in rankings.items():
    for k in TOP_K:
        values = [
            jaccard(a[:k], b[:k])
            for a, b in combinations(
                method_rankings, 2
            )
        ]

        rows.append({
            "Dataset": NAME,
            "Method": method,
            "TopK": k,
            "Splits": N_SPLITS,
            "Pairs": len(values),
            "Mean_Jaccard": float(np.mean(values)),
            "Std_Jaccard": float(np.std(
                values, ddof=1
            )),
            "Min_Jaccard": float(np.min(values)),
            "Max_Jaccard": float(np.max(values))
        })

results = pd.DataFrame(rows)

report_path = OUT / "stability_jaccard.csv"
temp_path = OUT / "stability_jaccard.csv.tmp"

results.to_csv(temp_path, index=False)
os.replace(temp_path, report_path)

print("\n" + "=" * 72)
print("ESTABILIDADE — JACCARD ENTRE DIVISÕES")
print("=" * 72)

print(
    results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nCheckpoints:", OUT)
print("Relatório:", report_path)
print("Teste externo utilizado: NÃO")
print("\nCÉLULA 43 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 44
# ESTABILIDADE DO CFIS ATUAL — BREAST_CANCER
#
# 10 divisões internas idênticas à Célula 43
# Top-K: 3, 5, 8, 10, 15, 20, 23
# Checkpoints individuais e retomada automática
# Teste externo: NÃO
# ============================================================

from pathlib import Path
from itertools import combinations
from datetime import datetime, timezone
import importlib.util
import inspect
import hashlib
import json
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"
NAME = "Breast_Cancer"

SEED = 42
N_REPEATS = 10
N_SPLITS = 10
TEST_SIZE = 0.25
TOP_K = [3, 5, 8, 10, 15, 20, 23]

OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Checkpoints"
    / "stability_v1_breast_cancer"
)

OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. IDENTIFICAR A IMPLEMENTAÇÃO ATUAL
# ------------------------------------------------------------

# Caso a versão atual já tenha sido importada no notebook,
# ela pode ser identificada por seu módulo de origem.
#
# Não utilizamos automaticamente a implementação histórica.

candidate_paths = []

for folder in [
    ROOT / "09_Logs",
    ROOT / "04_Efficiency",
]:
    if folder.exists():
        candidate_paths.extend(
            p for p in folder.rglob("*.py")
            if (
                "cfis" in p.name.lower()
                and any(
                    word in p.name.lower()
                    for word in [
                        "optimized",
                        "optimizado",
                        "otimizado",
                        "fast",
                        "v4",
                        "current"
                    ]
                )
            )
        )

candidate_paths = sorted(set(candidate_paths))

# É possível definir explicitamente a versão oficial
# caso haja mais de uma implementação candidata.
CFIS_CURRENT_PATH = None

if CFIS_CURRENT_PATH is not None:
    source_path = Path(CFIS_CURRENT_PATH)

elif len(candidate_paths) == 1:
    source_path = candidate_paths[0]

else:
    print("=" * 72)
    print("IDENTIFICAÇÃO DA IMPLEMENTAÇÃO ATUAL DO CFIS")
    print("=" * 72)
    print("Candidatas encontradas:", len(candidate_paths))

    for path in candidate_paths:
        print(path)

    raise RuntimeError(
        "Não foi possível identificar inequivocamente "
        "a implementação atual. Defina CFIS_CURRENT_PATH "
        "com o caminho da versão oficial já validada."
    )

if not source_path.is_file():
    raise FileNotFoundError(source_path)

source_hash = hashlib.sha256(
    source_path.read_bytes()
).hexdigest()

spec = importlib.util.spec_from_file_location(
    "cfis_current_benchmark",
    source_path
)

module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

if not hasattr(module, "CFIS_feature_selection"):
    raise RuntimeError(
        "A implementação selecionada não disponibiliza "
        "CFIS_feature_selection."
    )

cfis_function = module.CFIS_feature_selection

print("=" * 72)
print("CFIS — ESTABILIDADE EM 10 DIVISÕES")
print("=" * 72)
print("Implementação:", source_path)
print("SHA256:", source_hash)
print("Assinatura:", inspect.signature(cfis_function))

# ------------------------------------------------------------
# 2. CARREGAR SOMENTE O TREINAMENTO EXTERNO
# ------------------------------------------------------------

with open(
    DATA / "Raw" / f"{NAME}_metadata.json",
    encoding="utf-8"
) as f:
    metadata = json.load(f)

raw = pd.read_parquet(
    DATA / "Raw" / f"{NAME}.parquet"
)

outer_indices = np.load(
    DATA / "Splits" / f"{NAME}_train_indices.npy",
    allow_pickle=False
)

outer = raw.iloc[outer_indices].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed"
    / NAME / "preprocessor.joblib"
)

# ------------------------------------------------------------
# 3. INTERFACE EQUIVALENTE À FORMULAÇÃO VALIDADA
# ------------------------------------------------------------

signature = inspect.signature(cfis_function)

required = {
    "X",
    "y",
    "task",
    "n_repeats",
    "random_state"
}

missing = required - set(signature.parameters)

if missing:
    raise RuntimeError(
        "Assinatura incompatível. Argumentos ausentes: "
        + ", ".join(sorted(missing))
    )

def run_cfis(X, y):
    kwargs = {
        "X": X,
        "y": y,
        "task": "classification",
        "n_repeats": N_REPEATS,
        "random_state": SEED
    }

    if "model" in signature.parameters:
        kwargs["model"] = None

    return cfis_function(**kwargs)

def extract_ranking(result, feature_names):
    """
    Aceita apenas saídas que permitam identificar
    inequivocamente a ordenação completa.
    """

    n = len(feature_names)

    if isinstance(result, dict):
        for key in [
            "ranking",
            "ranked_features",
            "selected_features",
            "feature_ranking"
        ]:
            if key in result:
                return extract_ranking(
                    result[key], feature_names
                )

        for key in [
            "scores",
            "feature_importances",
            "importance_scores"
        ]:
            if key in result:
                return extract_ranking(
                    result[key], feature_names
                )

    if isinstance(result, pd.Series):
        if (
            len(result) == n
            and set(result.index.astype(str))
                == set(feature_names)
        ):
            scores = result.reindex(
                feature_names
            ).to_numpy(dtype=float)

            if not np.isfinite(scores).all():
                raise ValueError(
                    "Importâncias não finitas."
                )

            order = np.lexsort((
                np.arange(n), -scores
            ))
            return feature_names[order].tolist()

        result = result.to_numpy()

    arr = np.asarray(result)

    if arr.ndim != 1 or len(arr) != n:
        raise RuntimeError(
            "Formato de saída não reconhecido. "
            "Não é seguro inferir o ranking."
        )

    if np.issubdtype(arr.dtype, np.number):
        values = arr.astype(float)

        if not np.isfinite(values).all():
            raise RuntimeError(
                "Valores não finitos no retorno."
            )

        # Vetores numéricos são tratados como scores.
        order = np.lexsort((
            np.arange(n), -values
        ))
        return feature_names[order].tolist()

    names = arr.astype(str).tolist()

    if (
        len(set(names)) == n
        and set(names) == set(feature_names)
    ):
        return names

    raise RuntimeError(
        "Não foi possível validar a ordenação "
        "completa das características."
    )

def save_atomic(path, payload):
    tmp = path.with_name(path.name + ".tmp")

    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(
            payload, f,
            ensure_ascii=False,
            indent=2
        )
        f.flush()
        os.fsync(f.fileno())

    os.replace(tmp, path)

# ------------------------------------------------------------
# 4. EXECUTAR AS MESMAS DEZ DIVISÕES
# ------------------------------------------------------------

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS,
    test_size=TEST_SIZE,
    random_state=SEED
)

rankings = []
execution_times = []

for split_number, (fit_idx, val_idx) in enumerate(
    splitter.split(X_raw, y), start=1
):
    checkpoint = OUT / (
        f"split_{split_number:02d}__CFIS.json"
    )

    if checkpoint.exists():
        with open(checkpoint, encoding="utf-8") as f:
            saved = json.load(f)

        if saved.get("source_sha256") != source_hash:
            raise RuntimeError(
                f"Implementação diferente no checkpoint "
                f"da divisão {split_number}."
            )

        ranking = saved["ranking_names"]
        elapsed = saved["selection_seconds"]

        print(
            f"Divisão {split_number}: "
            "checkpoint reutilizado"
        )

    else:
        preprocessor = clone(
            artifact["preprocessor"]
        )

        variance = clone(
            artifact["variance_filter"]
        )

        X_fit_raw = X_raw.iloc[fit_idx]
        y_fit = y[fit_idx]

        X_trans = preprocessor.fit_transform(
            X_fit_raw, y_fit
        )

        X_fit = variance.fit_transform(X_trans)

        X_fit = np.asarray(
            X_fit, dtype=np.float64
        )

        try:
            names = np.asarray(
                preprocessor.get_feature_names_out(),
                dtype=str
            )
        except (AttributeError, ValueError):
            names = np.asarray(
                X_fit_raw.columns, dtype=str
            )

        names = names[variance.get_support()]

        if len(names) != X_fit.shape[1]:
            raise RuntimeError(
                "Dimensão das características inconsistente."
            )

        start = time.perf_counter()

        result = run_cfis(
            pd.DataFrame(
                X_fit,
                columns=names
            ),
            pd.Series(y_fit)
        )

        elapsed = time.perf_counter() - start

        ranking = extract_ranking(
            result, names
        )

        save_atomic(checkpoint, {
            "dataset": NAME,
            "method": "CFIS",
            "split_number": split_number,
            "seed": SEED,
            "n_repeats": N_REPEATS,
            "source_sha256": source_hash,
            "n_fit": len(fit_idx),
            "n_validation": len(val_idx),
            "ranking_names": ranking,
            "selection_seconds": elapsed,
            "outer_test_used": False,
            "created_utc": datetime.now(
                timezone.utc
            ).isoformat()
        })

        print(
            f"Divisão {split_number}/{N_SPLITS}: "
            f"{elapsed:.3f} s"
        )

    rankings.append(ranking)
    execution_times.append(elapsed)

# ------------------------------------------------------------
# 5. ESTABILIDADE JACCARD
# ------------------------------------------------------------

def jaccard(a, b):
    a, b = set(a), set(b)
    union = a | b
    return len(a & b) / len(union) if union else 1.0

rows = []

for k in TOP_K:
    values = [
        jaccard(a[:k], b[:k])
        for a, b in combinations(rankings, 2)
    ]

    rows.append({
        "Dataset": NAME,
        "Method": "CFIS",
        "TopK": k,
        "Splits": N_SPLITS,
        "Pairs": len(values),
        "Mean_Jaccard": float(np.mean(values)),
        "Std_Jaccard": float(np.std(
            values, ddof=1
        )),
        "Min_Jaccard": float(np.min(values)),
        "Max_Jaccard": float(np.max(values))
    })

results = pd.DataFrame(rows)

report = OUT / "stability_cfis_jaccard.csv"
results.to_csv(report, index=False)

print("\n" + "=" * 72)
print("CFIS — ESTABILIDADE TOP-K")
print("=" * 72)

print(
    results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nTempo médio de seleção:",
      f"{np.mean(execution_times):.4f} s")

print("Checkpoints:", OUT)
print("Relatório:", report)
print("Teste externo utilizado: NÃO")
print("\nCÉLULA 44 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 44A
# LOCALIZAR IMPLEMENTAÇÃO ATUAL
# Somente diagnóstico — nenhum treinamento
# ============================================================

from pathlib import Path
import inspect
import ast
import sys
import hashlib

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")

print("=" * 72)
print("CFIS — LOCALIZAÇÃO DA IMPLEMENTAÇÃO")
print("=" * 72)

# ------------------------------------------------------------
# 1. PROCURAR FUNÇÕES NO NOTEBOOK ATUAL
# ------------------------------------------------------------

print("\n1. FUNÇÕES DISPONÍVEIS NO NOTEBOOK")

found_functions = []

for name, obj in list(globals().items()):
    if (
        "cfis" in name.lower()
        and inspect.isfunction(obj)
    ):
        try:
            signature = str(inspect.signature(obj))
        except (ValueError, TypeError):
            signature = "indisponível"

        try:
            source_file = inspect.getsourcefile(obj)
        except (TypeError, OSError):
            source_file = None

        found_functions.append(name)

        print(f"\nFunção: {name}")
        print("Assinatura:", signature)
        print("Origem:", source_file or "Notebook/memória")

if not found_functions:
    print("Nenhuma função CFIS encontrada no namespace.")

# ------------------------------------------------------------
# 2. PROCURAR NOS MÓDULOS JÁ CARREGADOS
# ------------------------------------------------------------

print("\n2. MÓDULOS CARREGADOS")

for module_name, module in list(sys.modules.items()):
    if module is None:
        continue

    if "cfis" in module_name.lower():
        print(
            module_name,
            "|",
            getattr(module, "__file__", None)
        )

# ------------------------------------------------------------
# 3. PROCURAR DEFINIÇÕES NOS ARQUIVOS PYTHON
# ------------------------------------------------------------

print("\n3. DEFINIÇÕES NOS ARQUIVOS DO PROJETO")

matches = []

for path in ROOT.rglob("*.py"):
    try:
        source = path.read_text(encoding="utf-8")
        tree = ast.parse(source)
    except (OSError, UnicodeError, SyntaxError):
        continue

    for node in ast.walk(tree):
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        ) and "cfis" in node.name.lower():

            record = {
                "path": str(path),
                "function": node.name,
                "line": node.lineno,
                "sha256": hashlib.sha256(
                    path.read_bytes()
                ).hexdigest()
            }

            matches.append(record)

            print(
                f"\nArquivo: {path}"
                f"\nFunção: {node.name}"
                f"\nLinha: {node.lineno}"
                f"\nSHA256: {record['sha256']}"
            )

# ------------------------------------------------------------
# 4. RESUMO
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("RESUMO")
print("=" * 72)

print(
    "Funções no notebook:",
    len(found_functions)
)

print(
    "Definições em arquivos:",
    len(matches)
)

print("Treinamentos executados: 0")
print("Arquivos modificados: 0")

print("\nCÉLULA 44A CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 44B
# IDENTIFICAR A IMPLEMENTAÇÃO ATUAL VALIDADA
# Apenas inspeção, sem execução de experimentos
# ============================================================

import inspect
import hashlib
import ast

print("=" * 72)
print("CFIS — IDENTIFICAÇÃO DA VERSÃO ATUAL")
print("=" * 72)

# ------------------------------------------------------------
# 1. INSPECIONAR FUNÇÕES DE AUDITORIA E EXECUÇÃO
# ------------------------------------------------------------

functions_to_inspect = [
    "execute_cfis",
    "audit_cfis",
    "instrumented_cfis"
]

for name in functions_to_inspect:
    obj = globals().get(name)

    print("\n" + "-" * 72)
    print("FUNÇÃO:", name)

    if obj is None:
        print("Não encontrada.")
        continue

    try:
        print("Assinatura:", inspect.signature(obj))
        source = inspect.getsource(obj)
        print("\nCÓDIGO:")
        print(source[:16000])
    except (OSError, TypeError) as exc:
        print("Código indisponível:", repr(exc))

# ------------------------------------------------------------
# 2. LOCALIZAR VARIÁVEIS RELACIONADAS ÀS VERSÕES
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("VARIÁVEIS RELACIONADAS AO CFIS")
print("=" * 72)

keywords = [
    "cfis",
    "optimized",
    "optim",
    "variant",
    "adapter",
    "equivalent"
]

for name, obj in sorted(
    list(globals().items()),
    key=lambda item: item[0].lower()
):
    if name.startswith("_"):
        continue

    if not any(
        word in name.lower()
        for word in keywords
    ):
        continue

    if inspect.ismodule(obj):
        continue

    if inspect.isfunction(obj):
        print(
            f"{name}: function "
            f"{inspect.signature(obj)}"
        )
    elif isinstance(
        obj,
        (str, int, float, bool, type(None))
    ):
        print(f"{name}: {repr(obj)[:300]}")
    elif isinstance(obj, dict):
        print(
            f"{name}: dict "
            f"keys={list(obj.keys())[:20]}"
        )
    elif isinstance(obj, (list, tuple)):
        print(
            f"{name}: {type(obj).__name__} "
            f"length={len(obj)}"
        )

# ------------------------------------------------------------
# 3. LOCALIZAR CÉLULAS QUE DEFINIRAM O CFIS
# ------------------------------------------------------------

print("\n" + "=" * 72)
print("DEFINIÇÕES NO HISTÓRICO DO NOTEBOOK")
print("=" * 72)

ip = get_ipython()
history = ip.history_manager.input_hist_raw

matches = []

for index, code in enumerate(history):
    if not isinstance(code, str):
        continue

    try:
        tree = ast.parse(code)
    except SyntaxError:
        continue

    definitions = [
        node.name
        for node in ast.walk(tree)
        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        )
        and (
            "cfis" in node.name.lower()
            or "optimized" in node.name.lower()
            or "adapter" in node.name.lower()
        )
    ]

    if definitions:
        matches.append((index, definitions))

for index, definitions in matches[-30:]:
    print(
        f"Célula histórica {index}: "
        f"{', '.join(definitions)}"
    )

print("\nTotal de células com definições:", len(matches))
print("Treinamentos executados: 0")
print("Arquivos modificados: 0")
print("\nCÉLULA 44B CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 44C
# ESTABILIDADE CFIS + COMPARAÇÃO MI / ANOVA
#
# Breast_Cancer | 10 divisões | 7 valores Top-K
# Jaccard médio + estabilidade de Nogueira
# CFIS atual: consolidated_permutation_adapter
# ============================================================

from pathlib import Path
from itertools import combinations
from datetime import datetime, timezone
import hashlib
import json
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"

NAME = "Breast_Cancer"
SEED = 42
N_REPEATS = 10
N_SPLITS = 10
TEST_SIZE = 0.25
TOP_K = [3, 5, 8, 10, 15, 20, 23]

OUT = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
    / "Checkpoints"
    / "stability_v1_breast_cancer"
)

OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. VERIFICAR A IMPLEMENTAÇÃO ATUAL
# ------------------------------------------------------------

required_globals = [
    "cfis_function",
    "cfis_globals",
    "original_permutation",
    "consolidated_permutation_adapter"
]

missing = [
    name for name in required_globals
    if name not in globals()
]

if missing:
    raise RuntimeError(
        "Funções necessárias ausentes do notebook: "
        + ", ".join(missing)
    )

if not callable(cfis_function):
    raise RuntimeError("cfis_function não é executável.")

if not callable(consolidated_permutation_adapter):
    raise RuntimeError(
        "Adaptador atual não é executável."
    )

if globals().get("exact_cfis") is not True:
    raise RuntimeError(
        "Auditoria de equivalência exata não confirmada."
    )

# Confirma que o dicionário pertence à função principal.
if cfis_function.__globals__ is not cfis_globals:
    raise RuntimeError(
        "O namespace do CFIS não corresponde "
        "ao namespace auditado."
    )

original_adapter = cfis_globals["permutation_importance"]

if original_adapter is not original_permutation:
    raise RuntimeError(
        "O adaptador original não está restaurado. "
        "Verifique o estado do notebook."
    )

print("=" * 72)
print("CFIS — ESTABILIDADE EM 10 DIVISÕES")
print("=" * 72)
print("Implementação atual: CFIS")
print("Adaptador validado: SIM")
print("Equivalência exata: SIM")
print("Teste externo: NÃO")

# ------------------------------------------------------------
# 2. CARREGAR SOMENTE TREINO EXTERNO
# ------------------------------------------------------------

with open(
    DATA / "Raw" / f"{NAME}_metadata.json",
    encoding="utf-8"
) as f:
    metadata = json.load(f)

raw = pd.read_parquet(
    DATA / "Raw" / f"{NAME}.parquet"
)

outer_indices = np.load(
    DATA / "Splits" / f"{NAME}_train_indices.npy",
    allow_pickle=False
)

outer = raw.iloc[outer_indices].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed"
    / NAME / "preprocessor.joblib"
)

# ------------------------------------------------------------
# 3. EXECUTAR CFIS ATUAL
# ------------------------------------------------------------

def run_current_cfis(X, y):
    try:
        cfis_globals["permutation_importance"] = (
            consolidated_permutation_adapter
        )

        return cfis_function(
            X,
            y,
            task="classification",
            model=None,
            n_repeats=N_REPEATS,
            random_state=SEED
        )

    finally:
        cfis_globals["permutation_importance"] = (
            original_adapter
        )

def extract_cfis_ranking(result, names):
    """
    O retorno histórico do CFIS é um DataFrame
    contendo a coluna Feature.
    """
    if not isinstance(result, pd.DataFrame):
        raise TypeError(
            "Retorno CFIS inesperado: "
            f"{type(result).__name__}"
        )

    if "Feature" not in result.columns:
        raise RuntimeError(
            "Coluna Feature não encontrada."
        )

    ranking = result["Feature"].astype(str).tolist()

    if (
        len(ranking) != len(names)
        or len(set(ranking)) != len(ranking)
        or set(ranking) != set(names)
    ):
        raise RuntimeError(
            "Ranking CFIS incompleto ou incompatível."
        )

    return ranking

def save_atomic(path, payload):
    tmp = path.with_name(path.name + ".tmp")

    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(
            payload, f,
            ensure_ascii=False,
            indent=2
        )
        f.flush()
        os.fsync(f.fileno())

    os.replace(tmp, path)

# ------------------------------------------------------------
# 4. MESMAS DIVISÕES DA CÉLULA 43
# ------------------------------------------------------------

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS,
    test_size=TEST_SIZE,
    random_state=SEED
)

rankings_by_method = {
    "CFIS": [],
    "Mutual_Information": [],
    "ANOVA_or_F_Regression": []
}

times = []

for split_number, (fit_idx, val_idx) in enumerate(
    splitter.split(X_raw, y), start=1
):
    checkpoint = OUT / (
        f"split_{split_number:02d}__CFIS.json"
    )

    if checkpoint.exists():
        with open(checkpoint, encoding="utf-8") as f:
            saved = json.load(f)

        if (
            saved.get("method") != "CFIS"
            or saved.get("seed") != SEED
            or saved.get("n_repeats") != N_REPEATS
        ):
            raise RuntimeError(
                f"Checkpoint CFIS incompatível: "
                f"{checkpoint}"
            )

        ranking = saved["ranking_names"]
        elapsed = saved["selection_seconds"]

        print(
            f"Divisão {split_number:02d}: "
            "CFIS recuperado"
        )

    else:
        preprocessor = clone(
            artifact["preprocessor"]
        )

        variance = clone(
            artifact["variance_filter"]
        )

        X_fit_raw = X_raw.iloc[fit_idx]
        y_fit = y[fit_idx]

        X_trans = preprocessor.fit_transform(
            X_fit_raw, y_fit
        )

        X_fit = variance.fit_transform(X_trans)
        X_fit = np.asarray(X_fit, dtype=np.float64)

        try:
            names = np.asarray(
                preprocessor.get_feature_names_out(),
                dtype=str
            )
        except (AttributeError, ValueError):
            names = np.asarray(
                X_fit_raw.columns, dtype=str
            )

        names = names[variance.get_support()]

        X_cfis = pd.DataFrame(
            X_fit,
            columns=names
        )

        start = time.perf_counter()

        result = run_current_cfis(
            X_cfis,
            pd.Series(y_fit)
        )

        elapsed = time.perf_counter() - start

        ranking = extract_cfis_ranking(
            result, names
        )

        save_atomic(checkpoint, {
            "dataset": NAME,
            "method": "CFIS",
            "split_number": split_number,
            "seed": SEED,
            "n_repeats": N_REPEATS,
            "n_fit": len(fit_idx),
            "n_validation": len(val_idx),
            "ranking_names": ranking,
            "selection_seconds": elapsed,
            "outer_test_used": False,
            "created_utc": datetime.now(
                timezone.utc
            ).isoformat()
        })

        print(
            f"Divisão {split_number:02d}: "
            f"CFIS {elapsed:.4f} s"
        )

    rankings_by_method["CFIS"].append(ranking)
    times.append(elapsed)

    # Carregar rankings existentes de MI e ANOVA.
    for method in [
        "Mutual_Information",
        "ANOVA_or_F_Regression"
    ]:
        path = OUT / (
            f"split_{split_number:02d}__{method}.json"
        )

        if not path.exists():
            raise FileNotFoundError(path)

        with open(path, encoding="utf-8") as f:
            record = json.load(f)

        if (
            record.get("split_number") != split_number
            or record.get("seed") != SEED
            or record.get("method") != method
        ):
            raise RuntimeError(
                f"Checkpoint incompatível: {path}"
            )

        rankings_by_method[method].append(
            record["ranking_names"]
        )

# ------------------------------------------------------------
# 5. MÉTRICAS DE ESTABILIDADE
# ------------------------------------------------------------

def jaccard(a, b):
    a, b = set(a), set(b)
    union = a | b
    return len(a & b) / len(union) if union else 1.0

def nogueira_stability(rankings, k):
    """
    Índice de estabilidade de Nogueira.

    Para subconjuntos de tamanho fixo k,
    com p características no universo comum:

    estabilidade = 1 - variância_observada /
                       variância_esperada

    Usa variância amostral (ddof=1).
    """
    universe = sorted(
        set().union(
            *(set(r) for r in rankings)
        )
    )

    p = len(universe)
    m = len(rankings)

    if not (0 < k < p):
        return float("nan")

    selection_matrix = np.zeros(
        (m, p), dtype=np.float64
    )

    index = {
        name: j
        for j, name in enumerate(universe)
    }

    for i, ranking in enumerate(rankings):
        for feature in ranking[:k]:
            selection_matrix[
                i, index[feature]
            ] = 1.0

    observed = np.var(
        selection_matrix,
        axis=0,
        ddof=1
    ).mean()

    expected = (k / p) * (1 - k / p)

    return float(1 - observed / expected)

rows = []

for method, method_rankings in rankings_by_method.items():
    for k in TOP_K:
        pairwise = [
            jaccard(a[:k], b[:k])
            for a, b in combinations(
                method_rankings, 2
            )
        ]

        rows.append({
            "Dataset": NAME,
            "Method": method,
            "TopK": k,
            "Splits": N_SPLITS,
            "Pairs": len(pairwise),
            "Mean_Jaccard": float(
                np.mean(pairwise)
            ),
            "Std_Jaccard": float(
                np.std(pairwise, ddof=1)
            ),
            "Nogueira": nogueira_stability(
                method_rankings, k
            )
        })

comparison = pd.DataFrame(rows)

report = OUT / "stability_comparison_cfis_mi_anova.csv"
temp = report.with_name(report.name + ".tmp")

comparison.to_csv(temp, index=False)
os.replace(temp, report)

print("\n" + "=" * 72)
print("ESTABILIDADE — COMPARAÇÃO DOS TRÊS MÉTODOS")
print("=" * 72)

print(
    comparison.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nTEMPO CFIS")
print(
    "Média:",
    f"{np.mean(times):.4f} s"
)
print(
    "Desvio padrão:",
    f"{np.std(times, ddof=1):.4f} s"
)

print("\nRelatório:", report)
print("Teste externo utilizado: NÃO")
print("\nCÉLULA 44C CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 45
# QUALIDADE PREDITIVA PAREADA — BREAST_CANCER
#
# 10 splits × 3 métodos × 7 K × 2 modelos = 420 jobs
# Rankings existentes: CFIS, MI, ANOVA
# Pré-processamento ajustado somente no treino interno
# Teste externo: NÃO
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import os
import hashlib
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    f1_score, accuracy_score, balanced_accuracy_score
)

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"
BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

NAME = "Breast_Cancer"
SEED = 42
N_SPLITS = 10
TOP_K = [3, 5, 8, 10, 15, 20, 23]

METHODS = [
    "CFIS",
    "Mutual_Information",
    "ANOVA_or_F_Regression"
]

MODELS = ["DecisionTree", "RandomForest"]

RANK_DIR = (
    BASE / "Checkpoints"
    / "stability_v1_breast_cancer"
)

RUN_ID = "predictive_paired_v1_breast_cancer"
CHECKPOINT_DIR = BASE / "Checkpoints" / RUN_ID
RESULT_DIR = BASE / "Results" / RUN_ID

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)
RESULT_DIR.mkdir(parents=True, exist_ok=True)

CONFIG = {
    "dataset": NAME,
    "seed": SEED,
    "n_splits": N_SPLITS,
    "inner_validation_fraction": 0.25,
    "top_k": TOP_K,
    "methods": METHODS,
    "models": MODELS,
    "rf_estimators": 100,
    "outer_test_used": False
}

def digest(value):
    return hashlib.sha256(
        json.dumps(
            value, sort_keys=True
        ).encode("utf-8")
    ).hexdigest()

CONFIG_HASH = digest(CONFIG)

def atomic_json(path, payload):
    temp = path.with_name(path.name + ".tmp")
    with open(temp, "w", encoding="utf-8") as f:
        json.dump(payload, f, indent=2)
        f.flush()
        os.fsync(f.fileno())
    os.replace(temp, path)

config_file = CHECKPOINT_DIR / "config.json"

if config_file.exists():
    with open(config_file, encoding="utf-8") as f:
        if digest(json.load(f)) != CONFIG_HASH:
            raise RuntimeError(
                "Configuração incompatível com checkpoints."
            )
else:
    atomic_json(config_file, CONFIG)

with open(
    DATA / "Raw" / f"{NAME}_metadata.json",
    encoding="utf-8"
) as f:
    metadata = json.load(f)

raw = pd.read_parquet(
    DATA / "Raw" / f"{NAME}.parquet"
)

outer_idx = np.load(
    DATA / "Splits" / f"{NAME}_train_indices.npy",
    allow_pickle=False
)

outer = raw.iloc[outer_idx].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed"
    / NAME / "preprocessor.joblib"
)

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS,
    test_size=0.25,
    random_state=SEED
)

def model_factory(name):
    if name == "DecisionTree":
        return DecisionTreeClassifier(
            random_state=SEED
        )
    return RandomForestClassifier(
        n_estimators=100,
        random_state=SEED,
        n_jobs=1
    )

def checkpoint_path(split_no, method, model, k):
    return CHECKPOINT_DIR / (
        f"split_{split_no:02d}__"
        f"{method}__{model}__K{k}.json"
    )

print("=" * 72)
print("CFIS — QUALIDADE PREDITIVA PAREADA")
print("=" * 72)
print("Jobs previstos:", 10 * 3 * 7 * 2)
print("Teste externo utilizado: NÃO")

done = 0
skipped = 0
failures = []

for split_no, (fit_idx, val_idx) in enumerate(
    splitter.split(X_raw, y), start=1
):
    # Verifica se todos os rankings existem antes
    # de preparar a divisão.
    saved_rankings = {}

    for method in METHODS:
        rank_file = RANK_DIR / (
            f"split_{split_no:02d}__{method}.json"
        )

        if not rank_file.exists():
            raise FileNotFoundError(rank_file)

        with open(rank_file, encoding="utf-8") as f:
            record = json.load(f)

        if (
            record.get("dataset") != NAME
            or record.get("method") != method
            or record.get("split_number") != split_no
            or record.get("seed") != SEED
            or record.get("outer_test_used") is not False
        ):
            raise RuntimeError(
                f"Ranking incompatível: {rank_file}"
            )

        saved_rankings[method] = record["ranking_names"]

    pending = [
        (method, model, k)
        for method in METHODS
        for model in MODELS
        for k in TOP_K
        if not checkpoint_path(
            split_no, method, model, k
        ).exists()
    ]

    if not pending:
        skipped += len(METHODS) * len(MODELS) * len(TOP_K)
        print(
            f"Split {split_no:02d}: completo anteriormente",
            flush=True
        )
        continue

    # Reajusta o pré-processamento em cada treino interno.
    prep = clone(artifact["preprocessor"])
    variance = clone(artifact["variance_filter"])

    X_fit_raw = X_raw.iloc[fit_idx]
    X_val_raw = X_raw.iloc[val_idx]

    X_fit_trans = prep.fit_transform(
        X_fit_raw, y[fit_idx]
    )
    X_val_trans = prep.transform(X_val_raw)

    X_fit = variance.fit_transform(X_fit_trans)
    X_val = variance.transform(X_val_trans)

    try:
        names = np.asarray(
            prep.get_feature_names_out(),
            dtype=str
        )
    except (AttributeError, ValueError):
        names = np.asarray(
            X_fit_raw.columns, dtype=str
        )

    names = names[variance.get_support()]

    if len(names) != X_fit.shape[1]:
        raise RuntimeError(
            f"Dimensão inconsistente no split {split_no}"
        )

    if len(set(names)) != len(names):
        raise RuntimeError(
            "Nomes de características duplicados."
        )

    name_to_index = {
        name: index
        for index, name in enumerate(names)
    }

    for method in METHODS:
        ranking = saved_rankings[method]

        if (
            len(ranking) != len(names)
            or len(set(ranking)) != len(ranking)
            or set(ranking) != set(names)
        ):
            raise RuntimeError(
                f"Ranking incompatível com as features "
                f"do split {split_no}: {method}"
            )

        for k in TOP_K:
            selected_names = ranking[:k]
            columns = [
                name_to_index[name]
                for name in selected_names
            ]

            for model_name in MODELS:
                path = checkpoint_path(
                    split_no, method, model_name, k
                )

                if path.exists():
                    skipped += 1
                    continue

                try:
                    model = model_factory(model_name)

                    start = time.perf_counter()
                    model.fit(
                        X_fit[:, columns],
                        y[fit_idx]
                    )
                    fit_seconds = time.perf_counter() - start

                    prediction = model.predict(
                        X_val[:, columns]
                    )

                    payload = {
                        "Config_Hash": CONFIG_HASH,
                        "Dataset": NAME,
                        "Split": split_no,
                        "Method": method,
                        "Model": model_name,
                        "TopK": k,
                        "N_Train": len(fit_idx),
                        "N_Validation": len(val_idx),
                        "Macro_F1": float(
                            f1_score(
                                y[val_idx],
                                prediction,
                                average="macro",
                                zero_division=0
                            )
                        ),
                        "Balanced_Accuracy": float(
                            balanced_accuracy_score(
                                y[val_idx], prediction
                            )
                        ),
                        "Accuracy": float(
                            accuracy_score(
                                y[val_idx], prediction
                            )
                        ),
                        "Fit_Seconds": fit_seconds,
                        "Outer_Test_Used": False,
                        "Created_UTC": datetime.now(
                            timezone.utc
                        ).isoformat()
                    }

                    atomic_json(path, payload)
                    done += 1

                except Exception as exc:
                    failures.append(
                        (
                            split_no, method,
                            model_name, k, repr(exc)
                        )
                    )

    print(
        f"Split {split_no:02d}/{N_SPLITS} | "
        f"novos jobs: {done} | falhas: {len(failures)}",
        flush=True
    )

# ------------------------------------------------------------
# CONSOLIDAÇÃO
# ------------------------------------------------------------

records = []

for split_no in range(1, N_SPLITS + 1):
    for method in METHODS:
        for model_name in MODELS:
            for k in TOP_K:
                path = checkpoint_path(
                    split_no, method, model_name, k
                )

                if path.exists():
                    with open(path, encoding="utf-8") as f:
                        record = json.load(f)

                    if record["Config_Hash"] != CONFIG_HASH:
                        raise RuntimeError(
                            f"Checkpoint incompatível: {path}"
                        )

                    records.append(record)

results = pd.DataFrame(records)

output = RESULT_DIR / "predictive_paired_results.csv"
tmp = output.with_name(output.name + ".tmp")
results.to_csv(tmp, index=False)
os.replace(tmp, output)

print("\n" + "=" * 72)
print("RESUMO")
print("=" * 72)

print("Jobs esperados:", 420)
print("Novos jobs:", done)
print("Checkpoints reutilizados:", skipped)
print("Checkpoints disponíveis:", len(results))
print("Falhas:", len(failures))

if failures:
    for item in failures[:20]:
        print("ERRO:", item)

if not results.empty:
    summary = (
        results.groupby(
            ["Method", "Model", "TopK"],
            as_index=False
        )
        .agg(
            Mean_F1=("Macro_F1", "mean"),
            Std_F1=("Macro_F1", "std"),
            Mean_Balanced_Accuracy=(
                "Balanced_Accuracy", "mean"
            ),
            Splits=("Split", "nunique")
        )
    )

    # Apenas um resumo descritivo por modelo e método.
    # Não representa escolha definitiva de K.
    print("\nMELHOR MÉDIA OBSERVADA POR MÉTODO/MODELO")
    best = (
        summary.sort_values(
            "Mean_F1", ascending=False
        )
        .drop_duplicates(["Method", "Model"])
    )

    print(
        best.to_string(
            index=False,
            float_format=lambda v: f"{v:.5f}"
        )
    )

print("\nArquivo:", output)
print(
    "Status:",
    "COMPLETE" if len(results) == 420
    else "INCOMPLETE"
)
print("\nCÉLULA 45 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 46
# COMPARAÇÃO PAREADA + CONCORDÂNCIA DE RANKINGS
#
# Reutiliza 420 avaliações e 30 rankings.
# Nenhum novo treinamento.
# Nenhum acesso ao teste externo.
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from scipy.stats import spearmanr, kendalltau
import json
import os

import numpy as np
import pandas as pd

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")

BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

RESULTS_FILE = (
    BASE / "Results"
    / "predictive_paired_v1_breast_cancer"
    / "predictive_paired_results.csv"
)

RANK_DIR = (
    BASE / "Checkpoints"
    / "stability_v1_breast_cancer"
)

OUT = (
    BASE / "Results"
    / "paired_analysis_v1_breast_cancer"
)

OUT.mkdir(parents=True, exist_ok=True)

METHODS = [
    "Mutual_Information",
    "ANOVA_or_F_Regression"
]

TOP_K = [3, 5, 8, 10, 15, 20, 23]
N_SPLITS = 10

def atomic_csv(df, path):
    tmp = path.with_name(path.name + ".tmp")
    df.to_csv(tmp, index=False)
    os.replace(tmp, path)

# ------------------------------------------------------------
# 1. VERIFICAR OS 420 RESULTADOS
# ------------------------------------------------------------

df = pd.read_csv(RESULTS_FILE)

expected = 10 * 3 * 2 * 7

if len(df) != expected:
    raise RuntimeError(
        f"Esperados {expected} resultados; "
        f"encontrados {len(df)}."
    )

keys = ["Split", "Method", "Model", "TopK"]

if df.duplicated(keys).any():
    raise RuntimeError(
        "Há combinações experimentais duplicadas."
    )

if not df["Outer_Test_Used"].eq(False).all():
    raise RuntimeError(
        "Foi detectado registro de uso do teste externo."
    )

# ------------------------------------------------------------
# 2. COMPARAÇÃO PAREADA DE MACRO-F1
# ------------------------------------------------------------

wide = df.pivot(
    index=["Split", "Model", "TopK"],
    columns="Method",
    values="Macro_F1"
)

required_methods = ["CFIS"] + METHODS

if (
    wide.isna().any().any()
    or not set(required_methods).issubset(wide.columns)
):
    raise RuntimeError(
        "Resultados pareados incompletos."
    )

wide = wide.reset_index()

rows = []
details = []

for model in ["DecisionTree", "RandomForest"]:
    for k in TOP_K:
        subset = wide[
            (wide["Model"] == model)
            & (wide["TopK"] == k)
        ].sort_values("Split")

        if (
            len(subset) != N_SPLITS
            or subset["Split"].nunique() != N_SPLITS
        ):
            raise RuntimeError(
                f"Divisões incompletas: {model}, K={k}"
            )

        for competitor in METHODS:
            delta = (
                subset["CFIS"].to_numpy()
                - subset[competitor].to_numpy()
            )

            for split, value in zip(
                subset["Split"], delta
            ):
                details.append({
                    "Model": model,
                    "TopK": k,
                    "Competitor": competitor,
                    "Split": int(split),
                    "Delta_F1": float(value)
                })

            rows.append({
                "Model": model,
                "TopK": k,
                "Competitor": competitor,
                "CFIS_Mean_F1": float(
                    subset["CFIS"].mean()
                ),
                "Competitor_Mean_F1": float(
                    subset[competitor].mean()
                ),
                "Mean_Delta_F1": float(
                    np.mean(delta)
                ),
                "Std_Delta_F1": float(
                    np.std(delta, ddof=1)
                ),
                "CFIS_Wins": int(
                    np.sum(delta > 1e-12)
                ),
                "Ties": int(
                    np.sum(np.abs(delta) <= 1e-12)
                ),
                "CFIS_Losses": int(
                    np.sum(delta < -1e-12)
                ),
                "Splits": len(delta)
            })

paired = pd.DataFrame(rows)
paired_details = pd.DataFrame(details)

atomic_csv(
    paired,
    OUT / "paired_macro_f1.csv"
)

atomic_csv(
    paired_details,
    OUT / "paired_macro_f1_by_split.csv"
)

# ------------------------------------------------------------
# 3. CONCORDÂNCIA DOS RANKINGS
# ------------------------------------------------------------

def load_ranking(split, method):
    path = RANK_DIR / (
        f"split_{split:02d}__{method}.json"
    )

    with open(path, encoding="utf-8") as f:
        record = json.load(f)

    if (
        record.get("method") != method
        or record.get("split_number") != split
        or record.get("outer_test_used") is not False
    ):
        raise RuntimeError(
            f"Ranking incompatível: {path}"
        )

    ranking = record["ranking_names"]

    if len(ranking) != len(set(ranking)):
        raise RuntimeError(
            f"Ranking duplicado: {path}"
        )

    return ranking

agreement_rows = []

for split in range(1, N_SPLITS + 1):
    cfis = load_ranking(split, "CFIS")

    for competitor in METHODS:
        other = load_ranking(split, competitor)

        if set(cfis) != set(other):
            raise RuntimeError(
                "Universos de características diferentes."
            )

        pos_cfis = {
            feature: i
            for i, feature in enumerate(cfis)
        }

        pos_other = {
            feature: i
            for i, feature in enumerate(other)
        }

        universe = sorted(pos_cfis)

        a = [pos_cfis[f] for f in universe]
        b = [pos_other[f] for f in universe]

        rho = spearmanr(a, b).statistic
        tau = kendalltau(a, b).statistic

        for k in TOP_K:
            A = set(cfis[:k])
            B = set(other[:k])

            overlap = len(A & B)
            union = len(A | B)

            agreement_rows.append({
                "Split": split,
                "Competitor": competitor,
                "TopK": k,
                "Jaccard": overlap / union,
                "Precision_at_K": overlap / k,
                "Spearman_Full": float(rho),
                "Kendall_Full": float(tau)
            })

agreement = pd.DataFrame(agreement_rows)

agreement_summary = (
    agreement.groupby(
        ["Competitor", "TopK"],
        as_index=False
    )
    .agg(
        Mean_Jaccard=("Jaccard", "mean"),
        Mean_Precision_at_K=(
            "Precision_at_K", "mean"
        ),
        Mean_Spearman=(
            "Spearman_Full", "mean"
        ),
        Mean_Kendall=(
            "Kendall_Full", "mean"
        ),
        Splits=("Split", "nunique")
    )
)

atomic_csv(
    agreement,
    OUT / "ranking_agreement_by_split.csv"
)

atomic_csv(
    agreement_summary,
    OUT / "ranking_agreement_summary.csv"
)

# ------------------------------------------------------------
# 4. RELATÓRIO
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — COMPARAÇÃO PAREADA")
print("=" * 72)

print("\n1. MACRO-F1 — CFIS MENOS CONCORRENTE")
print(
    paired[
        [
            "Model",
            "TopK",
            "Competitor",
            "Mean_Delta_F1",
            "CFIS_Wins",
            "Ties",
            "CFIS_Losses"
        ]
    ].to_string(
        index=False,
        float_format=lambda v: f"{v:.5f}"
    )
)

print("\n2. CONCORDÂNCIA DOS RANKINGS")
print(
    agreement_summary.to_string(
        index=False,
        float_format=lambda v: f"{v:.4f}"
    )
)

print("\n3. INTEGRIDADE")
print("Avaliações utilizadas:", len(df))
print("Comparações preditivas:", len(paired))
print("Comparações de rankings:", len(agreement))
print("Teste externo utilizado: NÃO")
print("Testes de significância executados: NÃO")

print("\nArquivos salvos em:")
print(OUT)

print("\nCÉLULA 46 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 47
# WINE: CFIS × MI × ANOVA
# 10 splits × 3 métodos × 6 K × 2 modelos = 360 jobs
# Rankings, predição e estabilidade com checkpoints
# ============================================================

from pathlib import Path
from itertools import combinations
import json, os, time, hashlib

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.feature_selection import mutual_info_classif, f_classif
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, balanced_accuracy_score

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"
BASE = ROOT / "04_Efficiency/CFIS_Final_Benchmark/03_Final_Execution"

NAME = "Wine"
SEED = 42
N_SPLITS = 10
TOP_K = [2, 3, 4, 5, 7, 10]
METHODS = ["CFIS", "Mutual_Information", "ANOVA_or_F_Regression"]
MODELS = ["DecisionTree", "RandomForest"]

RUN = "wine_paired_stability_v1"
CKPT = BASE / "Checkpoints" / RUN
OUT = BASE / "Results" / RUN
CKPT.mkdir(parents=True, exist_ok=True)
OUT.mkdir(parents=True, exist_ok=True)

# Validação da implementação CFIS atual.
for name in [
    "cfis_function", "cfis_globals",
    "original_permutation",
    "consolidated_permutation_adapter"
]:
    if name not in globals():
        raise RuntimeError(f"Função ausente: {name}")

if cfis_function.__globals__ is not cfis_globals:
    raise RuntimeError("Namespace CFIS incompatível.")

if globals().get("exact_cfis") is not True:
    raise RuntimeError("Equivalência CFIS não confirmada.")

if cfis_globals["permutation_importance"] is not original_permutation:
    raise RuntimeError("Estado do adaptador CFIS inesperado.")

CONFIG = {
    "dataset": NAME, "seed": SEED, "splits": N_SPLITS,
    "validation_fraction": 0.25, "top_k": TOP_K,
    "methods": METHODS, "models": MODELS,
    "cfis_repeats": 10, "rf_estimators": 100,
    "outer_test_used": False
}
CONFIG_HASH = hashlib.sha256(
    json.dumps(CONFIG, sort_keys=True).encode()
).hexdigest()

def atomic_json(path, value):
    tmp = path.with_name(path.name + ".tmp")
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(value, f, ensure_ascii=False, indent=2)
        f.flush()
        os.fsync(f.fileno())
    os.replace(tmp, path)

def atomic_csv(df, path):
    tmp = path.with_name(path.name + ".tmp")
    df.to_csv(tmp, index=False)
    os.replace(tmp, path)

config_path = CKPT / "config.json"
if config_path.exists():
    with open(config_path, encoding="utf-8") as f:
        if json.load(f) != CONFIG:
            raise RuntimeError("Configuração de execução alterada.")
else:
    atomic_json(config_path, CONFIG)

# Carregar apenas os índices do treino externo.
with open(DATA / "Raw" / f"{NAME}_metadata.json", encoding="utf-8") as f:
    metadata = json.load(f)

raw = pd.read_parquet(DATA / "Raw" / f"{NAME}.parquet")
outer_indices = np.load(
    DATA / "Splits" / f"{NAME}_train_indices.npy",
    allow_pickle=False
)
outer = raw.iloc[outer_indices].copy()
target = metadata["target_column"]
X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed" / NAME / "preprocessor.joblib"
)

def run_cfis(X, y_train):
    try:
        cfis_globals["permutation_importance"] = (
            consolidated_permutation_adapter
        )
        return cfis_function(
            X, pd.Series(y_train),
            task="classification",
            model=None,
            n_repeats=10,
            random_state=SEED
        )
    finally:
        cfis_globals["permutation_importance"] = original_permutation

def ranking_from_scores(scores, names):
    scores = np.nan_to_num(
        np.asarray(scores, dtype=float),
        nan=-np.inf, posinf=np.inf, neginf=-np.inf
    )
    order = np.argsort(-scores, kind="stable")
    return names[order].tolist()

def make_model(name):
    if name == "DecisionTree":
        return DecisionTreeClassifier(random_state=SEED)
    return RandomForestClassifier(
        n_estimators=100, random_state=SEED, n_jobs=1
    )

def nogueira(rankings, k):
    universe = sorted(set().union(*(set(r) for r in rankings)))
    p = len(universe)
    if not (0 < k < p):
        return float("nan")
    position = {f: j for j, f in enumerate(universe)}
    matrix = np.zeros((len(rankings), p))
    for i, ranking in enumerate(rankings):
        for feature in ranking[:k]:
            matrix[i, position[feature]] = 1
    observed = matrix.var(axis=0, ddof=1).mean()
    expected = (k / p) * (1 - k / p)
    return float(1 - observed / expected)

print("=" * 72)
print("CÉLULA 47 — WINE: BENCHMARK PAREADO")
print("=" * 72)
print("Treino externo:", len(outer), "amostras")
print("Teste externo: isolado")
print("Jobs previstos: 360")

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS, test_size=0.25, random_state=SEED
)

new_rankings = 0
new_models = 0

for split_no, (fit_idx, val_idx) in enumerate(
    splitter.split(X_raw, y), start=1
):
    prep = clone(artifact["preprocessor"])
    variance = clone(artifact["variance_filter"])

    X_fit_raw = X_raw.iloc[fit_idx]
    X_val_raw = X_raw.iloc[val_idx]
    y_fit, y_val = y[fit_idx], y[val_idx]

    X_fit_t = prep.fit_transform(X_fit_raw, y_fit)
    X_val_t = prep.transform(X_val_raw)
    X_fit = variance.fit_transform(X_fit_t)
    X_val = variance.transform(X_val_t)

    if hasattr(X_fit, "toarray"):
        X_fit = X_fit.toarray()
        X_val = X_val.toarray()

    X_fit = np.asarray(X_fit, dtype=np.float64)
    X_val = np.asarray(X_val, dtype=np.float64)

    try:
        names = np.asarray(prep.get_feature_names_out(), dtype=str)
    except (AttributeError, ValueError):
        names = np.asarray(X_fit_raw.columns, dtype=str)

    names = names[variance.get_support()]

    if len(names) != X_fit.shape[1] or len(set(names)) != len(names):
        raise RuntimeError(f"Features inválidas no split {split_no}")

    if max(TOP_K) > len(names):
        raise RuntimeError(f"K excede as features no split {split_no}")

    feature_index = {f: i for i, f in enumerate(names)}

    for method in METHODS:
        ranking_file = CKPT / f"split_{split_no:02d}__{method}.json"

        if ranking_file.exists():
            with open(ranking_file, encoding="utf-8") as f:
                record = json.load(f)
            if record.get("config_hash") != CONFIG_HASH:
                raise RuntimeError(f"Ranking incompatível: {ranking_file}")
            ranking = record["ranking_names"]
        else:
            start = time.perf_counter()

            if method == "CFIS":
                result = run_cfis(
                    pd.DataFrame(X_fit, columns=names),
                    y_fit
                )
                if not isinstance(result, pd.DataFrame) or "Feature" not in result:
                    raise RuntimeError("Retorno CFIS inesperado.")
                ranking = result["Feature"].astype(str).tolist()

            elif method == "Mutual_Information":
                scores = mutual_info_classif(
                    X_fit, y_fit, random_state=SEED
                )
                ranking = ranking_from_scores(scores, names)

            else:
                scores, _ = f_classif(X_fit, y_fit)
                ranking = ranking_from_scores(scores, names)

            elapsed = time.perf_counter() - start

            if (
                len(ranking) != len(names)
                or len(set(ranking)) != len(ranking)
                or set(ranking) != set(names)
            ):
                raise RuntimeError(f"Ranking inválido: {method}")

            atomic_json(ranking_file, {
                "config_hash": CONFIG_HASH,
                "dataset": NAME,
                "method": method,
                "split_number": split_no,
                "ranking_names": ranking,
                "selection_seconds": elapsed,
                "outer_test_used": False
            })
            new_rankings += 1

        if set(ranking) != set(names) or len(ranking) != len(names):
            raise RuntimeError(f"Ranking incompatível: {method}")

        for k in TOP_K:
            columns = [feature_index[f] for f in ranking[:k]]

            for model_name in MODELS:
                path = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )
                if path.exists():
                    continue

                model = make_model(model_name)
                start = time.perf_counter()
                model.fit(X_fit[:, columns], y_fit)
                fit_seconds = time.perf_counter() - start
                pred = model.predict(X_val[:, columns])

                atomic_json(path, {
                    "config_hash": CONFIG_HASH,
                    "Split": split_no,
                    "Method": method,
                    "Model": model_name,
                    "TopK": k,
                    "Macro_F1": float(f1_score(
                        y_val, pred, average="macro", zero_division=0
                    )),
                    "Balanced_Accuracy": float(
                        balanced_accuracy_score(y_val, pred)
                    ),
                    "Fit_Seconds": fit_seconds,
                    "Outer_Test_Used": False
                })
                new_models += 1

    print(
        f"Split {split_no:02d}/10 | "
        f"novos rankings: {new_rankings} | "
        f"novas avaliações: {new_models}",
        flush=True
    )

# ------------------------------------------------------------
# CONSOLIDAÇÃO
# ------------------------------------------------------------

records = []
rankings = {method: [] for method in METHODS}
selection_times = {method: [] for method in METHODS}

for split_no in range(1, N_SPLITS + 1):
    for method in METHODS:
        path = CKPT / f"split_{split_no:02d}__{method}.json"
        with open(path, encoding="utf-8") as f:
            record = json.load(f)
        if record["config_hash"] != CONFIG_HASH:
            raise RuntimeError("Checkpoint de ranking incompatível.")
        rankings[method].append(record["ranking_names"])
        selection_times[method].append(record["selection_seconds"])

        for model_name in MODELS:
            for k in TOP_K:
                path = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )
                with open(path, encoding="utf-8") as f:
                    record = json.load(f)
                if record["config_hash"] != CONFIG_HASH:
                    raise RuntimeError("Checkpoint preditivo incompatível.")
                records.append(record)

results = pd.DataFrame(records)

if len(results) != 360:
    raise RuntimeError("Execução incompleta.")

summary = (
    results.groupby(["Method", "Model", "TopK"], as_index=False)
    .agg(
        Mean_F1=("Macro_F1", "mean"),
        Std_F1=("Macro_F1", "std"),
        Mean_Balanced_Accuracy=("Balanced_Accuracy", "mean"),
        Splits=("Split", "nunique")
    )
)

stability_rows = []
for method in METHODS:
    for k in TOP_K:
        pairwise = [
            len(set(a[:k]) & set(b[:k])) /
            len(set(a[:k]) | set(b[:k]))
            for a, b in combinations(rankings[method], 2)
        ]
        stability_rows.append({
            "Method": method,
            "TopK": k,
            "Mean_Jaccard": float(np.mean(pairwise)),
            "Nogueira": nogueira(rankings[method], k)
        })

stability = pd.DataFrame(stability_rows)

atomic_csv(results, OUT / "predictive_results.csv")
atomic_csv(summary, OUT / "predictive_summary.csv")
atomic_csv(stability, OUT / "stability_summary.csv")

print("\n" + "=" * 72)
print("MELHOR MACRO-F1 MÉDIO OBSERVADO POR MÉTODO/MODELO")
print("=" * 72)
print(
    summary.sort_values("Mean_F1", ascending=False)
    .drop_duplicates(["Method", "Model"])
    .to_string(index=False, float_format=lambda x: f"{x:.5f}")
)

print("\n" + "=" * 72)
print("ESTABILIDADE")
print("=" * 72)
print(
    stability.to_string(
        index=False, float_format=lambda x: f"{x:.4f}"
    )
)

print("\nTEMPO MÉDIO DE SELEÇÃO (s)")
for method in METHODS:
    print(f"{method}: {np.mean(selection_times[method]):.4f}")

print("\nAvaliações:", len(results), "/ 360")
print("Teste externo utilizado: NÃO")
print("Resultados:", OUT)
print("\nCÉLULA 47 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 48
# COMPARAÇÃO PAREADA WINE + SÍNTESE ENTRE DATASETS
# Sem treinamento | Sem teste externo
# ============================================================

from pathlib import Path
from scipy.stats import spearmanr, kendalltau
import json
import os

import numpy as np
import pandas as pd

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

WINE_DIR = BASE / "Results" / "wine_paired_stability_v1"
WINE_CKPT = BASE / "Checkpoints" / "wine_paired_stability_v1"

BC_DIR = (
    BASE / "Results"
    / "predictive_paired_v1_breast_cancer"
)

OUT = BASE / "Results" / "cross_dataset_paired_v1"
OUT.mkdir(parents=True, exist_ok=True)

COMPETITORS = [
    "Mutual_Information",
    "ANOVA_or_F_Regression"
]

def save_csv(frame, filename):
    path = OUT / filename
    temp = path.with_name(path.name + ".tmp")
    frame.to_csv(temp, index=False)
    os.replace(temp, path)

def load_wine_ranking(split, method):
    path = WINE_CKPT / (
        f"split_{split:02d}__{method}.json"
    )
    with open(path, encoding="utf-8") as f:
        record = json.load(f)

    if (
        record["method"] != method
        or record["split_number"] != split
        or record["outer_test_used"] is not False
    ):
        raise RuntimeError(f"Checkpoint inválido: {path}")

    return record["ranking_names"]

# ------------------------------------------------------------
# 1. CARREGAR RESULTADOS DAS DUAS BASES
# ------------------------------------------------------------

wine = pd.read_csv(WINE_DIR / "predictive_results.csv")
bc = pd.read_csv(BC_DIR / "predictive_paired_results.csv")

if len(wine) != 360 or len(bc) != 420:
    raise RuntimeError(
        "Quantidade de avaliações diferente do esperado."
    )

if (
    not wine["Outer_Test_Used"].eq(False).all()
    or not bc["Outer_Test_Used"].eq(False).all()
):
    raise RuntimeError("Teste externo detectado.")

wine["Dataset"] = "Wine"
bc["Dataset"] = "Breast_Cancer"

all_results = pd.concat(
    [bc, wine], ignore_index=True
)

keys = ["Dataset", "Split", "Method", "Model", "TopK"]

if all_results.duplicated(keys).any():
    raise RuntimeError("Combinações duplicadas.")

# ------------------------------------------------------------
# 2. DIFERENÇAS PAREADAS DE MACRO-F1
# ------------------------------------------------------------

wide = all_results.pivot(
    index=["Dataset", "Split", "Model", "TopK"],
    columns="Method",
    values="Macro_F1"
)

if wide.isna().any().any():
    raise RuntimeError("Resultados pareados incompletos.")

wide = wide.reset_index()

rows = []
detail_rows = []

for (dataset, model, k), group in wide.groupby(
    ["Dataset", "Model", "TopK"]
):
    group = group.sort_values("Split")

    if len(group) != 10 or group["Split"].nunique() != 10:
        raise RuntimeError(
            f"Divisões incompletas: {dataset}, {model}, K={k}"
        )

    for competitor in COMPETITORS:
        delta = (
            group["CFIS"].to_numpy()
            - group[competitor].to_numpy()
        )

        rows.append({
            "Dataset": dataset,
            "Model": model,
            "TopK": int(k),
            "Competitor": competitor,
            "CFIS_Mean_F1": float(group["CFIS"].mean()),
            "Competitor_Mean_F1": float(
                group[competitor].mean()
            ),
            "Mean_Delta_F1": float(delta.mean()),
            "Std_Delta_F1": float(delta.std(ddof=1)),
            "Wins": int((delta > 1e-12).sum()),
            "Ties": int((np.abs(delta) <= 1e-12).sum()),
            "Losses": int((delta < -1e-12).sum())
        })

        for split, value in zip(group["Split"], delta):
            detail_rows.append({
                "Dataset": dataset,
                "Model": model,
                "TopK": int(k),
                "Competitor": competitor,
                "Split": int(split),
                "Delta_F1": float(value)
            })

paired = pd.DataFrame(rows)
details = pd.DataFrame(detail_rows)

save_csv(paired, "paired_comparison_two_datasets.csv")
save_csv(details, "paired_differences_by_split.csv")

# ------------------------------------------------------------
# 3. CONCORDÂNCIA DOS RANKINGS NO WINE
# ------------------------------------------------------------

agreement_rows = []

for split in range(1, 11):
    cfis = load_wine_ranking(split, "CFIS")

    for competitor in COMPETITORS:
        other = load_wine_ranking(split, competitor)

        if set(cfis) != set(other):
            raise RuntimeError(
                f"Universos diferentes: split {split}"
            )

        position_a = {
            feature: i for i, feature in enumerate(cfis)
        }
        position_b = {
            feature: i for i, feature in enumerate(other)
        }

        universe = sorted(position_a)

        rho = spearmanr(
            [position_a[f] for f in universe],
            [position_b[f] for f in universe]
        ).statistic

        tau = kendalltau(
            [position_a[f] for f in universe],
            [position_b[f] for f in universe]
        ).statistic

        for k in [2, 3, 4, 5, 7, 10]:
            A = set(cfis[:k])
            B = set(other[:k])
            overlap = len(A & B)

            agreement_rows.append({
                "Split": split,
                "Competitor": competitor,
                "TopK": k,
                "Jaccard": overlap / len(A | B),
                "Precision_at_K": overlap / k,
                "Spearman_Full": float(rho),
                "Kendall_Full": float(tau)
            })

agreement = pd.DataFrame(agreement_rows)

agreement_summary = (
    agreement.groupby(
        ["Competitor", "TopK"], as_index=False
    )
    .agg(
        Jaccard=("Jaccard", "mean"),
        Precision_at_K=("Precision_at_K", "mean"),
        Spearman=("Spearman_Full", "mean"),
        Kendall=("Kendall_Full", "mean")
    )
)

save_csv(
    agreement_summary,
    "wine_ranking_agreement.csv"
)

# ------------------------------------------------------------
# 4. RELATÓRIO COMPARATIVO
# ------------------------------------------------------------

print("=" * 72)
print("CÉLULA 48 — COMPARAÇÃO ENTRE DATASETS")
print("=" * 72)

for dataset in ["Breast_Cancer", "Wine"]:
    print(f"\nDATASET: {dataset}")
    subset = paired[paired["Dataset"] == dataset]

    print(
        subset[
            [
                "Model", "TopK", "Competitor",
                "Mean_Delta_F1", "Wins", "Ties", "Losses"
            ]
        ].to_string(
            index=False,
            float_format=lambda v: f"{v:+.5f}"
        )
    )

    print("\nResumo das comparações:")
    for model in ["DecisionTree", "RandomForest"]:
        for competitor in COMPETITORS:
            selected = subset[
                (subset["Model"] == model)
                & (subset["Competitor"] == competitor)
            ]
            print(
                f"{model} × {competitor}: "
                f"{(selected['Mean_Delta_F1'] > 0).sum()}"
                f"/{len(selected)} K favoráveis ao CFIS"
            )

print("\n" + "=" * 72)
print("CONCORDÂNCIA DE RANKINGS — WINE")
print("=" * 72)
print(
    agreement_summary.to_string(
        index=False,
        float_format=lambda v: f"{v:.4f}"
    )
)

print("\nINTEGRIDADE")
print("Avaliações utilizadas:", len(all_results))
print("Comparações pareadas:", len(paired))
print("Novos treinamentos: 0")
print("Teste externo utilizado: NÃO")
print("Inferência estatística: ainda não executada")
print("Resultados:", OUT)
print("\nCÉLULA 48 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 49
# EXECUTOR MULTIDATASET — PRIMEIRA EXECUÇÃO: SYNTHETIC_30
#
# 10 splits × 3 seletores × 7 K × 2 modelos = 420
# Checkpoint/resume
# Pré-processamento fit apenas no treino interno
# Teste externo intocado
# ============================================================

from pathlib import Path
from itertools import combinations
import hashlib
import json
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import (
    StratifiedShuffleSplit,
    ShuffleSplit
)
from sklearn.feature_selection import (
    mutual_info_classif,
    mutual_info_regression,
    f_classif,
    f_regression
)
from sklearn.tree import (
    DecisionTreeClassifier,
    DecisionTreeRegressor
)
from sklearn.ensemble import (
    RandomForestClassifier,
    RandomForestRegressor
)
from sklearn.metrics import (
    f1_score,
    balanced_accuracy_score,
    r2_score,
    mean_squared_error
)

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"
BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

# Alterar apenas esta variável em futuras execuções.
DATASET = "Synthetic_30"

TOPK_BY_DATASET = {
    "Breast_Cancer": [3, 5, 8, 10, 15, 20, 23],
    "Wine": [2, 3, 4, 5, 7, 10],
    "Synthetic_30": [3, 5, 8, 10, 15, 20, 23],
    "Synthetic_60": [3, 5, 6, 10, 15, 20, 30, 45],
    "Synthetic_120": [3, 5, 10, 12, 20, 30, 60, 90],
    "Iris": [1, 2, 3],
    "Pima_Diabetes": [1, 2, 3, 4, 5, 6],
    "California_Housing": [1, 2, 3, 4, 5, 6],
    "Diabetes_Reg": [1, 3, 5, 8, 10],
    "Credit_Approval": [3, 5, 7, 10, 16, 20, 31, 46],
    "Gisette": [3, 5, 10, 20, 47, 117, 234, 351],
    "20_Newsgroups": [3, 5, 10, 20, 50, 125, 250, 375]
}

SEED = 42
N_SPLITS = 10
TOP_K = TOPK_BY_DATASET[DATASET]
METHODS = [
    "CFIS",
    "Mutual_Information",
    "ANOVA_or_F_Regression"
]
MODELS = ["DecisionTree", "RandomForest"]

# Identificação explícita da tarefa.
REGRESSION_DATASETS = {
    "California_Housing",
    "Diabetes_Reg"
}
TASK = (
    "regression"
    if DATASET in REGRESSION_DATASETS
    else "classification"
)

RUN = f"multidataset_v1_{DATASET.lower()}"
CKPT = BASE / "Checkpoints" / RUN
OUT = BASE / "Results" / RUN

CKPT.mkdir(parents=True, exist_ok=True)
OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# IMPLEMENTAÇÃO ATUAL DO CFIS
# ------------------------------------------------------------

required = [
    "cfis_function",
    "cfis_globals",
    "original_permutation",
    "consolidated_permutation_adapter"
]

for name in required:
    if name not in globals():
        raise RuntimeError(
            f"Implementação CFIS não carregada: {name}"
        )

if cfis_function.__globals__ is not cfis_globals:
    raise RuntimeError("Namespace CFIS incompatível.")

if cfis_globals["permutation_importance"] is not original_permutation:
    raise RuntimeError(
        "O adaptador original precisa estar restaurado."
    )

def current_cfis(X, y):
    try:
        cfis_globals["permutation_importance"] = (
            consolidated_permutation_adapter
        )
        return cfis_function(
            X,
            pd.Series(y),
            task=TASK,
            model=None,
            n_repeats=10,
            random_state=SEED
        )
    finally:
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

def atomic_json(path, data):
    tmp = path.with_name(path.name + ".tmp")
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
        f.flush()
        os.fsync(f.fileno())
    os.replace(tmp, path)

def atomic_csv(frame, path):
    tmp = path.with_name(path.name + ".tmp")
    frame.to_csv(tmp, index=False)
    os.replace(tmp, path)

# ------------------------------------------------------------
# CONFIGURAÇÃO IMUTÁVEL
# ------------------------------------------------------------

CONFIG = {
    "dataset": DATASET,
    "task": TASK,
    "seed": SEED,
    "n_splits": N_SPLITS,
    "validation_fraction": 0.25,
    "top_k": TOP_K,
    "methods": METHODS,
    "models": MODELS,
    "cfis_repeats": 10,
    "rf_estimators": 100,
    "outer_test_used": False
}

CONFIG_HASH = hashlib.sha256(
    json.dumps(CONFIG, sort_keys=True).encode()
).hexdigest()

config_file = CKPT / "config.json"

if config_file.exists():
    with open(config_file, encoding="utf-8") as f:
        if json.load(f) != CONFIG:
            raise RuntimeError(
                "Configuração diferente dos checkpoints."
            )
else:
    atomic_json(config_file, CONFIG)

# ------------------------------------------------------------
# DADOS: SOMENTE TREINO EXTERNO
# ------------------------------------------------------------

def locate_raw(dataset):
    for folder in ["Synthetic", "Raw"]:
        path = DATA / folder / f"{dataset}.parquet"
        if path.exists():
            return path, DATA / folder / f"{dataset}_metadata.json"
    raise FileNotFoundError(
        f"Dataset não encontrado: {dataset}"
    )

raw_file, metadata_file = locate_raw(DATASET)

with open(metadata_file, encoding="utf-8") as f:
    metadata = json.load(f)

raw = pd.read_parquet(raw_file)
outer_idx = np.load(
    DATA / "Splits" / f"{DATASET}_train_indices.npy",
    allow_pickle=False
)

outer = raw.iloc[outer_idx].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed"
    / DATASET / "preprocessor.joblib"
)

if TASK == "classification":
    splitter = StratifiedShuffleSplit(
        n_splits=N_SPLITS,
        test_size=0.25,
        random_state=SEED
    )
    splits = list(splitter.split(X_raw, y))
else:
    splitter = ShuffleSplit(
        n_splits=N_SPLITS,
        test_size=0.25,
        random_state=SEED
    )
    splits = list(splitter.split(X_raw))

def model_factory(name):
    if TASK == "classification":
        if name == "DecisionTree":
            return DecisionTreeClassifier(random_state=SEED)
        return RandomForestClassifier(
            n_estimators=100,
            random_state=SEED,
            n_jobs=1
        )
    if name == "DecisionTree":
        return DecisionTreeRegressor(random_state=SEED)
    return RandomForestRegressor(
        n_estimators=100,
        random_state=SEED,
        n_jobs=1
    )

def score_ranking(scores, names):
    values = np.asarray(scores, dtype=float)
    values = np.nan_to_num(
        values,
        nan=-np.inf,
        posinf=np.inf,
        neginf=-np.inf
    )
    return names[
        np.argsort(-values, kind="stable")
    ].tolist()

def nogueira(rankings, k):
    universe = sorted(
        set().union(*(set(r) for r in rankings))
    )
    p = len(universe)
    if not 0 < k < p:
        return float("nan")

    pos = {feature: j for j, feature in enumerate(universe)}
    matrix = np.zeros((len(rankings), p))

    for i, ranking in enumerate(rankings):
        for feature in ranking[:k]:
            matrix[i, pos[feature]] = 1

    observed = matrix.var(axis=0, ddof=1).mean()
    expected = (k / p) * (1 - k / p)

    return float(1 - observed / expected)

print("=" * 72)
print("CFIS — EXECUTOR MULTIDATASET")
print("=" * 72)
print("Dataset:", DATASET)
print("Tarefa:", TASK)
print("Treino externo:", len(outer))
print("K:", TOP_K)
print("Jobs previstos:", N_SPLITS * 3 * 2 * len(TOP_K))
print("Teste externo: isolado")

new_rankings = 0
new_models = 0

# ------------------------------------------------------------
# EXECUÇÃO COM CHECKPOINTS
# ------------------------------------------------------------

for split_no, (fit_idx, val_idx) in enumerate(splits, 1):
    prep = clone(artifact["preprocessor"])
    variance = clone(artifact["variance_filter"])

    X_fit_raw = X_raw.iloc[fit_idx]
    X_val_raw = X_raw.iloc[val_idx]

    y_fit = y[fit_idx]
    y_val = y[val_idx]

    X_fit_t = prep.fit_transform(X_fit_raw, y_fit)
    X_val_t = prep.transform(X_val_raw)

    X_fit = variance.fit_transform(X_fit_t)
    X_val = variance.transform(X_val_t)

    if hasattr(X_fit, "toarray"):
        X_fit = X_fit.toarray()
        X_val = X_val.toarray()

    X_fit = np.asarray(X_fit, dtype=np.float64)
    X_val = np.asarray(X_val, dtype=np.float64)

    try:
        names = np.asarray(
            prep.get_feature_names_out(),
            dtype=str
        )
    except (AttributeError, ValueError):
        names = np.asarray(
            X_fit_raw.columns,
            dtype=str
        )

    names = names[variance.get_support()]

    if (
        len(names) != X_fit.shape[1]
        or len(set(names)) != len(names)
        or max(TOP_K) > len(names)
    ):
        raise RuntimeError(
            f"Características inconsistentes no split {split_no}"
        )

    feature_index = {
        feature: j for j, feature in enumerate(names)
    }

    for method in METHODS:
        rank_file = CKPT / (
            f"split_{split_no:02d}__{method}.json"
        )

        if rank_file.exists():
            with open(rank_file, encoding="utf-8") as f:
                record = json.load(f)
            if record.get("config_hash") != CONFIG_HASH:
                raise RuntimeError(
                    f"Ranking incompatível: {rank_file}"
                )
            ranking = record["ranking_names"]
        else:
            start = time.perf_counter()

            if method == "CFIS":
                result = current_cfis(
                    pd.DataFrame(X_fit, columns=names),
                    y_fit
                )
                if (
                    not isinstance(result, pd.DataFrame)
                    or "Feature" not in result.columns
                ):
                    raise RuntimeError(
                        "Formato inesperado de retorno CFIS."
                    )
                ranking = result["Feature"].astype(str).tolist()

            elif method == "Mutual_Information":
                if TASK == "classification":
                    scores = mutual_info_classif(
                        X_fit, y_fit, random_state=SEED
                    )
                else:
                    scores = mutual_info_regression(
                        X_fit, y_fit, random_state=SEED
                    )
                ranking = score_ranking(scores, names)

            else:
                if TASK == "classification":
                    scores, _ = f_classif(X_fit, y_fit)
                else:
                    scores, _ = f_regression(X_fit, y_fit)
                ranking = score_ranking(scores, names)

            selection_seconds = time.perf_counter() - start

            if (
                len(ranking) != len(names)
                or len(set(ranking)) != len(ranking)
                or set(ranking) != set(names)
            ):
                raise RuntimeError(
                    f"Ranking inválido: {method}, split {split_no}"
                )

            atomic_json(rank_file, {
                "config_hash": CONFIG_HASH,
                "dataset": DATASET,
                "method": method,
                "split_number": split_no,
                "ranking_names": ranking,
                "selection_seconds": selection_seconds,
                "outer_test_used": False
            })
            new_rankings += 1

        if len(ranking) != len(names) or set(ranking) != set(names):
            raise RuntimeError(
                f"Ranking incompatível: {method}, split {split_no}"
            )

        for k in TOP_K:
            columns = [
                feature_index[feature]
                for feature in ranking[:k]
            ]

            for model_name in MODELS:
                result_file = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )

                if result_file.exists():
                    with open(result_file, encoding="utf-8") as f:
                        old = json.load(f)
                    if old.get("config_hash") != CONFIG_HASH:
                        raise RuntimeError(
                            f"Checkpoint incompatível: {result_file}"
                        )
                    continue

                model = model_factory(model_name)

                start = time.perf_counter()
                model.fit(X_fit[:, columns], y_fit)
                fit_seconds = time.perf_counter() - start

                predictions = model.predict(
                    X_val[:, columns]
                )

                result = {
                    "config_hash": CONFIG_HASH,
                    "Dataset": DATASET,
                    "Task": TASK,
                    "Split": split_no,
                    "Method": method,
                    "Model": model_name,
                    "TopK": k,
                    "Fit_Seconds": fit_seconds,
                    "Outer_Test_Used": False
                }

                if TASK == "classification":
                    result["Macro_F1"] = float(
                        f1_score(
                            y_val, predictions,
                            average="macro",
                            zero_division=0
                        )
                    )
                    result["Balanced_Accuracy"] = float(
                        balanced_accuracy_score(
                            y_val, predictions
                        )
                    )
                else:
                    result["R2"] = float(
                        r2_score(y_val, predictions)
                    )
                    result["RMSE"] = float(
                        np.sqrt(
                            mean_squared_error(
                                y_val, predictions
                            )
                        )
                    )

                atomic_json(result_file, result)
                new_models += 1

    print(
        f"Split {split_no:02d}/{N_SPLITS} | "
        f"novos rankings={new_rankings} | "
        f"novas avaliações={new_models}",
        flush=True
    )

# ------------------------------------------------------------
# CONSOLIDAR TODOS OS CHECKPOINTS
# ------------------------------------------------------------

rankings = {method: [] for method in METHODS}
selection_times = {method: [] for method in METHODS}
records = []

for split_no in range(1, N_SPLITS + 1):
    for method in METHODS:
        path = CKPT / f"split_{split_no:02d}__{method}.json"
        with open(path, encoding="utf-8") as f:
            record = json.load(f)

        if record["config_hash"] != CONFIG_HASH:
            raise RuntimeError("Ranking incompatível.")

        rankings[method].append(record["ranking_names"])
        selection_times[method].append(
            record["selection_seconds"]
        )

        for model_name in MODELS:
            for k in TOP_K:
                path = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )
                with open(path, encoding="utf-8") as f:
                    record = json.load(f)

                if record["config_hash"] != CONFIG_HASH:
                    raise RuntimeError("Resultado incompatível.")

                records.append(record)

results = pd.DataFrame(records)

expected = N_SPLITS * len(METHODS) * len(MODELS) * len(TOP_K)

if len(results) != expected:
    raise RuntimeError(
        f"Execução incompleta: {len(results)}/{expected}"
    )

metric = "Macro_F1" if TASK == "classification" else "R2"

summary = (
    results.groupby(
        ["Dataset", "Method", "Model", "TopK"],
        as_index=False
    )
    .agg(
        Mean_Score=(metric, "mean"),
        Std_Score=(metric, "std"),
        Splits=("Split", "nunique")
    )
)

stability_rows = []

for method in METHODS:
    for k in TOP_K:
        comparisons = [
            len(set(a[:k]) & set(b[:k])) /
            len(set(a[:k]) | set(b[:k]))
            for a, b in combinations(
                rankings[method], 2
            )
        ]

        stability_rows.append({
            "Dataset": DATASET,
            "Method": method,
            "TopK": k,
            "Mean_Jaccard": float(np.mean(comparisons)),
            "Nogueira": nogueira(rankings[method], k)
        })

stability = pd.DataFrame(stability_rows)

wide = results.pivot(
    index=["Split", "Model", "TopK"],
    columns="Method",
    values=metric
).reset_index()

paired_rows = []

for (model, k), group in wide.groupby(["Model", "TopK"]):
    for competitor in METHODS[1:]:
        delta = (
            group["CFIS"].to_numpy()
            - group[competitor].to_numpy()
        )

        paired_rows.append({
            "Dataset": DATASET,
            "Model": model,
            "TopK": k,
            "Competitor": competitor,
            "Mean_Delta": float(np.mean(delta)),
            "Wins": int(np.sum(delta > 1e-12)),
            "Ties": int(np.sum(np.abs(delta) <= 1e-12)),
            "Losses": int(np.sum(delta < -1e-12))
        })

paired = pd.DataFrame(paired_rows)

atomic_csv(results, OUT / "predictive_results.csv")
atomic_csv(summary, OUT / "predictive_summary.csv")
atomic_csv(stability, OUT / "stability_summary.csv")
atomic_csv(paired, OUT / "paired_comparison.csv")

print("\n" + "=" * 72)
print("MELHORES MÉDIAS OBSERVADAS")
print("=" * 72)
print(
    summary.sort_values("Mean_Score", ascending=False)
    .drop_duplicates(["Method", "Model"])
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)

print("\n" + "=" * 72)
print("ESTABILIDADE")
print("=" * 72)
print(
    stability.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 72)
print("COMPARAÇÃO PAREADA — CFIS MENOS CONCORRENTE")
print("=" * 72)
print(
    paired.to_string(
        index=False,
        float_format=lambda x: f"{x:+.5f}"
    )
)

print("\nTEMPO MÉDIO DE SELEÇÃO")
for method in METHODS:
    print(
        f"{method}: "
        f"{np.mean(selection_times[method]):.4f} s"
    )

print("\nAvaliações:", len(results), "/", expected)
print("Teste externo utilizado: NÃO")
print("Resultados:", OUT)
print("\nCÉLULA 49 CONCLUÍDA.")
```

```python

# ============================================================
# CFIS — CÉLULA 50
# SYNTHETIC_60 — CFIS × MUTUAL INFORMATION
#
# ANOVA EXCLUÍDA
# 10 splits × 2 métodos × 8 K × 2 modelos = 320 jobs
#
# - Checkpoint/resume
# - Pré-processamento ajustado apenas no treino interno
# - Teste externo isolado
# - CFIS original com adaptador numericamente equivalente
# - Estabilidade e comparação pareada
# ============================================================

from pathlib import Path
from itertools import combinations
import hashlib
import json
import os
import time

import joblib
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.feature_selection import mutual_info_classif
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, balanced_accuracy_score

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")
DATA = ROOT / "01_Datasets"

BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

DATASET = "Synthetic_60"
SEED = 42
N_SPLITS = 10
TOP_K = [3, 5, 6, 10, 15, 20, 30, 45]

METHODS = ["CFIS", "Mutual_Information"]
MODELS = ["DecisionTree", "RandomForest"]

RUN = "multidataset_v2_synthetic_60"
CKPT = BASE / "Checkpoints" / RUN
OUT = BASE / "Results" / RUN

CKPT.mkdir(parents=True, exist_ok=True)
OUT.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# VALIDAR CFIS
# ------------------------------------------------------------

required = [
    "cfis_function",
    "cfis_globals",
    "original_permutation",
    "consolidated_permutation_adapter"
]

for name in required:
    if name not in globals():
        raise RuntimeError(
            f"Implementação CFIS ausente: {name}. "
            "Execute primeiro a célula que carrega o CFIS."
        )

if cfis_function.__globals__ is not cfis_globals:
    raise RuntimeError("Namespace CFIS incompatível.")

if cfis_globals["permutation_importance"] is not original_permutation:
    raise RuntimeError("Estado inicial CFIS incompatível.")

if globals().get("exact_cfis") is not True:
    raise RuntimeError(
        "Equivalência da implementação CFIS não confirmada."
    )

def run_cfis(X, y):
    try:
        cfis_globals["permutation_importance"] = (
            consolidated_permutation_adapter
        )

        result = cfis_function(
            X,
            pd.Series(y),
            task="classification",
            model=None,
            n_repeats=10,
            random_state=SEED
        )

        return result

    finally:
        cfis_globals["permutation_importance"] = (
            original_permutation
        )

# ------------------------------------------------------------
# CHECKPOINTS
# ------------------------------------------------------------

def atomic_json(path, value):
    tmp = path.with_name(path.name + ".tmp")

    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(value, f, ensure_ascii=False, indent=2)
        f.flush()
        os.fsync(f.fileno())

    os.replace(tmp, path)

def atomic_csv(frame, path):
    tmp = path.with_name(path.name + ".tmp")
    frame.to_csv(tmp, index=False)
    os.replace(tmp, path)

CONFIG = {
    "dataset": DATASET,
    "seed": SEED,
    "splits": N_SPLITS,
    "validation_fraction": 0.25,
    "top_k": TOP_K,
    "methods": METHODS,
    "models": MODELS,
    "cfis_repeats": 10,
    "rf_estimators": 100,
    "outer_test_used": False
}

CONFIG_HASH = hashlib.sha256(
    json.dumps(CONFIG, sort_keys=True).encode()
).hexdigest()

config_path = CKPT / "config.json"

if config_path.exists():
    with open(config_path, encoding="utf-8") as f:
        if json.load(f) != CONFIG:
            raise RuntimeError(
                "Configuração incompatível com execução anterior."
            )
else:
    atomic_json(config_path, CONFIG)

# ------------------------------------------------------------
# DADOS — APENAS TREINO EXTERNO
# ------------------------------------------------------------

raw_path = DATA / "Synthetic" / f"{DATASET}.parquet"
meta_path = DATA / "Synthetic" / f"{DATASET}_metadata.json"

with open(meta_path, encoding="utf-8") as f:
    metadata = json.load(f)

raw = pd.read_parquet(raw_path)

outer_idx = np.load(
    DATA / "Splits" / f"{DATASET}_train_indices.npy",
    allow_pickle=False
)

outer = raw.iloc[outer_idx].copy()
target = metadata["target_column"]

X_raw = outer.drop(columns=[target])
y = outer[target].to_numpy().ravel()

artifact = joblib.load(
    DATA / "Processed" / DATASET / "preprocessor.joblib"
)

splitter = StratifiedShuffleSplit(
    n_splits=N_SPLITS,
    test_size=0.25,
    random_state=SEED
)

splits = list(splitter.split(X_raw, y))

def make_model(name):
    if name == "DecisionTree":
        return DecisionTreeClassifier(random_state=SEED)

    return RandomForestClassifier(
        n_estimators=100,
        random_state=SEED,
        n_jobs=1
    )

def ranking_from_scores(scores, names):
    scores = np.nan_to_num(
        np.asarray(scores, dtype=float),
        nan=-np.inf,
        posinf=np.inf,
        neginf=-np.inf
    )

    order = np.argsort(-scores, kind="stable")
    return names[order].tolist()

def nogueira(rankings, k):
    universe = sorted(
        set().union(*(set(r) for r in rankings))
    )

    p = len(universe)

    if not 0 < k < p:
        return float("nan")

    position = {
        feature: j
        for j, feature in enumerate(universe)
    }

    matrix = np.zeros((len(rankings), p))

    for i, ranking in enumerate(rankings):
        for feature in ranking[:k]:
            matrix[i, position[feature]] = 1

    observed = matrix.var(axis=0, ddof=1).mean()
    expected = (k / p) * (1 - k / p)

    return float(1 - observed / expected)

print("=" * 72)
print("CFIS — SYNTHETIC_60")
print("=" * 72)
print("Métodos:", METHODS)
print("ANOVA: EXCLUÍDA")
print("Treino externo:", len(outer))
print("Teste externo: ISOLADO")
print("Jobs previstos:", 320)

new_rankings = 0
new_jobs = 0
reused_jobs = 0

# ------------------------------------------------------------
# EXECUÇÃO
# ------------------------------------------------------------

for split_no, (fit_idx, val_idx) in enumerate(splits, 1):

    prep = clone(artifact["preprocessor"])
    variance = clone(artifact["variance_filter"])

    X_fit_raw = X_raw.iloc[fit_idx]
    X_val_raw = X_raw.iloc[val_idx]

    y_fit = y[fit_idx]
    y_val = y[val_idx]

    X_fit_t = prep.fit_transform(X_fit_raw, y_fit)
    X_val_t = prep.transform(X_val_raw)

    X_fit = variance.fit_transform(X_fit_t)
    X_val = variance.transform(X_val_t)

    if hasattr(X_fit, "toarray"):
        X_fit = X_fit.toarray()
        X_val = X_val.toarray()

    X_fit = np.asarray(X_fit, dtype=np.float64)
    X_val = np.asarray(X_val, dtype=np.float64)

    try:
        names = np.asarray(
            prep.get_feature_names_out(),
            dtype=str
        )
    except (AttributeError, ValueError):
        names = np.asarray(
            X_fit_raw.columns,
            dtype=str
        )

    names = names[variance.get_support()]

    if (
        len(names) != X_fit.shape[1]
        or len(set(names)) != len(names)
        or max(TOP_K) > len(names)
    ):
        raise RuntimeError(
            f"Características inconsistentes no split {split_no}"
        )

    feature_index = {
        feature: j
        for j, feature in enumerate(names)
    }

    for method in METHODS:

        rank_file = CKPT / (
            f"split_{split_no:02d}__{method}.json"
        )

        if rank_file.exists():
            with open(rank_file, encoding="utf-8") as f:
                record = json.load(f)

            if record.get("config_hash") != CONFIG_HASH:
                raise RuntimeError(
                    f"Ranking incompatível: {rank_file}"
                )

            ranking = record["ranking_names"]

        else:
            start = time.perf_counter()

            if method == "CFIS":
                result = run_cfis(
                    pd.DataFrame(X_fit, columns=names),
                    y_fit
                )

                if (
                    not isinstance(result, pd.DataFrame)
                    or "Feature" not in result.columns
                ):
                    raise RuntimeError(
                        "Formato inesperado do CFIS."
                    )

                ranking = (
                    result["Feature"].astype(str).tolist()
                )

            else:
                scores = mutual_info_classif(
                    X_fit,
                    y_fit,
                    random_state=SEED
                )

                ranking = ranking_from_scores(
                    scores, names
                )

            elapsed = time.perf_counter() - start

            if (
                len(ranking) != len(names)
                or len(set(ranking)) != len(ranking)
                or set(ranking) != set(names)
            ):
                raise RuntimeError(
                    f"Ranking inválido: {method}"
                )

            atomic_json(rank_file, {
                "config_hash": CONFIG_HASH,
                "dataset": DATASET,
                "method": method,
                "split_number": split_no,
                "ranking_names": ranking,
                "selection_seconds": elapsed,
                "outer_test_used": False
            })

            new_rankings += 1

        if (
            len(ranking) != len(names)
            or set(ranking) != set(names)
        ):
            raise RuntimeError(
                f"Ranking incompatível: {method}"
            )

        for k in TOP_K:

            columns = [
                feature_index[feature]
                for feature in ranking[:k]
            ]

            for model_name in MODELS:

                result_file = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )

                if result_file.exists():
                    with open(result_file, encoding="utf-8") as f:
                        old = json.load(f)

                    if old.get("config_hash") != CONFIG_HASH:
                        raise RuntimeError(
                            "Checkpoint incompatível."
                        )

                    reused_jobs += 1
                    continue

                model = make_model(model_name)

                start = time.perf_counter()

                model.fit(
                    X_fit[:, columns],
                    y_fit
                )

                fit_seconds = time.perf_counter() - start

                pred = model.predict(
                    X_val[:, columns]
                )

                record = {
                    "config_hash": CONFIG_HASH,
                    "Dataset": DATASET,
                    "Split": split_no,
                    "Method": method,
                    "Model": model_name,
                    "TopK": k,
                    "Macro_F1": float(
                        f1_score(
                            y_val,
                            pred,
                            average="macro",
                            zero_division=0
                        )
                    ),
                    "Balanced_Accuracy": float(
                        balanced_accuracy_score(
                            y_val,
                            pred
                        )
                    ),
                    "Fit_Seconds": fit_seconds,
                    "Outer_Test_Used": False
                }

                atomic_json(result_file, record)
                new_jobs += 1

    print(
        f"Split {split_no:02d}/10 | "
        f"rankings novos={new_rankings} | "
        f"jobs novos={new_jobs} | "
        f"reutilizados={reused_jobs}",
        flush=True
    )

# ------------------------------------------------------------
# CONSOLIDAÇÃO
# ------------------------------------------------------------

records = []
rankings = {method: [] for method in METHODS}
selection_times = {method: [] for method in METHODS}

for split_no in range(1, N_SPLITS + 1):
    for method in METHODS:

        path = CKPT / f"split_{split_no:02d}__{method}.json"

        with open(path, encoding="utf-8") as f:
            record = json.load(f)

        if record["config_hash"] != CONFIG_HASH:
            raise RuntimeError("Ranking incompatível.")

        rankings[method].append(
            record["ranking_names"]
        )

        selection_times[method].append(
            record["selection_seconds"]
        )

        for model_name in MODELS:
            for k in TOP_K:

                path = CKPT / (
                    f"split_{split_no:02d}__{method}"
                    f"__{model_name}__K{k}.json"
                )

                with open(path, encoding="utf-8") as f:
                    record = json.load(f)

                if record["config_hash"] != CONFIG_HASH:
                    raise RuntimeError(
                        "Resultado incompatível."
                    )

                records.append(record)

results = pd.DataFrame(records)

if len(results) != 320:
    raise RuntimeError(
        f"Execução incompleta: {len(results)}/320"
    )

summary = (
    results.groupby(
        ["Method", "Model", "TopK"],
        as_index=False
    )
    .agg(
        Mean_F1=("Macro_F1", "mean"),
        Std_F1=("Macro_F1", "std"),
        Mean_Balanced_Accuracy=(
            "Balanced_Accuracy", "mean"
        ),
        Splits=("Split", "nunique")
    )
)

stability_rows = []

for method in METHODS:
    for k in TOP_K:

        values = [
            len(set(a[:k]) & set(b[:k])) /
            len(set(a[:k]) | set(b[:k]))
            for a, b in combinations(
                rankings[method], 2
            )
        ]

        stability_rows.append({
            "Method": method,
            "TopK": k,
            "Mean_Jaccard": float(np.mean(values)),
            "Nogueira": nogueira(rankings[method], k)
        })

stability = pd.DataFrame(stability_rows)

wide = results.pivot(
    index=["Split", "Model", "TopK"],
    columns="Method",
    values="Macro_F1"
).reset_index()

paired_rows = []

for (model, k), group in wide.groupby(
    ["Model", "TopK"]
):
    delta = (
        group["CFIS"].to_numpy()
        - group["Mutual_Information"].to_numpy()
    )

    paired_rows.append({
        "Model": model,
        "TopK": k,
        "Mean_Delta_F1": float(delta.mean()),
        "Wins": int((delta > 1e-12).sum()),
        "Ties": int((np.abs(delta) <= 1e-12).sum()),
        "Losses": int((delta < -1e-12).sum())
    })

paired = pd.DataFrame(paired_rows)

atomic_csv(results, OUT / "predictive_results.csv")
atomic_csv(summary, OUT / "predictive_summary.csv")
atomic_csv(stability, OUT / "stability_summary.csv")
atomic_csv(paired, OUT / "paired_comparison.csv")

print("\n" + "=" * 72)
print("MELHORES MÉDIAS OBSERVADAS")
print("=" * 72)

print(
    summary.sort_values(
        "Mean_F1", ascending=False
    )
    .drop_duplicates(["Method", "Model"])
    .to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)

print("\nESTABILIDADE")
print(
    stability.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nCOMPARAÇÃO PAREADA")
print(
    paired.to_string(
        index=False,
        float_format=lambda x: f"{x:+.5f}"
    )
)

print("\nTEMPO MÉDIO DE SELEÇÃO")
for method in METHODS:
    print(
        f"{method}: "
        f"{np.mean(selection_times[method]):.4f} s"
    )

print("\nAvaliações:", len(results), "/ 320")
print("Novos jobs:", new_jobs)
print("Jobs reutilizados:", reused_jobs)
print("Teste externo utilizado: NÃO")
print("Resultados:", OUT)
print("\nCÉLULA 50 CONCLUÍDA.")
```

```python

# ============================================================
# CÉLULA 51 — BENCHMARK FINAL CFIS / DECISION TREE
# Congelamento do protocolo e manifesto de execução
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os

import numpy as np
import pandas as pd

ROOT = Path("/content/drive/MyDrive/CFIS_Artigo_Diogo")

BASE = (
    ROOT / "04_Efficiency"
    / "CFIS_Final_Benchmark"
    / "03_Final_Execution"
)

RUN_NAME = "decision_tree_final_v1"

RUN_DIR = BASE / RUN_NAME
MANIFEST_DIR = RUN_DIR / "Manifests"
CHECKPOINT_DIR = RUN_DIR / "Checkpoints"
RESULTS_DIR = RUN_DIR / "Results"

for directory in [
    MANIFEST_DIR,
    CHECKPOINT_DIR,
    RESULTS_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

SEED = 42
N_INNER_SPLITS = 10
CFIS_REPEATS = 10

TOPK = {
    "20_Newsgroups": [3, 5, 10, 20, 50, 125, 250, 375],
    "Breast_Cancer": [3, 5, 8, 10, 15, 20, 23],
    "California_Housing": [1, 2, 3, 4, 5, 6],
    "Credit_Approval": [3, 5, 7, 10, 16, 20, 31, 46],
    "Diabetes_Reg": [1, 3, 5, 8, 10],
    "Gisette": [3, 5, 10, 20, 47, 117, 234, 351],
    "Iris": [1, 2, 3],
    "Pima_Diabetes": [1, 2, 3, 4, 5, 6],
    "Synthetic_120": [3, 5, 10, 12, 20, 30, 60, 90],
    "Synthetic_30": [3, 5, 8, 10, 15, 20, 23],
    "Synthetic_60": [3, 5, 6, 10, 15, 20, 30, 45],
    "Wine": [2, 3, 4, 5, 7, 10]
}

REGRESSION_DATASETS = {
    "California_Housing",
    "Diabetes_Reg"
}

METHODS = [
    "CFIS",
    "Mutual_Information",
    "mRMR",
    "L1",
    "CFS",
    "RFE",
    "PFI",
    "TreeSHAP",
    "MDI",
    "Random_TopK"
]

MODEL = "DecisionTree"

PROTOCOL = {
    "name": RUN_NAME,
    "version": 1,
    "seed": SEED,
    "model": MODEL,
    "datasets": list(TOPK.keys()),
    "topk": TOPK,
    "methods": METHODS,
    "excluded_main_comparison": ["ANOVA_or_F_Regression"],
    "regression_datasets": sorted(REGRESSION_DATASETS),
    "inner_splits": N_INNER_SPLITS,
    "inner_validation_fraction": 0.25,
    "cfis_repeats": CFIS_REPEATS,
    "preprocessing": "fit_on_inner_training_only",
    "feature_selection": "fit_on_inner_training_only",
    "outer_test": "reserved_for_final_evaluation",
    "prediction_metrics": {
        "classification": [
            "Macro_F1",
            "Balanced_Accuracy"
        ],
        "regression": [
            "R2",
            "RMSE"
        ]
    },
    "ranking_metrics": [
        "Jaccard_at_K",
        "Precision_at_K",
        "Spearman",
        "Kendall"
    ],
    "stability_metrics": [
        "Mean_Pairwise_Jaccard",
        "Nogueira"
    ],
    "efficiency_metrics": [
        "Selection_Wall_Seconds",
        "Selection_CPU_Seconds",
        "Peak_RSS_MiB",
        "Training_Wall_Seconds",
        "Total_Wall_Seconds",
        "Estimated_Energy_Wh"
    ],
    "statistical_analysis": [
        "Friedman_across_datasets",
        "Paired_Wilcoxon_across_datasets",
        "Holm_correction",
        "Confidence_Intervals",
        "Effect_Sizes"
    ],
    "ablation": [
        "CFIS_Full",
        "CFIS_PFI_Component",
        "CFIS_Coefficient_Component",
        "CFIS_Statistical_Component"
    ],
    "random_topk_policy": (
        "deterministic_seeded_subset_per_split_and_k"
    ),
    "model_dependent_selector_policy": (
        "fit_reference_model_on_inner_training_only"
    ),
    "energy_policy": (
        "estimated_only_unless_hardware_measurement_available"
    )
}

PROTOCOL_HASH = hashlib.sha256(
    json.dumps(
        PROTOCOL,
        sort_keys=True,
        ensure_ascii=False
    ).encode("utf-8")
).hexdigest()

PROTOCOL["protocol_sha256"] = PROTOCOL_HASH

protocol_file = MANIFEST_DIR / "protocol.json"

if protocol_file.exists():
    with open(protocol_file, encoding="utf-8") as f:
        existing = json.load(f)

    if existing != PROTOCOL:
        raise RuntimeError(
            "Já existe um protocolo diferente neste diretório. "
            "Não sobrescreveremos um protocolo congelado."
        )
else:
    tmp = protocol_file.with_suffix(".json.tmp")

    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(
            PROTOCOL,
            f,
            ensure_ascii=False,
            indent=2
        )
        f.flush()
        os.fsync(f.fileno())

    os.replace(tmp, protocol_file)

# ------------------------------------------------------------
# MANIFESTO INTERNO
# ------------------------------------------------------------

rows = []

for dataset, ks in TOPK.items():

    task = (
        "regression"
        if dataset in REGRESSION_DATASETS
        else "classification"
    )

    for split in range(1, N_INNER_SPLITS + 1):

        for method in METHODS:

            for k in ks:

                rows.append({
                    "Dataset": dataset,
                    "Task": task,
                    "Model": MODEL,
                    "Split": split,
                    "Method": method,
                    "TopK": k,
                    "Protocol_SHA256": PROTOCOL_HASH,
                    "Status": "pending"
                })

manifest = pd.DataFrame(rows)

manifest_path = (
    MANIFEST_DIR / "inner_benchmark_manifest.csv"
)

if manifest_path.exists():
    previous = pd.read_csv(manifest_path)

    identity_columns = [
        "Dataset",
        "Task",
        "Model",
        "Split",
        "Method",
        "TopK",
        "Protocol_SHA256"
    ]

    if not previous[identity_columns].equals(
        manifest[identity_columns]
    ):
        raise RuntimeError(
            "Manifesto existente incompatível."
        )
else:
    tmp = manifest_path.with_suffix(".csv.tmp")
    manifest.to_csv(tmp, index=False)
    os.replace(tmp, manifest_path)

# ------------------------------------------------------------
# MANIFESTO EXTERNO — NÃO EXECUTAR AGORA
# ------------------------------------------------------------

outer_rows = []

for dataset, ks in TOPK.items():

    task = (
        "regression"
        if dataset in REGRESSION_DATASETS
        else "classification"
    )

    for method in METHODS:

        outer_rows.append({
            "Dataset": dataset,
            "Task": task,
            "Model": MODEL,
            "Method": method,
            "Selected_K": None,
            "K_Selection_Rule": (
                "maximize_inner_validation_primary_metric"
            ),
            "Outer_Test_Status": "LOCKED",
            "Protocol_SHA256": PROTOCOL_HASH
        })

outer_manifest = pd.DataFrame(outer_rows)

outer_path = (
    MANIFEST_DIR / "outer_final_manifest.csv"
)

if not outer_path.exists():
    tmp = outer_path.with_suffix(".csv.tmp")
    outer_manifest.to_csv(tmp, index=False)
    os.replace(tmp, outer_path)

# ------------------------------------------------------------
# RESUMO
# ------------------------------------------------------------

print("=" * 72)
print("CFIS — BENCHMARK FINAL / DECISION TREE")
print("=" * 72)

print("Modelo:", MODEL)
print("Datasets:", len(TOPK))
print("Métodos:", len(METHODS))
print("ANOVA na comparação principal: NÃO")
print("Divisões internas:", N_INNER_SPLITS)

print("\nMétodos:")
for i, method in enumerate(METHODS, 1):
    print(f"{i:02d}. {method}")

print("\nJobs internos:", len(manifest))
print("Configurações externas:", len(outer_manifest))
print("Teste externo: BLOQUEADO")

print("\nJobs por dataset:")
print(
    manifest.groupby("Dataset")
    .size()
    .rename("Jobs")
    .to_string()
)

print("\nProtocolo SHA256:", PROTOCOL_HASH)
print("Manifesto:", manifest_path)
print("Protocolo:", protocol_file)
print("Resultados:", RESULTS_DIR)

print("\nCÉLULA 51 CONCLUÍDA.")
```

```python
from pathlib import Path
import ast

path = Path(
    "/content/drive/MyDrive/CFIS_Artigo_Diogo/"
    "09_Logs/CFIS_replication_v1.py"
)

source = path.read_text(encoding="utf-8")
tree = ast.parse(source)

fn = next(
    node for node in tree.body
    if isinstance(node, ast.FunctionDef)
    and node.name == "CFIS_feature_selection"
)

print("ASSINATURA:")
print(ast.unparse(fn.args))

print("\nIMPORTAÇÕES:")
for node in ast.walk(fn):
    if isinstance(node, (ast.Import, ast.ImportFrom)):
        print(ast.unparse(node))

print("\nREFERÊNCIAS À PERMUTAÇÃO:")
for node in ast.walk(fn):
    if isinstance(node, ast.Call):
        name = ast.unparse(node.func)
        if "permutation" in name.lower():
            print(f"Linha {node.lineno}: {ast.unparse(node)[:500]}")

print("\nCÓDIGO COMPLETO DA FUNÇÃO:")
print(ast.get_source_segment(source, fn))
```
